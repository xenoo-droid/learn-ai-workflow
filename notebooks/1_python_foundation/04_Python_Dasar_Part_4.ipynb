# Bab 1: String Mendalam

String sudah dikenalkan sekilas di Part 1. Di bab ini kita bahas lebih dalam: indexing, slicing, dan method-method yang sering dipakai sehari-hari.

## 1.1 Indexing & Slicing

String adalah **sequence** (urutan karakter), sehingga tiap karakternya punya posisi/index, dimulai dari `0`.

```python
kata = "Python"

print(kata[0])       # karakter pertama
print(kata[-1])      # karakter terakhir (index negatif dihitung dari belakang)
print(kata[0:3])     # slicing: index 0 sampai 2 (index 3 TIDAK ikut)
print(kata[3:])      # dari index 3 sampai akhir
print(kata[:3])      # dari awal sampai index 2
print(kata[::-1])    # trik populer: membalik string (reverse)
```

🧠 **Penjelasan Output**: pola slicing adalah `string[mulai:berhenti:langkah]`. Semua bagian ini opsional — kalau dikosongkan, Python otomatis memakai nilai default (mulai dari awal, sampai akhir, langkah 1). Langkah `-1` membuat slicing berjalan mundur, sehingga `kata[::-1]` membalik seluruh string.

## 1.2 String Bersifat Immutable

String **tidak bisa diubah sebagian**. Setiap "perubahan" sebenarnya membuat string baru.

```python
kata = "Python"

try:
    kata[0] = "J"
except TypeError as e:
    print("TypeError:", e)
    print("String bersifat immutable -> harus membuat string baru, tidak bisa diubah sebagian")

kata_baru = "J" + kata[1:]
print(kata_baru)
```

## 1.3 String Methods yang Sering Dipakai

```python
teks = "  Belajar Python itu Menyenangkan  "

print(repr(teks.upper()))
print(repr(teks.lower()))
print(repr(teks.strip()))                 # menghapus spasi di awal & akhir
print(teks.strip().split())               # memecah jadi list kata (default pemisah: spasi)
print("-".join(["Data", "Science", "AI"]))  # menggabungkan list jadi string
print(teks.strip().replace("Python", "Coding"))
print(teks.lower().find("python"))        # posisi index kemunculan pertama, -1 kalau tidak ketemu
print(teks.lower().count("e"))            # menghitung kemunculan
print(teks.strip().startswith("Belajar"))
print(teks.strip().endswith("kan"))
```

⚠️ **Common Mistakes**
1. Mencoba mengubah string secara langsung lewat index (`kata[0] = "J"`) — ingat, string **immutable**
2. Index di luar jangkauan panjang string → `IndexError`
3. Lupa bahwa hampir semua string method **mengembalikan string baru**, bukan mengubah string aslinya — hasilnya harus ditampung ke variabel

```python
teks = "halo"
teks.upper()       # tidak mengubah 'teks' aslinya!
print(teks)         # masih "halo"

teks = teks.upper() # cara yang benar: tampung hasilnya kembali ke variabel
print(teks)          # sekarang "HALO"

try:
    print(teks[100])   # index jauh melebihi panjang string
except IndexError as e:
    print("IndexError:", e)
```

💡 **Tips & Best Practice**
- Gunakan slicing `[::-1]` untuk membalik string dengan cepat
- Method bisa "dirantai" (*chaining*): `teks.strip().lower().split()` — dijalankan berurutan dari kiri ke kanan
- Gunakan `in` untuk mengecek keberadaan substring, lebih ringkas daripada `.find() != -1`: `"Py" in "Python"` → `True`

# Bab 2: List

## 2.1 Membuat List, Indexing, Slicing

List adalah koleksi data yang **berurutan** dan **bisa diubah** (*mutable*), bisa berisi campuran berbagai tipe data.

```python
buah = ["apel", "jeruk", "mangga", "pisang"]

print(buah[0])
print(buah[-1])
print(buah[1:3])
print(len(buah))
```

## 2.2 List Bersifat Mutable

Berbeda dari string, isi list **bisa diubah langsung** tanpa perlu membuat list baru.

```python
buah = ["apel", "jeruk", "mangga", "pisang"]
buah[0] = "nanas"
print(buah)
```

## 2.3 List Methods yang Sering Dipakai

```python
angka = [3, 1, 4, 1, 5]

angka.append(9)          # tambah 1 elemen di akhir
print(angka)

angka.insert(0, 100)      # sisipkan elemen di posisi tertentu
print(angka)

angka.remove(1)           # hapus KEMUNCULAN PERTAMA nilai 1
print(angka)

dihapus = angka.pop()     # hapus & kembalikan elemen TERAKHIR
print("Dihapus:", dihapus, "| Sisa:", angka)

angka.sort()              # urutkan menaik (mengubah list aslinya)
print(angka)

angka.reverse()           # balik urutan
print(angka)

print("Jumlah kemunculan 1:", angka.count(1))
print("Index dari nilai 5:", angka.index(5))
```

## 2.4 List Comprehension

Cara ringkas membuat list baru dari list lain, sering menggantikan `for` loop biasa untuk kasus sederhana.

```python
angka = [1, 2, 3, 4, 5]

kuadrat = [x ** 2 for x in angka]
print(kuadrat)

genap = [x for x in angka if x % 2 == 0]
print(genap)
```

🧠 **Penjelasan Output**: pola umumnya `[ekspresi for item in urutan if kondisi]`. Bagian `if kondisi` bersifat opsional. Ini setara dengan `for` loop biasa yang menambahkan hasil ke list baru, hanya lebih ringkas.

## 2.5 Nested List (List Dua Dimensi)

```python
matriks = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(matriks[1][2])   # baris index 1, kolom index 2 -> nilai 6

for baris in matriks:
    for elemen in baris:
        print(elemen, end=" ")
    print()
```

## 2.6 Copy vs Reference

Ini salah satu jebakan paling umum bagi pemula: assignment biasa (`b = a`) **tidak** menyalin list, hanya membuat nama baru yang menunjuk ke objek yang **sama**.

```python
a = [1, 2, 3]
b = a          # b menunjuk objek yang SAMA dengan a, bukan salinan
b.append(4)
print("a:", a)  # ikut berubah, walau yang diubah 'b'!

c = a.copy()    # cara benar menyalin: .copy(), list(a), atau a[:]
c.append(100)
print("a:", a)  # a TIDAK berubah
print("c:", c)
```

⚠️ **Common Mistakes**
1. Lupa bahwa `b = a` bukan menyalin list, tapi membuat referensi ke objek yang sama
2. Index di luar jangkauan → `IndexError`
3. Menggunakan list sebagai nilai default parameter function — jebakan klasik karena Python hanya membuat objek default itu **sekali** saat function didefinisikan, lalu "diingat" terus di setiap pemanggilan

```python
def tambah_item(item, daftar=[]):   # BUG: default list dibagi antar pemanggilan!
    daftar.append(item)
    return daftar

print(tambah_item("a"))
print(tambah_item("b"))   # harusnya cuma berisi "b", tapi malah ikut membawa "a" dari panggilan sebelumnya

def tambah_item_benar(item, daftar=None):   # SOLUSI: gunakan None, baru buat list baru di dalam function
    if daftar is None:
        daftar = []
    daftar.append(item)
    return daftar

print(tambah_item_benar("a"))
print(tambah_item_benar("b"))
```

💡 **Tips & Best Practice**
- Gunakan list comprehension untuk transformasi/filter yang ringkas dan mudah dibaca
- Gunakan `.copy()`, `list(x)`, atau `x[:]` kapan pun benar-benar butuh **salinan independen**, bukan sekadar referensi
- Pilih method yang tepat: `.append()` untuk menambah satu elemen di akhir, `.extend()` untuk menggabungkan dengan list lain

# Bab 3: Tuple

## 3.1 Membuat Tuple & Immutability

Tuple mirip list, tapi bersifat **immutable** (tidak bisa diubah setelah dibuat). Ditulis dengan kurung `()`.

```python
koordinat = (10, 20)
print(koordinat)
print(type(koordinat))

titik_tunggal = (5,)   # koma WAJIB untuk tuple berisi 1 elemen
bukan_tuple = (5)       # tanpa koma, ini cuma int biasa di dalam kurung
print(type(titik_tunggal), type(bukan_tuple))

try:
    koordinat[0] = 99
except TypeError as e:
    print("TypeError:", e)
```

## 3.2 Tuple Unpacking

"Membongkar" isi tuple langsung ke beberapa variabel sekaligus.

```python
titik = (3, 7)
x, y = titik
print(x, y)

data_siswa = ("Budi", 20, "Data Science")
nama, umur, jurusan = data_siswa
print(nama, umur, jurusan)

# Trik populer: menukar (swap) nilai dua variabel tanpa variabel bantu
a, b = 1, 2
a, b = b, a
print(a, b)
```

## 3.3 Kapan Pakai Tuple, Kapan Pakai List?

- Gunakan **tuple** untuk data yang **tidak boleh berubah**: koordinat, kombinasi warna RGB, hari dalam seminggu — juga sedikit lebih hemat memori dan lebih cepat diakses dibanding list
- Gunakan **list** untuk koleksi data yang memang akan **sering diubah**: ditambah, dihapus, diurutkan ulang, dst.

⚠️ **Common Mistakes**
1. Lupa tanda koma untuk tuple berisi satu elemen — `(5)` bukan tuple, `(5,)` baru tuple
2. Mencoba mengubah isi tuple seperti mengubah list → `TypeError`
3. Memilih tuple padahal datanya justru akan sering berubah (harusnya pakai list), atau sebaliknya

💡 **Tips & Best Practice**
- Function yang perlu mengembalikan lebih dari satu nilai biasanya memakai tuple secara implisit lewat `return a, b, c`
- Gunakan tuple unpacking untuk kode yang lebih rapi dibanding mengakses lewat index satu-satu

# Bab 4: Dictionary

## 4.1 Membuat Dictionary (Key-Value)

Dictionary menyimpan data sebagai pasangan **key: value**, cocok untuk data yang punya "label" jelas, bukan sekadar posisi seperti list.

```python
mahasiswa = {
    "nama": "Xenno",
    "jurusan": "Data Science",
    "semester": 3
}

print(mahasiswa)
print(mahasiswa["nama"])
print(mahasiswa.get("jurusan"))
print(mahasiswa.get("ipk", "Belum ada data"))   # .get() bisa diberi nilai default kalau key tidak ada
```

## 4.2 Menambah, Mengubah, dan Menghapus

```python
mahasiswa["ipk"] = 3.75           # menambah key baru
print(mahasiswa)

mahasiswa["semester"] = 4          # mengubah value dari key yang sudah ada
print(mahasiswa)

del mahasiswa["ipk"]               # menghapus key tertentu
print(mahasiswa)

mahasiswa.update({"semester": 5, "status": "aktif"})   # update banyak key sekaligus
print(mahasiswa)
```

## 4.3 Looping pada Dictionary

```python
for key in mahasiswa:
    print(key)

print("---")

for key, value in mahasiswa.items():
    print(f"{key}: {value}")

print("---")

for value in mahasiswa.values():
    print(value)
```

## 4.4 Dictionary Comprehension

```python
angka = [1, 2, 3, 4, 5]
kuadrat_dict = {x: x ** 2 for x in angka}
print(kuadrat_dict)
```

## 4.5 Nested Dictionary (Dictionary Bersarang)

```python
data_siswa = {
    "Budi": {"nilai": 80, "kelas": "A"},
    "Ani": {"nilai": 95, "kelas": "B"}
}

print(data_siswa["Ani"]["nilai"])

for nama, info in data_siswa.items():
    print(f"{nama}: nilai={info['nilai']}, kelas={info['kelas']}")
```

⚠️ **Common Mistakes**
1. Mengakses key yang tidak ada dengan `[]` → `KeyError`. Kalau tidak yakin key-nya pasti ada, gunakan `.get()`
2. Key dictionary harus tipe data *immutable* (string, angka, tuple) — list **tidak bisa** dijadikan key
3. Mengira dictionary bisa diakses berdasarkan index angka seperti list — dictionary diakses berdasarkan **key**, bukan posisi

```python
try:
    print(mahasiswa["alamat"])
except KeyError as e:
    print("KeyError:", e)
    print("Solusi: gunakan mahasiswa.get('alamat', 'default value')")

print(mahasiswa.get("alamat", "Alamat belum diisi"))
```

💡 **Tips & Best Practice**
- Gunakan `.get()` untuk akses yang lebih aman saat tidak yakin sebuah key pasti ada
- Dictionary comprehension memudahkan transformasi data key-value secara ringkas
- Pilih dictionary saat data punya "nama/label" yang jelas untuk tiap nilainya; pilih list saat data hanya berupa kumpulan berurutan tanpa label khusus

# Bab 5: Set

## 5.1 Membuat Set

Set adalah koleksi data yang **unik** (tidak ada duplikat) dan **tidak terurut** (*unordered*).

```python
angka = {1, 2, 3, 2, 1}
print(angka)   # duplikat otomatis hilang

kosong_dict = {}      # PERHATIAN: ini DICT kosong, bukan set!
kosong_set = set()      # cara yang BENAR membuat set kosong
print(type(kosong_dict), type(kosong_set))
```

## 5.2 Operasi Himpunan (Union, Intersection, dkk.)

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print("Union (gabungan)              :", a | b)
print("Intersection (irisan)          :", a & b)
print("Difference (a tapi bukan di b) :", a - b)
print("Symmetric difference (hanya di salah satu):", a ^ b)
```

## 5.3 `add()`, `remove()`, `discard()`

```python
warna = {"merah", "hijau"}

warna.add("biru")
print(warna)

warna.remove("hijau")     # error kalau elemen yang dihapus tidak ada
print(warna)

warna.discard("kuning")   # TIDAK error meski elemen tidak ada di set
print(warna)
```

⚠️ **Common Mistakes**
1. Mengira `{}` adalah set kosong, padahal itu dictionary kosong — set kosong harus ditulis `set()`
2. Berharap urutan elemen set konsisten atau bisa diakses lewat index seperti list — set **tidak terurut**
3. Memakai `.remove()` pada elemen yang mungkin tidak ada → `KeyError`. Kalau tidak yakin elemennya pasti ada, gunakan `.discard()`

```python
angka = {1, 2, 3}
try:
    angka.remove(99)   # 99 tidak ada di set
except KeyError as e:
    print("KeyError:", e)
    print("Solusi: gunakan .discard(99) supaya tidak error meski elemen tidak ada")

angka.discard(99)   # aman, tidak error
print(angka)
```

💡 **Tips & Best Practice**
- Set sangat cepat untuk mengecek keanggotaan (`in`) dibanding list, terutama untuk data yang besar
- Cara cepat menghilangkan duplikat dari list: `list(set(daftar))` — tapi ingat, urutan aslinya bisa berubah
- Gunakan operasi himpunan (`|`, `&`, `-`, `^`) untuk membandingkan dua kumpulan data tanpa perlu loop manual

## 📚 Chapter 3 — `iloc[] (Row & Column Selection)`

Pada materi sebelumnya kamu sudah mempelajari:

- ✅ Memilih satu baris
- ✅ Memilih beberapa baris
- ✅ Slice baris (`0:3`, `-2:`)

Sekarang kita akan belajar memilih baris dan kolom sekaligus menggunakan `iloc`.

### 1. Dataset

```python
import pandas as pd

df = pd.DataFrame({
    "Nama": ["Andi", "Budi", "Citra", "Dewi", "Eko"],
    "Umur": [21, 23, 20, 22, 24],
    "Kota": ["Jakarta", "Bandung", "Surabaya", "Medan", "Yogyakarta"],
    "Nilai": [85, 90, 78, 88, 95]
})

print(df)
```

Output:

| Index | Nama  | Umur | Kota       | Nilai |
| ----: | ----- | ---: | ---------- | ----: |
|     0 | Andi  |   21 | Jakarta    |    85 |
|     1 | Budi  |   23 | Bandung    |    90 |
|     2 | Citra |   20 | Surabaya   |    78 |
|     3 | Dewi  |   22 | Medan      |    88 |
|     4 | Eko   |   24 | Yogyakarta |    95 |

---

### 2. Sintaks Dasar

```python
print(df.iloc[baris, kolom])
```

Urutannya selalu:

```text
[row, columns]
```

---

### 3. Mengambil Satu Nilai 

Baris ke-1, kolomke-2

```python
print(df.iloc[1, 2])
```

Output:

```text
Bandung
```

Karena:

```text
Baris 1 → Budi
Kolom 2 → Kota
```

---

### 4. Mengambil Satu Baris dan Beberapa Kolom

```python
print(df.iloc[2, 1:4])
```

Output:

```text
Umur         20
Kota    Surabaya
Nilai        78
```

---

### 5. Mengambil Beberapa Baris dan Beberapa Kolom

```python
print(df.iloc[1:4, 1:3])
```

Output:

| Umur | Kota     |
| ---: | -------- |
|   23 | Bandung  |
|   20 | Surabaya |
|   22 | Medan    |

---

### 6. Mengambil Semua Baris dan Kolom Tertentu

```python
print(df.iloc[:, 2])
```

Output:

```text
0      Jakarta
1      Bandung
2     Surabaya
3        Medan
4   Yogyakarta
```

`:` berarti semua baris.

---

### 7. Mengambil Semua Kolom dan Baris Tertentu

```python
print(df.iloc[3, :])
```

Output:

```text
Nama      Dewi
Umur        22
Kota     Medan
Nilai       88
```

---

### 8. Mengambil Blok Data

```python
print(df.iloc[1:4, 0:3])
```

Output

| Nama  | Umur | Kota     |
| ----- | ---: | -------- |
| Budi  |   23 | Bandung  |
| Citra |   20 | Surabaya |
| Dewi  |   22 | Medan    |

---

### Ringkasan 

| Kode               | Hasil                           |
| ------------------ | ------------------------------- |
| `df.iloc[1,2]`     | Satu nilai                      |
| `df.iloc[2,1:4]`   | Satu baris, beberapa kolom      |
| `df.iloc[1:4,1:3]` | Beberapa baris & beberapa kolom |
| `df.iloc[:,2]`     | Semua baris, satu kolom         |
| `df.iloc[3,:]`     | Satu baris, semua kolom         |
| `df.iloc[1:4,0:3]` | Blok data                       |

## 📚 Chapter 4 — `loc[]`

Jika `iloc[]` menggunakan posisi/index angka, maka `loc[]` menggunakan label (nama index dan nama kolom).

---
### 1. Dataset

```python
import pandas as pd

df = pd.DataFrame({
    "Nama": ["Andi", "Budi", "Citra", "Dewi", "Eko"],
    "Umur": [21, 23, 20, 22, 24],
    "Kota": ["Jakarta", "Bandung", "Surabaya", "Medan", "Yogyakarta"],
    "Nilai": [85, 90, 78, 88, 95]
})

print(df)
```

Output

| Index | Nama  | Umur | Kota       | Nilai |
| ----: | ----- | ---: | ---------- | ----: |
|     0 | Andi  |   21 | Jakarta    |    85 |
|     1 | Budi  |   23 | Bandung    |    90 |
|     2 | Citra |   20 | Surabaya   |    78 |
|     3 | Dewi  |   22 | Medan      |    88 |
|     4 | Eko   |   24 | Yogyakarta |    95 |

---

### 2. Sintaks Dasar

```python
df.loc[baris, kolom]
```

Urutannya tetap:

```text
[row_label, column_label]
```

Perbedaannya:

| `iloc[]`           | `loc[]`              |
| ------------------ | -------------------- |
| Berdasarkan posisi | Berdasarkan label    |
| `df.iloc[0, 3]`    | `df.loc[0, "Nilai"]` |

---

### 3. Mengambil Satu Nilai

```python
print(df.loc(0, 'Nilai'))
```

Output

```text
85
```

---

### 4. Mengambil Satu Kolom

```python
print(df.loc[:, 'Nama'])
```

Output:

```text
0     Andi
1     Budi
2    Citra
3     Dewi
4      Eko
```

Semua baris (`:`), hanya kolom `'Nama'`.

---

### 5. Mengambil Beberapa Kolom

```python
print(df.loc[:, ['Nama', 'Kota']])
```

Output

| Nama  | Kota       |
| ----- | ---------- |
| Andi  | Jakarta    |
| Budi  | Bandung    |
| Citra | Surabaya   |
| Dewi  | Medan      |
| Eko   | Yogyakarta |

---

### 6. Mengambil Beberapa Baris

```python
print(df.loc[1:3, :])
```

Output

| Nama  | Umur | Kota     | Nilai |
| ----- | ---: | -------- | ----: |
| Budi  |   23 | Bandung  |    90 |
| Citra |   20 | Surabaya |    78 |
| Dewi  |   22 | Medan    |    88 |

---

> Perbedaan penting dengan `iloc`:
>
> Pada `loc`, batas akhir ikut diambil (inclusive).
>
>    - `df.loc[1:3]` → indeks 1, 2, 3
>    - `df.iloc[1:3]` → indeks 1, 2

Ini adalah perbedaan yang paling sering membuat bingung pemula.

---

### 7. Mengambil Baris dan Kolom Sekaligus

```python
print(df.loc[1:3, ['Nama', 'Nilai']])
```

Output

| Nama  | Nilai |
| ----- | ----: |
| Budi  |    90 |
| Citra |    78 |
| Dewi  |    88 |

---

### 8. Mengambil Seluruh Data

```python
print(df.loc[:, :])
```

Output

Seluruh DataFrame akan ditampilkan.

---

### Ringkasan

| Kode                            | Hasil                             |
| ------------------------------- | --------------------------------- |
| `df.loc[0, "Nilai"]`            | Satu nilai                        |
| `df.loc[:, "Nama"]`             | Satu kolom                        |
| `df.loc[:, ["Nama","Kota"]]`    | Beberapa kolom                    |
| `df.loc[1:3, :]`                | Beberapa baris (batas akhir ikut) |
| `df.loc[1:3, ["Nama","Nilai"]]` | Baris & kolom sekaligus           |
| `df.loc[:, :]`                  | Seluruh DataFrame                 |

---

### ⚠️ Perbedaan `iloc` vs `loc`

| Aspek         | `iloc`               | `loc`                 |
| ------------- | -------------------- | --------------------- |
| Dasar seleksi | Posisi (integer)     | Label                 |
| Kolom         | Indeks kolom         | Nama kolom            |
| Slice baris   | Akhir **tidak ikut** | Akhir **ikut**        |
| Contoh        | `df.iloc[1:3]` → 1,2 | `df.loc[1:3]` → 1,2,3 |

# 📚 Pandas — Filtering Data

## 🎯 Materi 1: Operator Perbandingan

Filtering digunakan untuk memilih baris data yang memenuhi kondisi tertentu.

Misalkan kita memiliki DataFrame berikut:

```python
import pandas as pd

df = pd.DataFrame({
    "Nama": ["Andi", "Budi", "Citra", "Dina", "Eko"],
    "Umur": [20, 22, 19, 21, 23],
    "Nilai": [85, 70, 90, 75, 95]
})

print(df)
```

Output:

```text
    Nama  Umur  Nilai
0   Andi    20     85
1   Budi    22     70
2  Citra    19     90
3   Dina    21     75
4    Eko    23     95
```

---

### 1. Operator `>`

```python
print(df['Nilai'] > 80)
```

Output:

```text
0     True
1    False
2     True
3    False
4     True
Name: Nilai, dtype: bool
```

Belum memiliki data, tetapi menghasilkan **Series boolean**.

---

### 2. Menggunakan Boolean Mask untuk Filtering 

```python
print(df[df['Nilai'] > 80])
```

Output:

```text
    Nama  Umur  Nilai
0   Andi    20     85
2  Citra    19     90
4    Eko    23     95
```

---

### 3. Operator `<` 

```python
print(df['Umur'] < 21)
```

Output:

```text
    Nama  Umur  Nilai
0   Andi    20     85
2  Citra    19     90
```

---

### 4. Operator `>=`

```python
print(df[df['Nilai'] >= 90])
```

Output:

```text
    Nama  Umur  Nilai
2  Citra    19     90
4    Eko    23     95
```

---

### 5. Operator `<=`

```python
print(df[df['Umur'] <= 20])
```

Output:

```text
    Nama  Umur  Nilai
0   Andi    20     85
2  Citra    19     90
```

### 6. Operator `==`

```python
print(df[df['Nama'] == 'Dina'])
```

Output:

```text
   Nama  Umur  Nilai
3  Dina    21     75
```

### 7. Operator `!=`

```python
print(df[df['Nama'] != 'Budi'])
```

Output:

```text
    Nama  Umur  Nilai
0   Andi    20     85
2  Citra    19     90
3   Dina    21     75
4    Eko    23     95
```

---

### Ringkasan

| Operator | Arti                         |
| -------- | ---------------------------- |
| `>`      | Lebih besar dari             |
| `<`      | Lebih kecil dari             |
| `>=`     | Lebih besar atau sama dengan |
| `<=`     | Lebih kecil atau sama dengan |
| `==`     | Sama dengan                  |
| `!=`     | Tidak sama dengan            |

---

### 💡 Konsep Penting

Filtering di Pandas selalu mengikuti pola:

```python
df[kondisi]
```

Contoh:

```python
print(df[df['Nilai'] > 80])
```

- `df["Nilai"]` > 80 menghasilkan Series boolean.
- `df[...]` menggunakan Series tersebut untuk memilih baris yang bernilai `True`.

---

## 🎯 Materi 2: Filter Satu Kondisi

Pada materi sebelumnya kita langsung menulis kondisi di dalam `df[...]`.

Contoh:

```python
print(df[df['Nilai'] >= 80])
```

Cara ini benar, tetapi ketika kondisi mulai panjang, kode menjadi sulit dibaca.

Karena itu, kita bisa **menyimpan kondisi ke dalam variabel boolean**.

---

### 1. Membuat Boolean Mask

```python
import pandas as pd

df = pd.DataFrame({
    "Nama": ["Andi", "Budi", "Citra", "Dina", "Eko"],
    "Umur": [20, 22, 19, 21, 23],
    "Nilai": [85, 70, 90, 75, 95]
})

filter_nilai = df["Nilai"] >= 80

print(filter_nilai)
```

Output:

```text
0     True
1    False
2     True
3    False
4     True
Name: Nilai, dtype: bool
```

Variabel `filter_nilai` sekarang berisi **Series boolean**.

---

### 2. Menggunakan Boolean Mask

```python
print(df[filter_nilai])
```

Output:

```text
    Nama  Umur  Nilai
0   Andi    20     85
2  Citra    19     90
4    Eko    23     95
```

---

### 3. Contoh Lain

```python
filter_umur = df['Umur'] > 20

print(df[filter_umur])
```

Output:

```text
   Nama  Umur  Nilai
1  Budi    22     70
3  Dina    21     75
4   Eko    23     95
```

---

### 4. Filter Data String

```python
filter_nama = df['Nama'] == 'Citra'

print(df[filter_nama])
```

Output:

```text
    Nama  Umur  Nilai
2  Citra    19     90
```

---

### 5. Menggunakan Variabel Berulang Kali

Inilah keuntungan utamanya.

```python
filter_lulus = df['Nilai'] >= 80

print(df[filter_lulus])

print('\nJumlah siswa lulus:')
print(filter_lulus.sum())
```

Output:

```text
    Nama  Umur  Nilai
0   Andi    20     85
2  Citra    19     90
4    Eko    23     95

Jumlah siswa lulus:
3
```

Karena pada Python :
- `True = 1`
- `False = 0`

maka:

```python
filter_lulus.sum()
```

akan menghitung jumlah nilai `True`.

---

### 📌 Kapan Menggunakan Variabel?

Kurang disarankan:

```python
print(df[df['Nilai'] >= 80])
```

Lebih rapi:

```python
filter_lulus = df['Nilai'] >= 80

print(df[filter_lulus])
```

Terutama jika kondisi akan digunakan beberapa kali.

---

### Ringkasan

| Bentuk              | Keterangan                    |
| ------------------- | ----------------------------- |
| `df["Nilai"] >= 80` | Membuat boolean mask          |
| `mask = ...`        | Menyimpan kondisi ke variabel |
| `df[mask]`          | Memfilter DataFrame           |
| `mask.sum()`        | Menghitung jumlah `True`      |

---

## 🎯 Materi 3: Filter Multi Kondisi (`&` dan `|`)

Pada materi sebelumnya kita hanya menggunakan satu kondisi.

Contoh:

```python
df[df["Nilai"] >= 80]
```

Sekarang kita akan menggunakan **lebih dari satu kondisi**.

---

### Dataset

```python
import pandas as pd

df = pd.DataFrame({
    "Nama": ["Andi", "Budi", "Citra", "Dina", "Eko"],
    "Umur": [20, 22, 19, 21, 23],
    "Nilai": [85, 70, 90, 75, 95]
})

print(df)
```

Output:

```text
    Nama  Umur  Nilai
0   Andi    20     85
1   Budi    22     70
2  Citra    19     90
3   Dina    21     75
4    Eko    23     95
```

---

### 1. Operator `&` (AND)

Semua kondisi harus **True**.

```python
print(df[(df['Nilai'] >= 80) & (df['Umur'] > 20)])
```

Output:

```text
  Nama  Umur  Nilai
4  Eko    23     95
```

Mengapa hanya Eko?

| Nama  | Nilai ≥ 80 | Umur > 20 | Hasil |
| ----- | ---------- | --------- | ----- |
| Andi  | ✅          | ❌         | ❌     |
| Budi  | ❌          | ✅         | ❌     |
| Citra | ✅          | ❌         | ❌     |
| Dina  | ❌          | ✅         | ❌     |
| Eko   | ✅          | ✅         | ✅     |

---

### 2. Operator `|` (OR)

Minimal satu kondisi harus **True**.

```python
print(df[(df["Nilai"] >= 90) | (df["Umur"] < 20)])
```

Output:

```text
    Nama  Umur  Nilai
2  Citra    19     90
4    Eko    23     95
```

Karena:
- Citra memenuhi **dua kondisi**.
- Eko memenuhi kondisi **Nilai ≥ 90**.

---

### 3. Menggunakan Variabel Boolean 

```python
nilai_tinggi = df['Nilai'] >= 80
umur_dewasa = df['Umur'] > 20

print(df[nilai_tinggi & umur_dewasa])
```

Hasilnya sama seperti  contoh pertama.

### 4. Tiga Kondisi Sekaligus

```python
print(
    df[
        (df['Nilai'] >= 80) &
        (df['Umur'] > 20) &
        (df['Nama'] != 'Andi')
    ]
)
```

Output

```text
  Nama  Umur  Nilai
4  Eko    23     95
```

---

### ⚠️ Hal yang Paling Sering Salah

#### ❌ Salah

```python
df[df["Nilai"] >= 80 & df["Umur"] > 20]
```

Ini dapat menghasilkan error atau perilaku yang tidak diinginkan.

---

#### ✅ Benar

```python
df[(df["Nilai"] >= 80) & (df["Umur"] > 20)]
```

**Setiap kondisi wajib diberi tanda kurung `()`**.

---

### Ringkasan

| Operator | Arti                          |
| -------- | ----------------------------- |
| `&`      | AND (semua kondisi benar)     |
|  `Garis Vertikal ()`        | OR (minimal satu kondisi benar) |
| `()`     | Wajib mengapit setiap kondisi |

---

## 🎯 Materi 4: `isin()`

Pada materi sebelumnya, jika ingin memilih beberapa nilai, kita menulis seperti ini:

```python
df[
    (df["Nama"] == "Andi") |
    (df["Nama"] == "Citra") |
    (df["Nama"] == "Eko")
]
```

Cara ini benar, tetapi semakin banyak nilai yang dibandingkan, kodenya menjadi panjang.

Pandas menyediakan method `isin()` untuk mengatasi hal tersebut.

---

### Dataset

```python
import pandas as pd

df = pd.DataFrame({
    "Nama": ["Andi", "Budi", "Citra", "Dina", "Eko"],
    "Umur": [20, 22, 19, 21, 23],
    "Nilai": [85, 70, 90, 75, 95]
})

print(df)
```

Output:

```text
    Nama  Umur  Nilai
0   Andi    20     85
1   Budi    22     70
2  Citra    19     90
3   Dina    21     75
4    Eko    23     95
```

---

### 1. Memilih Beberapa Nama

```python
print(df[df['Nama'].isin(['Andi', 'Eko'])])
```

Output:

```text
   Nama  Umur  Nilai
0  Andi    20     85
4   Eko    23     95
```

---

### 2. Memilih Beberapa Nilai

```python
print(df[df['Nilai'].isin([70, 90])])
```

Output:

```text
    Nama  Umur  Nilai
1   Budi    22     70
2  Citra    19     90
```

---

### 3. Memilih Beberapa Umur

```python
print(df[df['Umur'].isin([19, 20, 23])])
```

Output:

```text
    Nama  Umur  Nilai
0   Andi    20     85
2  Citra    19     90
4    Eko    23     95
```

---

### 4. `isin()` Menghasilkan Boolean Mask

```python
mask = df['Nama'].isin(['Andi', 'Eko'])

print(mask)
```

Output:

```text
0     True
1    False
2    False
3    False
4     True
Name: Nama, dtype: bool
```

Kemudian dapat digunakan seperti biasa:

```python
print(df[mask])
```

Output:

```text
   Nama  Umur  Nilai
0  Andi    20     85
4   Eko    23     95
```

---

### 5. Menggabungkan `isin()` dengan `&`

Misalnyaingin memilih:
- Nama Andi atau Eko
- Nilai ≥ 90

```python
print(
    df[
        df['Nama'].isin(['Andi', 'Eko']) &
        (df['Nilai'] >= 90)
    ]
)
```

Output:

```text
  Nama  Umur  Nilai
4  Eko    23     95
```

---

### 6. Menggabungkan `isin()` dengan `|`

```python
print(
    df[
        df['Nama'].isin(['Andi', 'Budi']) |
        (df['Nilai'] >= 90)
    ]
)
```

Output:

```text
    Nama  Umur  Nilai
0   Andi    20     85
1   Budi    22     70
2  Citra    19     90
4    Eko    23     95
```

---

### 📌 Kapan Menggunakan isin()?

❌ Kurang praktis:

```python
df[
    (df["Nama"] == "Andi") |
    (df["Nama"] == "Citra") |
    (df["Nama"] == "Eko")
]
```

✅ Lebih ringkas:

```python
df[df['Nama'].isin(['Andi', 'Citra', 'Eko'])]
```

---

### Ringkasan

| Method              | Fungsi                                        |
| ------------------- | --------------------------------------------- |
| `.isin([...])`      | Mengecek apakah nilai termasuk dalam daftar   |
| `~.isin([...])`     | Mengecek nilai yang **tidak** termasuk daftar |
| `isin() & kondisi`  | Digabung dengan AND                           |
| `isin() \| kondisi` | Digabung dengan OR                            |

---

## 🎯 Materi 5: `between()`

Method `between()` digunakan untuk mengecek apakah suatu nilai berada **di antara dua batas**.

Daripada menulis:

```python
df[(df["Umur"] >= 20) & (df["Umur"] <= 23)]
```

Kita bisa menulis lebih singkat:

```python
df[df["Umur"].between(20, 23)]
```

Hasilnya sama, tetapi lebih mudah dibaca.

---

### Dataset

```python
import pandas as pd

df = pd.DataFrame({
    "Nama": ["Andi", "Budi", "Citra", "Dina", "Eko"],
    "Umur": [20, 22, 19, 21, 23],
    "Nilai": [85, 70, 90, 75, 95]
})

print(df)
```

Output:

```text
    Nama  Umur  Nilai
0   Andi    20     85
1   Budi    22     70
2  Citra    19     90
3   Dina    21     75
4    Eko    23     95
```

---

### 1. Filter Umur dalam Rentang

```python
print(df[df['Umur'].between(20, 23)])
```

Output:

```text
   Nama  Umur  Nilai
0  Andi    20     85
1  Budi    22     70
3  Dina    21     75
```

> `between(20, 23)` berarti 0 ≤ Umur ≤ 22.

---

### 2. Filter Nilai dalam Rentang

```python
print(
    df[
        df['Nilai'].between(75, 90)
    ]
)
```

Output:

```text
    Nama  Umur  Nilai
0   Andi    20     85
2  Citra    19     90
3   Dina    21     75
```

---

### 3. `between()` Menghasilkan Boolean Mask

```python
mask = df['Umur'].between(20, 22)

print(mask)
```

Output:

```text
0     True
1     True
2    False
3     True
4    False
Name: Umur, dtype: bool
```

Kemudian digunakan seperti biasa:

```python
print(df[mask])
```

---

### 4. Menggabungkan `between()` dengan `&`

Misalnya:
- Umur diantara 20-23
- Nilai minimal 85

```python
print(
    df[
        df['Umur'].between(20, 23) &
        (df['Nilai'] >= 85)
    ]
)
```

Output:

```text
   Nama  Umur  Nilai
0  Andi    20     85
4   Eko    23     95
```

---

### 5. Menggabungkan `between()` dengan `|`

```python
print(
    df[
        df['Nilai'].between(70, 75) | 
        (df['Nama'] == 'Eko')
    ]
)
```

Output:

```text
   Nama  Umur  Nilai
1  Budi    22     70
3  Dina    21     75
4   Eko    23     95
```

---

### 📌 Kelebihan `between()`

Tanpa `between()`:

```python
df[
    (df["Harga"] >= 1_000_000) &
    (df["Harga"] <= 5_000_000)
]
```

Dengan `between()`:

```python
df[
    df["Harga"].between(1_000_000, 5_000_000)
]
```

Lebih singkat dan lebih mudah dipahami.

---

### Ringkasan

| Method                 | Fungsi                                       |
| ---------------------- | -------------------------------------------- |
| `between(a, b)`        | Memilih nilai dari `a` sampai `b` (inklusif) |
| `between() & kondisi`  | Digabung dengan AND                          |
| `between() \| kondisi` | Digabung dengan OR                           |
| `mask = between()`     | Menyimpan boolean mask                       |

## 🎯 Materi 6: `query()`

`query()` adalah cara melakukan filtering menggunakan ekspresi string yang mirip dengan sintaks SQL.

Daripada menulis:

```python
df[(df["Nilai"] >= 80) & (df["Umur"] > 20)]
```

Kita bisa menulis:

```python
df.query('Nilai >= 80 and Umur > 20')
```

Hasilnya sama, tetapi seringkali lebih mudah dibaca.

---

### Dataset

```python
import pandas as pd

df = pd.DataFrame({
    "Nama": ["Andi", "Budi", "Citra", "Dina", "Eko"],
    "Umur": [20, 22, 19, 21, 23],
    "Nilai": [85, 70, 90, 75, 95]
})

print(df)
```

Output:

```text
    Nama  Umur  Nilai
0   Andi    20     85
1   Budi    22     70
2  Citra    19     90
3   Dina    21     75
4    Eko    23     95
```

---

### 1. Satu Kondisi

```python
print(df.query('Nilai >= 85'))
```

Output:

```text
    Nama  Umur  Nilai
0   Andi    20     85
2  Citra    19     90
4    Eko    23     95
```

---

### 2. Menggunakan `and`

```python
print(
    df.query('Nilai >= 80 and Umur > 20')
)
```

Output:

```text
  Nama  Umur  Nilai
4  Eko    23     95
```

---

### 3. Menggunakan `or`

```python
print(
    df.query('Nilai >= 90 or  Umur < 20')
)
```

Output:

```text
    Nama  Umur  Nilai
2  Citra    19     90
4    Eko    23     95
```

---

### 4. Filter String

Perhatikan penggunaan tanda kutip.

```python
print(
    df.query("Nama == 'Eko' ")
)
```

Output:

```text
  Nama  Umur  Nilai
4  Eko    23     95
```

Untuk string:
- Ekspresi `query()` menggunakan tanda kutip luar `"..."`.
- Nilai string di dalam ekspresi menggunakan tanda kutip `'...'`.

---

### 5. Kombinasi Kondisi

```python
print(
    df.query("Nama != 'Budi' and Nilai >= 85")
)
```

Output:

```text
    Nama  Umur  Nilai
0   Andi    20     85
2  Citra    19     90
4    Eko    23     95
```

---

### Perbandingan

#### Boolean Indexing 

```python
df[
    (df["Nilai"] >= 80) &
    (df["Umur"] > 20)
]
```

---

#### Query 

```python
df.query("Nilai >= 80 and Umur > 20")
```

Keduanya menghasilkan output yaang sama.

---

### Kapan Sebaiknya Meenggunakan `query()`?

Gunakan `query()` ketika:
- Kondisi filtering cukup panjang.
- Ingin kode lebih mudah dibaca.
- Tidak membutuhkan operasi yang sulit diekspresikan dalam string.

Gunakan **boolean indexing** ketika:
- Kondisi sangat kompleks
- Menggunakan method seperti `isin()`, `between()`, atau operasi lain yang lebih nyaman ditulis langsung pada Series.

---

### Ringkasan 

| Query                     | Arti              |
| ------------------------- | ----------------- |
| `query("Nilai >= 80")`    | Satu kondisi      |
| `query("A and B")`        | AND               |
| `query("A or B")`         | OR                |
| `query("Nama == 'Eko'")`  | Filter string     |
| `query("Nama != 'Budi'")` | Tidak sama dengan |

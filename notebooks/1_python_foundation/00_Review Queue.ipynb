# 📚 Review #1 — Error Handling

---

## 1. Apa itu Error?

Error adalah kondisi ketika program tidak dapat berjalan sebagaimana mestinya.

Contoh:

```python
print(10/0)
```

Output

```text
ZeroDivisionError: division by zero
```

Program langsung berhenti karena terjadi **exception**.

---

## 2. Mengapa Menggunakan Error Handling?

Tanpa Error andling

```python
angka = int(input("Masukkan angka: "))
hasil = 100 / angka

print(hasil)
```

Jika user menginput:

```text
abc
```

Program langsung berhenti.

Dengan Error Handling:

```python
try:
    angka = int(input("Masukkan angka: "))
    hasil = 100 / angka
    print(hasil)

except:
    print("Terjadi kesalahan.")
```

Program tidak crash.

---

## 3. Blok `try`

Semua kode yang **berpotensi menghasilkan exception** ditempatkan di dalam `try`.

```python
try:
    nilai = 10/2
    print(nilai)
```

Jika tidak ada error, semua kode dijalankan.

---

## 4. Blok `except`

Digunakan ketika terjadi exception.

```python
try:
    angka = int(input())

except:
    print("Input harus berupa angka.")
```

---

Lebih baik menangkap exception tertentu.

```python
try:
    angka = int(input())

except ValueError:
    print("Input bukan angka.")
```

Kenapa?

Karena kita tahu jenis error yang sedang ditangani.

---

## 5. Menangani Banyak Exception

Misalnya:

```python
try:
    angka = int(input())

    hasil = 100 / angka

except ValueError:
    print("Input harus angka/")

except ZeroDivisionError:
    print("Tidak boleh nol.")
```

Jika user memasukkan:

```python
0
```

Output

```text
Tidak boleh nol.
```

Jika memasukkan:

```text
abc
```

Output

```text
Input harus angka.
```

---

## 6. Menyimpan Objek Error

Kadang kita ingin mengetahui pesan error sebenarnya.

```python
try:
    angka = int(input())

except ValueError as e:
    print(e)
```

Misalnya input:

```text
abc
```

Output

```text
invalid literal for int()...
```

Ini berguna saat debugging.

---

## 7. Blok `else`

`else` hanya dijalankan **jika tidak terjadi exception**.

```python
try:
    angka = int(input())

except ValueError:
    print("Input salah.")

else:
    print("Input berhasil.")
```

Contoh:

Input

```text
25
```

Output

```text
Input berhasil.
```

---

## 8. Blok `finally`

`finally` **selalu digunakan**, baik ada error maupun tidak.

```python
try:
    print("Membuka file")

except:
    print("Error")

finally:
    print("Menutup file")
```

Output

```text
Mebuka file
Menutup file
```

Ini biasanya digunakan untuk:
- menutup file,
- menutup koneksi database,
- membersihkan resource.

---

## 9. Menggunakan `raise`

Kita juga bisa membuat exception sendiri.

```python
umur = -5

if umur < 0:
    raise ValueError("Umur tidak boleh negatif.")
```

Output:

```text
ValueError: Umur tidak boleh negatif.
```

---

## 10. Exception yang Sering Dijumpai 

| Exception           | Penyebab                                |
| ------------------- | --------------------------------------- |
| `ValueError`        | Nilai tidak sesuai tipe yang diharapkan |
| `TypeError`         | Operasi pada tipe data yang tidak cocok |
| `ZeroDivisionError` | Pembagian dengan nol                    |
| `IndexError`        | Index di luar batas list                |
| `KeyError`          | Key dictionary tidak ditemukan          |
| `FileNotFoundError` | File tidak ditemukan                    |

---

## 11. Best Practice

✅ Tangkap exception yang spesifik.

```python
except ValueError:
```

lebih baik daripada

```python
except:
```

---

✅ Gunakan finally untuk membersihkan resource.

---

✅ Jangan menggunakan exception untuk logika program biasa.

Kurang baik:

```python
try:
    data[10]

except:
    ...
```

Lebih baik:

```python
if len(data) > 10:
    ...
```

---

## 12.Ringkasan

```text
try
│
├── Tidak ada error
│      │
│      ├── else
│      │
│      └── finally
│
└── Ada error
       │
       ├── except
       │
       └── finally
```

---

````python
def kalkulator_pembagian():
    try:
        # 1. Minta pengguna memasukkan dua angka
        angka1 = float(input("Masukkan angka pertama: "))
        angka2 = float(input("Masukkan angka kedua (pembagi): "))
        
        # Operasi pembagian
        hasil = angka1 / angka2
        
    except ValueError:
        # 2. Tangani jika input bukan angka
        print("Error: Input yang Anda masukkan bukan angka yang valid!")
        
    except ZeroDivisionError:
        # 3. Tangani jika pembagi bernilai 0
        print("Error: Tidak dapat melakukan pembagian dengan angka nol (0)!")
        
    else:
        # 4. Jika berhasil, tampilkan hasil pembagian
        print(f"Hasil pembagian: {hasil}")
        # 5. Gunakan else untuk menampilkan pesan keberhasilan
        print("Perhitungan berhasil.")
        
    finally:
        # 6. Gunakan finally untuk menampilkan pesan selesai
        print("Program selesai.")

# Menjalankan program
kalkulator_pembagian()
````

# 📚 Review #2 — File I/O

---

## 1. Apa itu File I/O?

**File I/O (Input/Output)** adalah proses membaca data dari file atau menulis data ke file.

Contoh penggunaan:

- Menyimpan log aplikasi
- Menyimpan hasil perhitungan
- Membaca dataset CSV
- Menyimpan konfigurasi

---

## 2. Membuka FIle

Sintaks dasa:

```python
file = open("data.txt". "r")
```

Parameter kedua adalah **mode**.

| Mode  | Fungsi                                  |
| ----- | --------------------------------------- |
| `"r"` | Read (membaca)                          |
| `"w"` | Write (menulis, file lama akan ditimpa) |
| `"a"` | Append (menambah isi file)              |
| `"x"` | Membuat file baru, error jika sudah ada |

---

## 3. Membaca File

Misal isi `data.txt`:

```text
Andi
Budi
Citra
```

Kode:

```python
file = open("data.txt", "r")

isi = file.read()

print(isi)

file.close()
```

Output

```text
Andi
Budi
Citra
```

---

## 4. `.readline()`

Membaca satu baris.

```python
file = open("data.txt", "r")

print(file.readline())
print(file.readline())

file.close()
```

Output:

```text
Andi
Budi
```

---

## 5. `.readlines()`

Menghasilkan list.

```python
file = open("data.txt", "r")

isi = file.readlines()

print(isi)

file.close()
```

Output:

```python
['Andi\n', 'Budi\n', 'Citra']
```

---

## 6. Menulis File (`w`)

```python
file = open("data.txt", "w")

file.write("Halo Dunia")

file.close
```

⚠️ Perlu diingat:

Mode `"w"` menghapus isi file lama sebelum menulis data baru.

---

## 7. Menambahkan Isi File (`a`)

```python
file = open("data.txt", "a")

file.write("\nPython")

file.close()
```

Isi file menjadi:

```text
Halo Dunia
Python
```

---

## 8. Menggunakan `with`

Cara lain yang direkomendasikan:

```python
with open("data.txt", "r") as file:
    isi = file.read()

print(isi)
```

Keuntungan:

- Tidak perlu memanggil `close()`.
- File otomatis ditutup meskipun terjadi exception.
- Kode lebih rapi.

---

## 9. Menangani Error

Jika file tidak ada:

```python
with open("abc.txt", "r") as file:
    print(file.read())
```

Output:

```text
FileNotFoundError
```

Gunakan Error Handling:

```python
try:
    with open("abc.txt", "r") as file:
        print(file.read())

except FileNotFoundError:
    print("File tidak ditemukan.")
```

---

## 10. Best Prctice

✅ Gunakan `with open(...)`.

Lebih baik:

```python
with open("data.txt", "r") as file:
    print(file.read())
```

daripada:

```python
file = open("data.txt", "r")

print(file.read())

file.close()
```

Karena `with` memastikan file selalu ditutup dengan aman.

---

## 11. Ringkasan

```text
File I/O

open()
│
├── r → membaca
├── w → menulis (overwrite)
├── a → menambah
└── x → membuat file baru

↓

read()
readline()
readlines()

↓

write()

↓

with open()

↓

try-except
```

---

````python
def catatan_harian():
    # 1. Minta pengguna memasukkan sebuah catatan.
    catatan_baru = input("Masukkan catatan hari ini: ")
    
    try:
        # 2 & 3. Simpan catatan ke file catatan.txt menggunakan mode append ("a") dengan with open(...)
        with open("catatan.txt", "a") as file:
            # Menambahkan newline (\n) agar catatan berikutnya berada di baris baru
            file.write(catatan_baru + "\n")
            
        print("\n--- Isi Catatan Harian ---")
        
        # 4. Setelah berhasil menyimpan, baca kembali seluruh isi file dan tampilkan ke layar.
        # Menggunakan mode read ("r")
        with open("catatan.txt", "r") as file:
            isi_file = file.read()
            print(isi_file, end="") # end="" mencegah tambahan baris kosong ekstra dari print
            
    except FileNotFoundError:
        # 5. Tangani kemungkinan FileNotFoundError
        print("\nError: File 'catatan.txt' tidak ditemukan saat mencoba membaca.")
        
    except Exception as e:
        # Tangani exception lain yang mungkin terjadi
        print(f"\nTerjadi kesalahan yang tidak terduga: {e}")
        
    finally:
        # 6. Gunakan finally untuk menampilkan pesan selesai
        print("\nProgram selesai.")

# Menjalankan program
catatan_harian()
````

# 📚 Review #3 — Virtual Environment & `pip`

## 1. Apa itu Virtual Environment?

Virtual Environement (**venv**) adalah lingkungan Python yang **terisolasi**.

Artinya, setiap proyek dapat memiliki:
- versi Python (tergantung konfigurasi)
- package
- dependency
yang berbeda tanpa saling mengganggu.

Contoh:

```text
Project A
│
├── NumPy 1.26
└── Pandas 2.2

Project B
│
├── NumPy 2.0
└── TensorFlow
```

Keduanya bisa berjalan tanpa konflik.

---

## 2. Membuat Virtual Environent

Perintah:

```Bash
python -m venv venv
```

atau

```Bash
python3 -m venv venv
```

Setelah dijalankan akan muncul folder:

```text
project/

├── venv/
├── app.py
└── requirements.txt
```

---

## 3. Mengaktifkan Virtual Environment

### Windows

```Bash
venv\Scripts\activate
```

Jika berhaasil :

```text
(venv) C:\Project>
```

---

### Linux / macOS

```Bash
source venv/bin/activate
```

---

## 4. Menonaktifkan Virtual Environment

Cukup jalankan:

```Bash
deactivate
```

Prompt kembali normal.

---

## 5. Apa itu `pip`/

`pip` adalah package manager Python.

Digunakan untuk:
- install library
- update librar
- uninstall library
- melihat package yang terpasang

---

## 6. Install Package

Contoh:

```Bash
pip install numpy
```

atau

```Bash
pip install pandas
```

---

## 7. Melihat Package

```Bash
pip list
```

Output contoh:

```text
numpy
pandas
matplotlib
```

---

## 8. Melihat Informasi Package

```Bash
pip show numpy
```

Menampilkan informasi seperti:
- versi
- lokasi terkini
- dependency

---

## 9. Update Package

```Bash
pip install --upgrade numpy
```

---

## 10. Uninstall Package

```Bash
pip uninstall numpy
```

---

## 11. `requirements.txt`

Digunakan untuk menyimpan daftar dependency proyek.

Membuat file:

```Bash
pip freeze > reuirements.txt
```

Isi contoh:

```text
numpy==2.3.1
pandas==2.2.3
matplotlib==3.10.0
```

---

## 12. Install dari `requuirements.txt`

Pada komputer lain cukup jalankan:

```Bash
pip install -r requirements.txt
```

Semua package akan diinstal otomatis sesuai versinya.

---

## 13. Bes Practice

✅ Buat Virtual Environment untuk setiap proyek.

✅ Jangan mengunggah folder `venv/` ke GitHub.

Tambahkan ke `.gitignore`:

```text
venv/
```

✅ Simpan dependency dalam requirements.txt.

---

## 14. Ringkasan

```text
Project
│
├── python -m venv venv
│
├── activate
│
├── pip install ...
│
├── pip freeze > requirements.txt
│
├── pip install -r requirements.txt
│
└── deactivate
```

---

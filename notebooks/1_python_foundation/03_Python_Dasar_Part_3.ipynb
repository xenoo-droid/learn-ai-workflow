# Bab 1: Function (Fungsi)

## 1.1 Kenapa Butuh Function?

Bayangkan kamu perlu menghitung luas persegi panjang di 5 tempat berbeda dalam program. Tanpa function, kamu harus menulis ulang rumus yang sama berkali-kali. Function memungkinkan kode ditulis **sekali**, lalu dipakai ulang (*reusable*) kapan pun dibutuhkan — prinsip ini disebut **DRY** (*Don't Repeat Yourself*).

````python
def sapa():
    print("Halo, selamat datang!")

# Memanggil (menjalankan) function
sapa()
sapa()
sapa()
````

🧠 **Penjelasan Output**: `def` mendefinisikan function bernama `sapa`. Kode di dalam function **tidak langsung jalan** saat didefinisikan — baru dijalankan saat dipanggil dengan `sapa()`. Karena dipanggil 3 kali, pesannya tercetak 3 kali.

## 1.2 Parameter dan Return Value

Function bisa menerima input (**parameter**) dan mengembalikan hasil (**return value**).

````python
def luas_persegi_panjang(panjang, lebar):
    hasil = panjang * lebar
    return hasil

luas1 = luas_persegi_panjang(5, 3)
luas2 = luas_persegi_panjang(10, 2)

print(luas1)
print(luas2)
````

🧠 **Penjelasan Output**: `panjang` dan `lebar` adalah **parameter** — nilai yang "dititipkan" saat function dipanggil. `return` mengirim hasil perhitungan kembali ke tempat function dipanggil, sehingga bisa disimpan ke variabel (`luas1`, `luas2`) atau dipakai langsung.

⚠️ Function tanpa `return` sebenarnya tetap mengembalikan nilai, yaitu `None`.

````python
def tanpa_return(x):
    print(f"Nilai x adalah {x}")
    # tidak ada 'return'

hasil = tanpa_return(10)
print("Hasil yang disimpan:", hasil)   # None, karena function tidak mengembalikan apa pun secara eksplisit
````

## 1.3 Default Parameter Value

Parameter bisa diberi nilai default, sehingga boleh tidak diisi saat function dipanggil.

````python
def sapa_nama(nama, sapaan="Halo"):
    print(f"{sapaan}, {nama}!")

sapa_nama("Xenno")              # sapaan pakai nilai default "Halo"
sapa_nama("Xenno", "Selamat pagi")   # sapaan di-override
````

## 1.4 Positional Argument vs Keyword Argument

Argumen bisa dikirim berdasarkan **urutan** (positional) atau berdasarkan **nama parameter** (keyword) — keyword argument tidak harus berurutan dan lebih jelas maksudnya.

````python
def buat_profil(nama, umur, kota):
    print(f"{nama}, {umur} tahun, tinggal di {kota}")

# Positional -> urutan HARUS sesuai definisi
buat_profil("Xenno", 20, "Surabaya")

# Keyword -> urutan bebas, karena nama parameter disebutkan
buat_profil(kota="Surabaya", nama="Xenno", umur=20)
````

## 1.5 `*args` dan `**kwargs`

Dipakai saat jumlah argumen yang diterima function **tidak diketahui pasti** jumlahnya.
- `*args` menampung sejumlah argumen positional sebagai `tuple`
- `**kwargs` menampung sejumlah argumen keyword sebagai `dict`

````python
def jumlahkan_semua(*args):
    total = 0
    for angka in args:
        total += angka
    return total

print(jumlahkan_semua(1, 2, 3))
print(jumlahkan_semua(10, 20, 30, 40, 50))
````

````python
def tampilkan_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

tampilkan_info(nama="Xenno", jurusan="Data Science", semester=3)
````

## 1.6 Docstring

Docstring adalah dokumentasi function, ditulis dengan tiga tanda kutip (tiga petik tunggal atau tiga petik ganda berurutan) tepat di baris pertama setelah `def`, bisa diakses lewat `help()`.

````python
def hitung_bmi(berat_kg, tinggi_m):
    '''Menghitung Body Mass Index (BMI).

    Parameter:
        berat_kg (float): berat badan dalam kilogram
        tinggi_m (float): tinggi badan dalam meter

    Return:
        float: nilai BMI
    '''
    return berat_kg / (tinggi_m ** 2)

print(hitung_bmi(65, 1.7))
help(hitung_bmi)
````

⚠️ **Common Mistakes**
1. Lupa `return`, lalu bingung kenapa hasil function `None`
2. Menulis `print()` di dalam function padahal seharusnya `return` — `print` hanya menampilkan ke layar, tidak bisa disimpan atau dipakai perhitungan lain
3. Parameter tanpa default diletakkan **setelah** parameter yang punya default → `SyntaxError`
4. Lupa bahwa argumen positional harus dikirim **sebelum** argumen keyword saat memanggil function

````python
# Contoh error #3 (ditangani dengan try supaya notebook tetap jalan)
try:
    exec("def contoh(a, b=1, c):\n    pass")
except SyntaxError as e:
    print("SyntaxError:", e)
    print("Solusi: parameter tanpa default harus di depan -> def contoh(a, c, b=1):")
````

💡 **Tips & Best Practice**
- Beri nama function yang jelas menggambarkan **aksi**, biasanya berupa kata kerja: `hitung_luas()`, bukan `luas()`
- Satu function idealnya melakukan **satu tugas** saja (*Single Responsibility*)
- Gunakan docstring untuk function yang cukup kompleks atau akan dipakai orang lain
- Gunakan keyword argument saat memanggil function dengan banyak parameter, agar kode lebih mudah dibaca

# Bab 2: Scope

## 2.1 Local Scope vs Global Scope

**Scope** adalah aturan mengenai di bagian kode mana sebuah variabel bisa "dilihat"/diakses.
- **Local scope**: variabel yang dibuat di dalam function, hanya bisa diakses di dalam function itu sendiri
- **Global scope**: variabel yang dibuat di luar semua function, bisa diakses dari mana saja (termasuk di dalam function, untuk **dibaca**)

````python
nama_global = "Xenno"   # variabel global

def tampilkan_nama():
    print(nama_global)   # BISA membaca variabel global dari dalam function

tampilkan_nama()
````

````python
def buat_variabel_lokal():
    pesan_lokal = "Ini variabel lokal"
    print(pesan_lokal)

buat_variabel_lokal()

try:
    print(pesan_lokal)   # ERROR! variabel lokal tidak bisa diakses dari luar function
except NameError as e:
    print("NameError:", e)
````

## 2.2 Keyword `global`

Secara default, mengisi nilai ke sebuah nama variabel **di dalam function** akan membuat variabel **lokal baru**, bukan mengubah variabel global yang namanya kebetulan sama. Untuk benar-benar mengubah variabel global dari dalam function, gunakan keyword `global`.

````python
skor = 0   # variabel global

def tambah_skor_salah():
    skor = skor + 10   # ERROR! Python menganggap 'skor' di sini variabel lokal, tapi belum diisi nilai

try:
    tambah_skor_salah()
except UnboundLocalError as e:
    print("UnboundLocalError:", e)
````

````python
skor = 0   # variabel global

def tambah_skor_benar():
    global skor          # menyatakan: 'skor' yang dimaksud adalah variabel global, bukan lokal baru
    skor = skor + 10

tambah_skor_benar()
tambah_skor_benar()
print(skor)   # 20
````

## 2.3 Aturan LEGB (Sekilas)

Saat Python mencari sebuah nama variabel, urutan pencariannya adalah:

**L**ocal → **E**nclosing → **G**lobal → **B**uilt-in

Artinya Python akan mengecek scope lokal (dalam function saat ini) dulu, lalu scope di sekitarnya (kalau ada function di dalam function), lalu scope global, dan terakhir baru fungsi/nama bawaan Python (seperti `print`, `len`).

````python
x = "global"

def luar():
    x = "enclosing"
    def dalam():
        x = "local"
        print("Di dalam fungsi dalam():", x)
    dalam()
    print("Di dalam fungsi luar():", x)

luar()
print("Di scope global:", x)
````

⚠️ **Common Mistakes**
1. Mengira mengubah variabel di dalam function otomatis mengubah variabel global (padahal tanpa `global`, itu membuat variabel lokal baru)
2. Memakai `global` secara berlebihan — ini biasanya tanda desain program kurang baik, sebaiknya pakai `return` untuk mengirim nilai keluar dari function
3. Bingung kenapa variabel lokal "menghilang" setelah function selesai dijalankan (memang sengaja begitu — itulah gunanya scope)

````python
def contoh():
    hasil_sementara = 100 * 2   # variabel lokal, sengaja hanya "hidup" selama function berjalan
    return hasil_sementara

nilai = contoh()   # cara yang BENAR untuk "mengeluarkan" nilai dari function: pakai return
print(nilai)
````

💡 **Tips & Best Practice**
- Sebisa mungkin hindari `global` — lebih baik kirim data masuk lewat parameter, dan keluarkan lewat `return`
- Nama variabel lokal boleh sama dengan nama variabel global tanpa masalah — keduanya benar-benar terpisah kecuali sengaja dihubungkan dengan `global`
- Jika function mulai butuh banyak `global`, itu pertanda function tersebut sebaiknya didesain ulang

# Bab 3: Lambda (Fungsi Anonim)

## 3.1 Sintaks Lambda

Lambda adalah cara menulis function **ringkas** dalam satu baris, tanpa nama (*anonymous function*). Cocok untuk logika sederhana yang hanya dipakai sekali atau sebentar.

```
lambda parameter: ekspresi
```

````python
# Function biasa
def kuadrat(x):
    return x ** 2

# Function yang sama, ditulis sebagai lambda
kuadrat_lambda = lambda x: x ** 2

print(kuadrat(5))
print(kuadrat_lambda(5))
````

🧠 **Penjelasan Output**: kedua cara menghasilkan nilai yang sama persis. Lambda tidak butuh `return` — hasil ekspresi setelah `:` otomatis dikembalikan.

## 3.2 Lambda dengan Banyak Parameter

````python
tambah = lambda a, b: a + b
print(tambah(3, 4))

is_genap = lambda x: x % 2 == 0
print(is_genap(10))
print(is_genap(7))
````

## 3.3 Lambda sebagai Argumen Function Lain

Kegunaan lambda yang paling umum adalah sebagai argumen sementara untuk function lain, seperti `sorted()`, `map()`, dan `filter()`.

````python
siswa = [("Budi", 80), ("Ani", 95), ("Citra", 70)]

# Mengurutkan berdasarkan nilai (elemen index ke-1 dari tiap tuple), bukan berdasarkan nama
siswa_terurut = sorted(siswa, key=lambda item: item[1], reverse=True)
print(siswa_terurut)
````

````python
angka = [1, 2, 3, 4, 5]

# map() -> menerapkan lambda ke SETIAP elemen
kuadrat_semua = list(map(lambda x: x ** 2, angka))
print(kuadrat_semua)

# filter() -> menyaring elemen yang membuat lambda bernilai True
angka_genap = list(filter(lambda x: x % 2 == 0, angka))
print(angka_genap)
````

⚠️ **Common Mistakes**
1. Mencoba menulis banyak baris/statement di dalam lambda — lambda hanya boleh berisi **satu ekspresi**, tidak boleh ada `if` blok, `for`, atau banyak baris
2. Memaksakan lambda untuk logika yang rumit, sehingga kode malah sulit dibaca — sebaiknya pakai `def` biasa untuk logika kompleks
3. Lupa membungkus hasil `map()`/`filter()` dengan `list()` saat ingin melihat isinya langsung (keduanya mengembalikan objek *iterator*, bukan list)

````python
angka = [1, 2, 3]
hasil = map(lambda x: x * 2, angka)

print(hasil)          # <map object ...> -> bukan list yang langsung terlihat isinya
print(list(hasil))    # baru terlihat isinya setelah dibungkus list()
````

💡 **Tips & Best Practice**
- Gunakan lambda hanya untuk logika **pendek dan sederhana**, idealnya muat dalam satu baris
- Kalau logikanya butuh nama supaya jelas maksudnya, atau butuh lebih dari satu baris, gunakan `def` biasa
- Lambda paling sering dipakai sebagai argumen `key=` di `sorted()`, atau di dalam `map()`/`filter()`

# Bab 4: Modules

## 4.1 Apa itu Module?

Module adalah file Python (`.py`) berisi kumpulan function, variabel, atau class yang bisa dipakai di file lain dengan cara **import**. Python juga menyediakan banyak module bawaan (*standard library*) yang siap pakai tanpa install apa pun, misalnya `math`, `random`, dan `datetime`.

````python
import math

print(math.sqrt(16))     # akar kuadrat
print(math.pi)            # nilai konstanta pi
print(math.ceil(4.2))     # pembulatan ke atas
print(math.floor(4.8))    # pembulatan ke bawah
````

````python
import random

print(random.randint(1, 10))     # angka acak antara 1-10 (inklusif keduanya)
buah = ["apel", "jeruk", "mangga"]
print(random.choice(buah))        # pilih satu elemen acak dari list
````

````python
import datetime

sekarang = datetime.date(2026, 8, 12)
print(sekarang)
print(sekarang.year, sekarang.month, sekarang.day)
````

## 4.2 Variasi Cara Import

- `import nama_module` — panggil isinya dengan `nama_module.sesuatu`
- `import nama_module as alias` — beri nama pendek/alias supaya lebih ringkas ditulis
- `from nama_module import sesuatu` — langsung ambil bagian tertentu, tidak perlu tulis nama module lagi

````python
import math as m
print(m.sqrt(25))   # alias 'm' menggantikan 'math'

from math import sqrt, pi
print(sqrt(36))      # tidak perlu 'math.' lagi
print(pi)
````

⚠️ Hati-hati memakai `from module import *` (mengimpor semua isi module sekaligus) — ini bisa menimbulkan konflik nama antar variabel/function dari module berbeda, dan membuat kode sulit dilacak asalnya. Sebaiknya import spesifik saja yang dibutuhkan.

## 4.3 Membuat Module Sendiri

Module buatan sendiri cukup dibuat sebagai file `.py` biasa. Misalnya kamu membuat file `kalkulator.py` berisi beberapa function, lalu di file lain (dalam folder yang sama) kamu bisa memanggil `import kalkulator` untuk memakai isinya. Karena notebook ini adalah satu file tunggal, kita simulasikan konsepnya di sini dengan menuliskan "module" langsung sebagai kumpulan function dalam satu cell.

````python
# Simulasikan isi sebuah "module" bernama kalkulator
def kalkulator_tambah(a, b):
    return a + b

def kalkulator_kurang(a, b):
    return a - b

# Di file .py sungguhan, dua function di atas akan disimpan di kalkulator.py
# lalu dipanggil dari file lain dengan: import kalkulator
print(kalkulator_tambah(10, 5))
print(kalkulator_kurang(10, 5))
````

## 4.4 `dir()` dan `help()`

`dir(module)` menampilkan daftar semua nama (function, variabel, dst.) yang tersedia di dalam sebuah module. `help(module)` menampilkan dokumentasinya.

````python
import math

print(dir(math)[:15])   # tampilkan 15 nama pertama saja supaya tidak terlalu panjang
# help(math)             # aktifkan baris ini untuk melihat dokumentasi lengkap module math
````

⚠️ **Common Mistakes**
1. Lupa `import` module sebelum memakainya → `NameError: name 'math' is not defined`
2. Salah ketik nama module atau nama function di dalamnya → `ModuleNotFoundError` atau `AttributeError`
3. Menamai file `.py` sendiri sama seperti nama module bawaan (misalnya menyimpan file bernama `random.py`) — ini bisa membuat Python bingung dan meng-import file sendiri, bukan module bawaan

````python
try:
    print(matematika.sqrt(9))   # sengaja salah ketik & lupa import
except NameError as e:
    print("NameError:", e)
    print("Solusi: pastikan sudah 'import math' dan menulis 'math.sqrt(9)' dengan benar")
````

💡 **Tips & Best Practice**
- Letakkan semua `import` di **bagian paling atas** file/notebook, sebelum kode lainnya (konvensi standar Python/PEP 8)
- Gunakan alias (`as`) untuk module yang sering dipakai dan namanya panjang
- Hindari `from module import *` kecuali benar-benar tahu isinya dan yakin tidak ada konflik nama

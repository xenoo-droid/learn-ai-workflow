# Bab 1: Percabangan (Conditional Statement)

## 1.1 Pernyataan `if`

Percabangan memungkinkan program mengambil **keputusan** — menjalankan blok kode tertentu hanya jika suatu kondisi bernilai `True`.

Di Python, blok kode di dalam `if` ditandai dengan **indentasi** (biasanya 4 spasi), bukan kurung kurawal `{}` seperti di bahasa lain. Ini wajib, kalau salah indentasi program akan error.

```python
umur = 20

if umur >= 17:
    print("Kamu sudah cukup umur untuk membuat KTP")
    print("Baris ini juga bagian dari blok if, karena masih terindentasi")

print("Baris ini SELALU dijalankan, karena tidak terindentasi di dalam if")
```

🧠 **Penjelasan Output**: karena `umur >= 17` bernilai `True`, kedua baris `print()` di dalam blok `if` ikut dijalankan. Baris terakhir tetap dijalankan apa pun hasil kondisinya, karena berada di luar blok `if`.

## 1.2 `if-else`

Menjalankan blok kode berbeda tergantung kondisi `True` atau `False`.

```python
nilai = 65

if nilai >= 70:
    print("LULUS")
else:
    print("TIDAK LULUS")
```

## 1.3 `if-elif-else`

Untuk mengecek **lebih dari dua** kemungkinan kondisi secara berurutan. Python akan berhenti di kondisi pertama yang bernilai `True` dan mengabaikan sisanya.

```python
nilai = 82

if nilai >= 90:
    grade = "A"
elif nilai >= 80:
    grade = "B"
elif nilai >= 70:
    grade = "C"
elif nilai >= 60:
    grade = "D"
else:
    grade = "E"

print(f"Nilai {nilai} mendapat grade {grade}")
```

🧠 **Penjelasan Output**: meskipun `nilai = 82` juga memenuhi syarat `nilai >= 70` dan `nilai >= 60`, Python **berhenti** di kondisi pertama yang `True` (`nilai >= 80`) dan tidak mengecek kondisi-kondisi setelahnya. Urutan `elif` sangat penting!

## 1.4 Nested if (if di dalam if)

`if` bisa diletakkan di dalam blok `if` lain untuk mengecek kondisi yang lebih spesifik.

```python
umur = 25
punya_sim = True

if umur >= 17:
    print("Umur memenuhi syarat")
    if punya_sim:
        print("Boleh mengendarai mobil sendiri")
    else:
        print("Perlu membuat SIM dulu")
else:
    print("Umur belum memenuhi syarat")
```

## 1.5 Ternary Operator (Conditional Expression)

Cara ringkas menulis `if-else` sederhana dalam satu baris, biasanya dipakai saat langsung mengisi nilai ke variabel.

```python
umur = 20

# Cara biasa
if umur >= 18:
    status = "Dewasa"
else:
    status = "Anak-anak"
print(status)

# Cara ternary (satu baris, hasilnya sama persis)
status2 = "Dewasa" if umur >= 18 else "Anak-anak"
print(status2)
```

⚠️ **Common Mistakes**
1. Lupa titik dua (`:`) setelah kondisi `if`, `elif`, atau `else`
2. Indentasi tidak konsisten (campur tab dan spasi, atau jumlah spasi beda antar baris) → `IndentationError`
3. Menggunakan `=` (assignment) padahal maksudnya `==` (perbandingan) di dalam kondisi `if`
4. Urutan `elif` yang salah, sehingga kondisi yang lebih spesifik tidak pernah tercapai karena keburu ditangkap kondisi yang lebih umum di atasnya

```python
# Contoh kesalahan #4: urutan elif yang salah
nilai = 95

# Kondisi paling umum diletakkan di ATAS -> kondisi spesifik di bawahnya tidak akan pernah tercapai
if nilai >= 60:
    grade = "Lulus (tapi info grade detail hilang)"
elif nilai >= 90:
    grade = "A"   # baris ini TIDAK PERNAH tercapai untuk nilai >= 90, karena sudah ditangkap kondisi pertama
print(grade)

print("---")

# Versi yang benar: kondisi paling SPESIFIK/ketat diletakkan di ATAS
nilai = 95
if nilai >= 90:
    grade = "A"
elif nilai >= 60:
    grade = "Lulus"
print(grade)
```

💡 **Tips & Best Practice**
- Gunakan 4 spasi untuk setiap level indentasi (standar PEP 8), jangan campur dengan tab
- Urutkan kondisi `elif` dari yang **paling spesifik/ketat** ke yang paling umum
- Untuk `if-else` sederhana yang hanya mengisi satu variabel, pertimbangkan ternary operator agar kode lebih ringkas
- Hindari nested if yang terlalu dalam (lebih dari 2-3 level) — biasanya bisa disederhanakan dengan `elif` atau kondisi gabungan (`and`/`or`)

# Bab 2: Perulangan (Loop)

## 2.1 `for` Loop dan `range()`

`for` loop dipakai untuk mengulang blok kode sebanyak jumlah elemen dalam suatu urutan (sequence). Fungsi `range()` menghasilkan urutan angka yang sering dipakai bersama `for`.

```python
# range(n) -> menghasilkan angka dari 0 sampai n-1
for i in range(5):
    print(i)
```

```python
# range(mulai, berhenti) -> dari 'mulai' sampai 'berhenti - 1'
for i in range(2, 7):
    print(i)

print("---")

# range(mulai, berhenti, langkah)
for i in range(0, 10, 2):
    print(i)
```

🧠 **Penjelasan Output**: `range(mulai, berhenti, langkah)` **tidak pernah** menyertakan nilai `berhenti` itu sendiri — berhenti tepat sebelum nilai tersebut. `range(0, 10, 2)` menghasilkan `0, 2, 4, 6, 8`, bukan sampai `10`.

## 2.2 Looping Langsung pada Data

`for` juga bisa langsung mengulang isi string atau list, tanpa perlu `range()`. Pembahasan detail tentang list ada di **Part 4**.

```python
nama = "Budi"
for huruf in nama:
    print(huruf)

print("---")

buah = ["apel", "jeruk", "mangga"]
for item in buah:
    print(item)
```

## 2.3 `while` Loop

`while` mengulang blok kode **selama** kondisinya masih `True`. Berbeda dengan `for` yang jumlah pengulangannya biasanya sudah pasti, `while` cocok dipakai ketika jumlah pengulangan **tidak diketahui di awal**.

```python
hitung = 1
while hitung <= 5:
    print(hitung)
    hitung += 1   # WAJIB ada, kalau tidak akan infinite loop!

print("Selesai")
```

⚠️ Kalau baris `hitung += 1` dihapus atau lupa ditulis, kondisi `hitung <= 5` akan **selamanya** `True`, dan program akan terjebak di **infinite loop** (loop tak berhenti) sampai dihentikan paksa.

## 2.4 `break`, `continue`, dan `pass`

- `break` — menghentikan loop sepenuhnya, langsung keluar
- `continue` — melewati sisa kode di iterasi saat ini, lanjut ke iterasi berikutnya
- `pass` — tidak melakukan apa-apa, hanya "penanda tempat" (placeholder) supaya kode tetap valid secara sintaks

```python
# break -> berhenti total begitu ketemu angka 5
for i in range(10):
    if i == 5:
        break
    print(i)

print("--- contoh continue ---")

# continue -> lewati angka genap, hanya cetak yang ganjil
for i in range(10):
    if i % 2 == 0:
        continue
    print(i)
```

```python
# pass -> placeholder, berguna saat kerangka kode belum selesai ditulis
for i in range(5):
    if i == 3:
        pass   # TODO: nanti diisi logika khusus untuk i == 3
    print(i)
```

## 2.5 Klausa `else` pada Loop

Fitur unik Python: `for` dan `while` bisa punya blok `else`, yang dijalankan **hanya jika loop selesai secara normal** (tidak dihentikan oleh `break`).

```python
# else TIDAK dijalankan karena loop dihentikan oleh break
for i in range(5):
    if i == 3:
        break
    print(i)
else:
    print("Loop selesai tanpa break")

print("---")

# else DIJALANKAN karena loop selesai normal, tanpa break
for i in range(5):
    print(i)
else:
    print("Loop selesai tanpa break")
```

## 2.6 Nested Loop (Loop di Dalam Loop)

Sering dipakai untuk data dua dimensi, misalnya mencetak pola atau tabel.

```python
for baris in range(1, 4):
    for kolom in range(1, 4):
        print(f"({baris},{kolom})", end="  ")
    print()   # pindah baris baru setelah loop dalam selesai
```

```python
# Contoh pola segitiga bintang menggunakan nested loop
tinggi = 5
for baris in range(1, tinggi + 1):
    for kolom in range(baris):
        print("*", end="")
    print()
```

⚠️ **Common Mistakes**
1. Lupa mengupdate variabel penghitung di `while` → **infinite loop**
2. Salah paham urutan `range()`: `range(5)` menghasilkan `0` sampai `4`, **bukan** `1` sampai `5`
3. Menaruh `break`/`continue` di luar loop → `SyntaxError`
4. Lupa bahwa `continue` tetap melanjutkan ke iterasi berikutnya, bukan menghentikan loop sepenuhnya (sering tertukar dengan `break`)

```python
# Contoh: infinite loop yang DIPERBAIKI dengan batas pengaman
hitung = 0
batas_pengaman = 0

while hitung < 5:
    print(hitung)
    hitung += 1
    batas_pengaman += 1
    if batas_pengaman > 1000:   # jaga-jaga kalau logikanya salah
        print("Dihentikan paksa, kemungkinan ada bug!")
        break
```

💡 **Tips & Best Practice**
- Gunakan `for` ketika jumlah pengulangan sudah pasti/diketahui (misal: mengulang isi list, atau `range()` dengan angka tetap)
- Gunakan `while` ketika pengulangan bergantung pada kondisi yang bisa berubah saat program berjalan (misal: menunggu input valid dari pengguna)
- Saat menulis `while`, selalu pastikan ada sesuatu di dalam loop yang **mengubah** kondisi, agar loop pasti berhenti
- Hindari nested loop lebih dari 2–3 level — biasanya menandakan kode perlu dipecah jadi function (dibahas di **Part 3**)

# 📊 Pandas → Merge & Join

## ⭐ Materi 1: `merge()`

`merge()` digunakan untuk **menggabungkan dua DataFrame** berdasarkan kolom atau key yang memiliki hubungan.

Contoh sederhana:

```python
import pandas as pd

df_siswa = pd.DataFrame({
    'ID': [1, 2, 3],
    'Nama': ['Andi', 'Budi', 'Citra']
})

df_nilai = pd.DataFrame({
    'ID': [1, 2, 3],
    'Nilai': [85, 90, 95]
})

gabung = pd.merge(df_siswa, df_nilai, on='ID')

print(gabung)
```

Output:

```text
   ID   Nama  Nilai
0   1   Andi     85
1   2   Budi     90
2   3  Citra     95
```

### 🧠 Konsep inti

```text
df_siswa                 df_nilai
┌────┬───────┐           ┌────┬───────┐
│ ID │ Nama  │           │ ID │ Nilai │
├────┼───────┤           ├────┼───────┤
│ 1  │ Andi  │           │ 1  │ 85    │
│ 2  │ Budi  │     +     │ 2  │ 90    │
│ 3  │ Citra │           │ 3  │ 95    │
└────┴───────┘           └────┴───────┘
             │
             ▼
        pd.merge(..., on="ID")
             │
             ▼
┌────┬───────┬───────┐
│ ID │ Nama  │ Nilai │
├────┼───────┼───────┤
│ 1  │ Andi  │ 85    │
│ 2  │ Budi  │ 90    │
│ 3  │ Citra │ 95    │
└────┴───────┴───────┘
```

### 🔑 Parameter

```python
pd.merge(df1, df2, on='ID')
```

- `df1` → DataFrame pertama
- `df2` → DataFrame kedua
- `on="ID"` → kolom yang digunakan sebagai key

Untuk sekarang fokus dulu pada `merge()` + `on`. Kita belum masuk `how`, `concat()`, atau `join()`.

---

## ⭐ Materi 2: `concat()`

### 🎯 Konsep utama

`concat()` digunakan untuk menggabungkan DataFrame secara vertikal atau horizontal.

### 1. Vertical — menambah baris

```python
import pandas as pd

df1 = pd.DataFrame({
    'Nama': ['Andi', 'Budi'],
    'Nilai': [80, 85]
})

df2 = pd.DataFrame({
    'Nama': ['Citra', 'Doni'],
    'Nilai': [90, 88]
})

hasil = pd.concat([df1, df2])

print(hasil)
```

Output:

```text
    Nama  Nilai
0   Andi     80
1   Budi     85
0  Citra     90
1   Doni     88
```

Perhatikan bahwa index lama ikut terbawa.

### 2. Reset Index

Kalau ingin index dibuat ulang:

```python
hasil = pd.concat([df1, df2], ignore_index=True)
```

Output:

```text
    Nama  Nilai
0   Andi     80
1   Budi     85
2  Citra     90
3   Doni     88
```

Jadi:

```text
pd.concat([df1, df2])
        ↓
gabungkan DataFrame
        ↓
ignore_index=True
        ↓
buat index baru
```

---

### 3. Horizontal — menambah kolom

`concat()` juga bisa dilakukan berdasarkan kolom dengan:

```python
hasil = pd.concat([df1, df1], axis=1)
```

`axis=1` berarti DataFrame digabung berdampingan secara horizontal.

---

### 🧠 Bedakan dengan merge()

#### `merge()`

```text
merge()
   ↓
menggabungkan berdasarkan KEY
   ↓
contoh: ID
```

#### `concat()`

```text
concat()
   ↓
menyatukan DataFrame
   ↓
vertikal / horizontal
```

Jadi jangan menganggap `concat()` sebagai versi lain dari `merge()`.

## ⭐ Materi 3: `join()`

### Basic Join
`join()` digunakan untuk menggabungkan DataFrame berdasarkan index.

🔑 Sintaks:

```python
df1.join(df2)
```

Artinya `df2` digabungkan ke `df1` berdasarkan index.

Contoh:

```python
import pandas as pd

df_nama = pd.DataFrame({
    "Nama": ["Andi", "Budi", "Citra"]
}, index=[1, 2, 3])

df_nilai = pd.DataFrame({
    "Nilai": [80, 90, 95]
}, index=[1, 2, 3])

hasil = df_nama.join(df_nilai)

print(hasil)
```

Hasil:

```text
    Nama  Nilai
1   Andi     80
2   Budi     90
3  Citra     95
```

---

#### 🧠 Cara berpikirnya

```text
df_nama                    df_nilai
index                      index
  1 → Andi                  1 → 80
  2 → Budi                  2 → 90
  3 → Citra                 3 → 95
       │                       │
       └─────────┬─────────────┘
                 ▼
           df_nama.join(df_nilai)
                 │
                 ▼
          index menjadi acuan
```

Jadi untuk membedakan tiga method tadi :

```text
merge()
   → berdasarkan KEY / kolom

concat()
   → berdasarkan AXIS

join()
   → berdasarkan INDEX
```

---

### `how='inner'`

Pada Basic `join()`, kita sudah melihat bahwa DataFrame kiri menjadi acuan.

Sekarang `how='inner'` digunakan ketika kita hanya ingin mengambil index yang terdapat di kedua DataFrame.

Contoh:

```python
df_nama = pd.DataFrame({
    "Nama": ["Andi", "Budi", "Citra"]
}, index=[1, 2, 3])

df_nilai = pd.DataFrame({
    "Nilai": [90, 95, 88]
}, index=[2, 3, 4])
```

Jika:

```python
hasil = df_nama.join(
    df_nilai,
    how='inner'
)
```

Maka outputnya:

```text
   Nama  Nilai
2  Budi     90
3  Citra    95
```

Kenapa begitu?...karena:

```text
df_nama   → 1, 2, 3
df_nilai  → 2, 3, 4

index yang ADA DI KEDUANYA:
             2, 3
             ↓
          inner join
```

Jadi:
> `how='inner'` → **hanya mempertahankan index yang cocok di kedua DataFrame**.

#### 🧠 Bedakan dengan Basic `join()`

```text
df_nama.join(df_nilai)
        ↓
default
        ↓
mengikuti index df_nama
        ↓
1, 2, 3
```

Sedangkan:

```text
df_nama.join(df_nilai, how="inner")
        ↓
hanya index yang cocok
        ↓
2, 3
```

---

### `how='outer'`

`how="outer"` mempertahankan **semua index dari kedua DataFrame**.

Contoh:

```python
df1 = pd.DataFrame({
    "Nama": ["Andi", "Budi", "Citra"]
}, index=[1, 2, 3])

df2 = pd.DataFrame({
    "Nilai": [80, 90, 95]
}, index=[2, 3, 4])
```

Dengan:

```python
hasil = df1.join(
    df2,
    how="outer"
)
```

Output:

```text
   Nama  Nilai
1  Andi    NaN
2  Budi   80.0
3  Citra  90.0
4  NaN    95.0
```

#### 🧠 Cara berpikir

```text
df1 → 1, 2, 3
df2 →    2, 3, 4
       ─────────
outer → 1, 2, 3, 4
```

Berbeda dengan `inner`:

```text
inner → hanya 2, 3
outer → 1, 2, 3, 4
```

Jadi:

```text
join()
├── default
│   └── mengikuti index DataFrame kiri
│
├── how="inner"
│   └── hanya index yang sama di kedua DataFrame
│
└── how="outer"
    └── semua index dari kedua DataFrame
```

#### ⚠️ Perhatikan `NaN`

Ketika sebuah index hanya terdapat di salah satu DataFrame, kolom dari DataFrame lainnya tidak memiliki nilai.

Contoh:

```text
index 1 → hanya ada di df1
         → Nilai = NaN

index 4 → hanya ada di df2
         → Nama = NaN
```

Ini penting karena nanti akan berhubungan dengan materi Missing Value yang sudah pernah kita pelajari.

# 📁 Pandas → Import / Export

Import/export penting karena data nyata sering datang dari file, lalu setelah dianalisis hasilnya perlu disimpan kembali.

---

## ⭐ 1. CSV

### Membaca CSV

```python
df = pd.read_csv('data.csv')

print(df)
```

---

`read_csv()` digunakan untuk **mengimpor data CSV menjadi DataFrame**.

### Menyimpan CSV

```python
df.to_csv('hasil.csv', index=False)
```

`to_csv()` digunakan untuk **mengekspor DataFrame ke CSV**.

`index=False` berarti index DataFrame tidak ikut ditulis sebagai kolom tambahan.

Contoh:

```python
import pandas as pd

df = pd.DataFrame({
    'Nama': ['Andi', 'Budi', 'Citra'],
    'Nilai': [85, 90, 95]
})

df.to_csv('nilai.csv', index=False)

hasil = pd.read_csv('nilai.csv')

print(hasil)
```

Output:

```text
    Nama  Nilai
0   Andi     85
1   Budi     90
2  Citra     95
```

Perhatikan bahwa kita melakukan dua operasi berbeda:

```python
df.to_csv(...)
```

→ Export

```python
pd.read_csv(...)
```

→ Import

---

### 🧠 Catatan penting

Jangan tertukar:

```text
pd.read_csv()
    ↓
Import → File → DataFrame

df.to_csv()
    ↓
Export → DataFrame → File
```

---

## ⭐ 2. Excel

Pola dasarnya sama seperti CSV:

```text
Excel File
   ↓
pd.read_excel()
   ↓
DataFrame
```

dan:

```text
DataFrame
   ↓
df.to_excel()
   ↓
Excel File
```

---

### Membaca Excel — `read_excel()`

Untuk membaca file Excel:

```python
df = pd.read_excel('data.xlsx')
```

---

### Menyimpan Excel

Untuk export ke Excel:

```python
df.to_excel('hasil.xlsx', index=False)
```

Sama seperti CSV, `index=False` mencegah index DataFrame menjadi kolom tambahan.

---

## Perbedaan CSv dan Excel

| CSV | Excel |
|---|---|
| `pd.read_csv()` | `pd.read_excel()` |
| `df.to_csv()` | `df.to_excel()` |
| `.csv` | `.xlsx` |
| Format teks/tabular | Workbook Excel |

Jadi jangan sampai tertukar:

```python
pd.read_excel("data.xlsx")   # Import
df.to_excel("hasil.xlsx")    # Export
```

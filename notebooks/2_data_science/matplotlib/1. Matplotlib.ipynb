# 📈 Matplotlib → Pengenalan & Line Plot

## 1. `import matplotlib.pyplot as plt`

Matplotlib adalah library Python yang digunakan untuk membuat visualisas.

Kita menggunnakan:

```python
import matplotlib.pyplot as plt
```

`plt` adalah alias yang umum digunakan untuk mengakses fungsi-fungsi plotting.

Pada tahap ini belum ada grafik yang dibuat, kita baru **mengimpor modul plotting**.

---

## 2. `plt.plot()`

Untuk membuat **line plot**, kita menggunakan:

```python
plt.plot(x, y)
```

Contoh sederhana:

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y)
plt.show()
```

Secara konsep:

```text
x → posisi horizontal
y → nilai vertikal

(1,10) ── (2,20)
             \
              (3,15) ── (4,30) ── (5,25)
```

`plt.plot(x, y)` menghubungkan titik-titikk tersebut menjadi sebuah **grafis**.

---

## 📌 Pola yang perlu kamu ingat

```python
pl.plot(x, y)
```

bukan:

```python
plt.plot(y, x)
```

Karena secara umum:

> x = **data horizontal**
> y = **data vertikal**

## `plt.show()`

### 1. Apa fungsi `plt.show()`

`plt.show()` digunakan untuk menampilkan grafik yang sudah dibuat.

Contoh:

```python
import matplotlib.pyplot as plt

hari = [1, 2, 3, 4, 5]
suhu = [28, 29, 30, 31, 29]

plt.plot(hari, suhu), 
plt.show()
```

Urutannya:

```text
plt.plot()
    ↓
Membuat/menyiapkan grafik
    ↓
plt.show()
    ↓
Menampilkan grafik
```

#### 🧠 Bedakan dengan `plt.plot()`

```python
plt.plot(hari, suhu)
```

> → menentukan/membuat line plot berdasarkan data.

```python
plt.show()
```

> → menampilkan grafik tersebut.

Jadi keduanya memiliki fungsi yang berbeda.

---

### 2. Apakah `plt.show()` mengubah data?

**_Tidak_**.

Misalnya:

```python
hari = [1, 2, 3]
suhu = [28, 30, 29]

plt.plot(hari, suhu)
plt.show()
```

`hari` dan `suhu` tetap berisi data yang sama.

`plt.show()` hanya berkaitan dengan **penampilan visualisasi**, bukan manipulasi data.

---

### 3. Posisi `plt.show()`

Biasanya `plt.show()` diletakkan **setelah seluruh konfigurasi grafik selesai**:

```python
plt.plot(...)
plt.xlabel(...)
plt.ylabel(...)
plt.title(...)
plt.grid()

plt.show()
```

Jadi pola dasar yang sekarang sudah kita pelajari adalah:

```text
import
  ↓
plot
  ↓
xlabel
  ↓
ylabel
  ↓
title
  ↓
grid
  ↓
show
```

# 📈 Matplotlib → Bar Plot

# 📈 Matplotlib → Histogram

## Apa itu Histogram?

Histogram digunakan untuk melihat distribusi data numerik.

Berbeda dengan `plt.bar()`:

- Bar Plot → membandingkan kategori.
- Histogram → melihat bagaimana angka tersebar dalam beberapa rentang nilai (bins).

Contoh sederhana:

```python
import matplotlib.pyplot as plt

nilai = [55, 60, 62, 65, 70, 72, 75, 78, 80, 85, 90, 95]

plt.hist(nilai)

plt.show()
```

Histogram akan mengelompokkan nilai ke dalam beberapa rentang secara otomatis.

---

### 🔑 `plt.hist()`

Struktur dasarnya:

```python
plt.hist(data)
```

Contoh:

```python
nilai = [55, 60, 62, 65, 70, 72, 75, 78, 80, 85, 90, 95]

plt.hist(nilai)

plt.show()
```

**`bins`**

Kita juga bisa menentukan jumlah kelompok:

```python
plt.hist(nilai, bins=5)
```

Semakin besar jumlah `bins`, semakin banyak rentang yang digunakan untuk mengelompokkan data.

---

## 🧠 Bedakan dengan Bar Plot

Misalnya:

```python
produk = ["Laptop", "Mouse", "Keyboard"]
terjual = [15, 40, 25]
```

ini cocok untuk:

```python
plt.bar(produk, penjualan)
```

karena `Laptop`, `Mouse`, dan `Keyboard` adalah **kaategoori**.

Sedangkan:

```python
nilai = [55, 60, 62, 65, 70, 72, 75, 78, 80, 85, 90, 95]
```

cocok untuk:

```python
pl.hist(nilai)
```

karena kita ingin melihat distribusi nilai.

# 📈 Matplotlib → Subplots

`subplot()` digunakan untuk menampilkan **beberapa plot dalam satu figure**.

Sintaks dasar:

```python

```

Contoh:

```python

y1 = [10, 20, 15, 25]
y2 = [5, 15, 10, 20]

```

Artinya:

```text
subplot(1, 2, 1)
        │  │  └── plot ke-1
        │  └───── 2 kolom
        └──────── 1 baris

subplot(1, 2, 2)
        │  │  └── plot ke-2
        │  └───── 2 kolom
        └──────── 1 baris
```

Hasilnya kira-kira:

```text
┌──────────────┬──────────────┐
│    Data 1    │    Data 2    │
│     plot     │     plot     │
└──────────────┴──────────────┘
```

## Pola yang perlu kamu pahami

Untuk 2 baris x 1 kolom:

```python

```

Hasil:

```text
┌──────────────┐
│    Plot 1    │
├──────────────┤
│    Plot 2    │
└──────────────┘
```

Jadi angka terakhir menentukan posisi plot.

# 🖼️ Figure & Axes (`fig`, `ax/axes`)

## 1️⃣ Mental Model Paling Penting

Bayangkan Matplotlib seperti sebuah **kanvas dengan area gambar**.

```text
🖼️ Figure
│
└── 📊 Axes
      ├── 📈 Plot
      ├── 🏷️ Title
      ├── ↔️ X-axis
      ├── ↕️ Y-axis
      ├── 🏷️ Labels
      └── 📌 Legend
```

### 🖼️ `Figure`

`Figure` adalah **wadah atau keseluruhan kanvas**.Satu figure dapat memiliki satu atau beberapa `Axes`.

### 📊 `Axes`

`Axes` adalah area **plot individual** tempat data benar-benar divisualisasikan.

> ⚠️ Ini bagian yang sering membingungkan: `Axes` **bukan berarti sekadar garis sumbu X dan Y**. Dalam Matplotlib, `Axes` adalah seluruh area plot individual.

Dokumentasi resmi Matplotlib membedakan `Figure` sebagai container tingkat atas dan`Axes` sebagai objek yang membungkus elemen-elemen sebuah subplot individual.

---

## 2️⃣ Cara Membuat `Figure` dan `Axes`

Cara yang paling umum dan direkomendasikan untuk pola object-oriented:

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
```

Output:

```text
fig → Figure
ax  → Axes
```

Kemudian kita bekerja secara eksplisit dengan `ax`:

```python
ax.plot([1, 2, 3], [2, 4, 6])
ax.set_title('Grafik Saya')

plt.show()
```

Matplotlib secara resmi membedakan pendekatan **object-based/explicit (Axes interface)** dan pendekatan `pyplot` yang lebih implicit. Untuk proyek yang semakin kompleks, memhamai pedektana `fig, ax` sangat penting.

---

## 3️⃣ Bandingkan dengan Cara yang Sudah Kamu Gunakan

Sebelumnya kamu mungkin sering mengetik:

```python
plt.plot(x, y)
plt.title('Penjualan')
plt.xlabel('Bulan')
plt.ylabel('Pendapatan')
```

Dengan pendekatan `Figure & Axes`:

```python
fig, ax = plt.subplot()

ax.plot(x, y)
ax.set_title('Penjualan')
ax.set_xlabel('Bulan')
ax.set_ylabel('Pendapatan')

plt.show()
```

### Perhatikan polanya 👇

| `pyplot` | Object-Oriented |
|---|---|
| `plt.plot()` | `ax.plot()` |
| `plt.title()` | `ax.set_title()` |
| `plt.xlabel()` | `ax.set_xlabel()` |
| `plt.ylabel()` | `ax.set_ylabel()` |
| `plt.legend()` | `ax.legend()` |

Intinya: sekarang kita tidak hanya memberi instruksi ke Matplotlib secara umum. Kita secara eksplisit mengatakan **Axes mana yang ingin kita ubah**.

---

## 🧠 Kenapa Ini Penting?

Bayangkan kamu punya 4 subplot:

```text
Figure
├── Axes 1 📈
├── Axes 2 📊
├── Axes 3 📉
└── Axes 4 🔵
```

Dengan `Axes`, kita bisa menentukan dengan jelask plot mana yang ingin dimodifikasi:

```python
fig, axes = plt.subplots(2, 2)

axes[0, 0].plot(...)
axes[0, 1].bar(...)
axes[1, 0].hist(...)
axes[1, 1].scatter(...)
```

Objek axes dapat diakses menggunakan indeks matriks 2D [baris, kolom] yang dimulai dari angka 0:

- `axes[0, 0]`: Baris ke-0, Kolom ke-0 (grafik kiri atas)
- `axes[0, 1]`: Baris ke-0, Kolom ke-1 (grafik kanan atas)
- `axes[1, 0]`: Baris ke-1, Kolom ke-0 (grafik kiri bawah)
- `axes[1, 1]`: Baris ke-1, Kolom ke-1 (grafik kanan bawah)

`plt.subplots()` memang menghasilkan Figure beserta satu atau kumpulan Axes, tergantung layout yang dibuat.

---

## 🔥 Contoh Lengkap

```python
import matplotlib.pyplot as plt

bulan = ["Jan", "Feb", "Mar", "Apr"]
penjualan = [120, 150, 130, 180]

fig, ax = plt.subplots(figsize=(8, 4))

ax.plot(bulan, penjualan, marker="o")
ax.set_title("Penjualan Bulanan")
ax.set_xlabel("Bulan")
ax.set_ylabel("Jumlah Penjualan")
ax.grid()

plt.show()
```

Mari baca kodenya seperti manusia:

```text
plt.subplots()
      ↓
Buat Figure + Axes
      ↓
ax.plot()
      ↓
Plot data pada Axes
      ↓
ax.set_...
      ↓
Atur elemen pada Axes tersebut
```

---

## ⚠️ Satu Konsep yang Harus Kamu Pegang

Jangan menghafal:

> "`fig` untuk ini, `ax` untuk itu."

Lebih baik pahami:

> `fig` = **keseluruhan wadah visual**
> `ax` = **tempat spesifik tempat visualisasi dibuat**

Mental model ini akan terus berguna saat kita masuk ke Boxplot, Errorbar, dan visualisasi yang lebih kompleks.

## 🔥 Mengapa Kadang Namanya `ax`, Kadang `axes`?

Perhatikan dua kode dibawah ini:

### Satu PLot

```python
fig, ax = plt.subplots()
```

Di sini kita punya satu Axes, sehingga variabel `ax` menyimpan satu objek Axes.

### Banyak Plot

```python
fig, axes = plt.subplots(2, 2)
```

Di sini kit punya banyak Axes, sehingga `axes` menyimpan kumpulan obejk Axes.

Secara sederhana:

```text
1 Axes  → ax
banyak Axes → axes
```

Contohnya:

```python
fig, axes = plt.subplots(2, 2)

axes[0, 0].plot(...)
axes[0, 1].plot(...)
axes[1, 0].plot(...)
axes[1, 1].plot(...)
```

⚠️ Catatan penting: nama `ax` dan `axes` sebenarnya hanyalah nama variabel yang kita pilih. Yang penting adalah memahami objek yang dikembalikan oleh `plt.subplots()`.

## 📐 Mengatur Ukuran Figure dengan `figsize`

`figsize` digunakan untuk menentukan ukuran FIgure:

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 6))
```

Angka tersebut berart:

```text
figsize = (lebar, tinggi)
          (10,     6)
```

Satuannya menggunakan **_inch_**, bukan pixel.

**Kenapa ini penting?**

Bayangkan kamu punya4 subplot dalam Figure yang kecil:

```text
🖼️ Figure kecil
┌─────┬─────┐
│plot │plot │
├─────┼─────┤
│plot │plot │
└─────┴─────┘
```

Plot dan tulisan bisa terasa sempit. Dengan `figsize`, kita memberikan ruang yang lebih nyaman dan luas:

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
```

> 💡 `figsize` **mengatur ukuran keseluruhan Figure, bukan ukuran satu Axes secara individual**.

---

## 🧠 Hubungan `figsize` dengan Layout

Bandingkan:

```python
fig, axes = plt.subplots(1, 1, figsize=(10, 5))
```

dengan:

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 5))
```

Keduanya memiliki **Figure size yang sama**, tetapi pada kasus kedua, ruang tersebut harus dibagi ke **4 Axes**.

Jadi mental modelnya:

```text
Ukuran Figure tetap
        ↓
Jumlah Axes bertambah
        ↓
Ruang per Axes menjadi lebih kecil
```

Ini penting saat kamu mulai membuat dashboard atau visualisasi dengan banyak plot.

---

### 🔥 Contoh Praktis

Misalnya:

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
```

Artinya:

```text
🖼️ Figure berukuran 12 × 8
│
├── 📊 Axes[0, 0]
├── 📊 Axes[0, 1]
├── 📊 Axes[1, 0]
└── 📊 Axes[1, 1]
```

Keempat Axes berbagi ruang di dalam Figure tersebut.

## Merapikan Layout dengan `tight_layout()`

Saat kita memiliki banyak subplot, kadang terjadi masalah seperti ini:

```text
Judul/label saling bertabrakan 😵
        ↓
Visualisasi sulit dibaca
```

Matplotlib menyediakan:

```python
plt.thight_layout()
```

Fungsinya adalah membantu **mengatur jarak antar-subplot secara otomatis** agar elemen seperti judul dan label tidak saling bertumpuk.

Contoh:

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 6))

# kode untuk membuat plot...

plt.tight_layout()
plt.show()
```

### ⚠️ Hubungannya dengan fig.suptitle()

Kalau kita menggunakan judul utama Figure:

```python
fig,suptitle('Laporan Visualisasi')
```

kadang `tight_layout()` bisa membuat ruang untuk judul utama terasa kurang ideal. Salah satu cara yang bisa digunakan:

```python
plt.tight_layout(rect=[0, 0, 1, 0.95])
```

Artinya secara sederhana: kita meminta layout subplot menggunakan sebagian besar Figure sambil menyisakan ruang di bagian atas untuk `suptitle`.

# 📦 MATPLOTLIB — BOX PLOT

## 1️⃣ Apa itu Boxplot?

Boxplot atau **box-and-whisker plot** adalah visualisasi statistik yang membantu kita melihat:
- pusat data → **median**
- penyebaran data → **quartile & IQR**
- nilai ekstrem → **outlier**
- perbandingan distribusi antar kelompok

Bayangkan data nilai ujian:

```text
50, 60, 65, 70, 72, 75, 80, 85, 95
```

Daripada melihat semua angka satu per satu boxplot merangkum pola distribusinya.

---

## 2️⃣ Anatomi Boxplot 🧠

Komponen utamanya:

```text
          Outlier
             ●

        │──────────│
        │          │
     ┌──┴──────────┴──┐
     │       │         │
     │      Median     │
     └─────────────────┘
       Q1          Q3
     ←──── IQR ────→
```

### 📌 Istilah penting

| Komponen | Arti |
|---|---|
| **Q1** | 25% data berada di bawah nilai ini |
| **Median (Q2)** | Nilai tengah / 50% data berada di bawahnya |
| **Q3** | 75% data berada di bawah nilai ini |
| **IQR** | `Q3 - Q1` |
| **Whisker** | Rentang data non-outlier |
| **Outlier/Flier** | Data yang berada di luar batas whisker |

Seacar default, Matplotlib menggunakan aturan *1.5 × IQR* untuk menentukan batas whisker, titik di luar batas tersebut ditampilkan sebagai outlier. Penting: whisker tidak selalu berarti nilai minimum dan maksimum absolut.

---

## 3️⃣ Membuat Boxplot Pertama

```python

plt.show()
```

## 🧠 Cara Membaca Boxplot

Jangan hanya melihat bentuknya. Biasakan bertanya:
1. Di mana Median?
→ Kelompok mana yang nilai tengahnya lebih tinggi?
2. Seberapa Besar Box-nya?
→ Box yang lebih besar berarti IQR lebih besar, sehingga 50% data tengah lebih tersebar.
3. Apakah ada Outlier?
→ Ada data yang jauh dari pola mayoritas.
4. Apakah distribusi antar kelompok berbeda?
→ Median sama belum tentu distribusinya sama.

⚠️ Insight penting: Boxplot adalah alat untuk memahami **distribusi**, bukan sekadar mencari nilai tertinggi atau terendah.

Ini berguna ketika ingin mengatur, misalnya, tampilan box secara lebih spesifik.

# ❌ Matplotlib → Errorbar

## 1. 🤔 Apa itu Error Bar?

Bayangkan kamu memiliki hasil rata-rata nilai ujian:

```text
Kelas A → rata-rata 80
Kelas B → rata-rata 82
```

Sekilas, Kelas B terlihat lebih baik. Tapi ada pertanyaan penting:

> **Seberapa yakin kita terhadap angka tersebut? Seberapa besar variasi atau ketidakpastiannya?**

Di sinilah **error bar berguna**.

Secara visual:

```text
        │
        │  ← error / uncertainty
        ●  ← nilai utama
        │
        │  ← error / uncertainty
```

Error bar dapat menunjukkan **variablitas atau ketidakpastian**, tetapi arti tepatnya bergantung pada data/statistik yang digunakan—misalnya standard deviation, standar error, atau confiendence interval. Karena itu, error bar harus selalu diberi konteks yang jelas.

> ⚠️ Mental model penting: Error bar bukan berarti datanya “salah”. Error bar memberi informasi tambahan tentang nilai yang sedang kita tampilkan.

---

## 2. 🧠 Struktur Dasar `errorbar()`

Di Matplotlib, kita bisa menggunakan:

```python
plt.errorbar(x, y, yerr=error)
```

Atau dengan pendekatan Figure & Axes yang sudah kamu kuasai:

```python
fig, ax = plt.subplots()

ax.errorbar(x, y, yerr=error)

plt.show()
```

Matplotlib secara resmi mendefinisikan `errorbar()` untuk memplot data `x` dan `y` dengan error bar yang melekat pada titik/garis data. `yerr` mengatur ukuran error vertikal, sedangkan `xerr` untuk error horizontal.


---

## 3. 📊 Contoh Pertama — Error Vertikal

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 15, 13, 18]

error = [1, 2, 1.5, 2]

plt.errorbar(x, y, yerr=error)

plt.show()
```

Mari baca titik pertama:

```text
y = 10
error = 1

Range visual:
9 ── 10 ── 11
```

Jadi secara visual, error bar membentan dari:

```text
y - error
```

sampai:

```text
y + error
```

Jika error bersifat simetris, Matlpotlib menerima satu nilai error per titik dan menggambar rentang `±` di sekitar nilai tersebut.

---

## 4. ↔️ `xerr` vs ↕️ `yerr`

Kita tidak hanya bisa menunjukkan ketidakpastian pada sumbu Y.

### `yerr`

Menampilkan error secara vertikal:

```python
plt.errorbar(x, y, yerr=error)
```

### `xerr`

Menampilkan error secara hroizontal:

```python
plt.errorbar(x, y, xerr=erro)
```

### Keduanya Sekaligus

```python
plt.errorbar(
    x, y,
    xerr=x_error,
    yerr=y_error
)
```

Matplotlib mendukung error pada kedua arah ini dalam satu plot.

---

## 5. 🎨 Membuat Error Bar Lebih Jelas

Biasanya kita ingin titik utama terlihat jelas dan ujung error bar memiliki 'cap'.

```python
plt.errorbar(
    x, y,
    yerr=error,
    fmt='o',
    capsize=5
)

plt.show()
```

### Parameter Penting:

| Parameter | Fungsi |
|---|---|
| `fmt='o'` | Menampilkan marker berbentuk titik |
| `capsize=5` | Memberikan garis kecil di ujung error bar |
| `yerr` | Ukuran error vertikal |
| `xerr` | Ukuran error horizontal |

Contoh resmi Matplotlib juga menggunakan `fmt` dan `capsize` untuk membuat error bar lebih mudah dibaca.

---

## 6. 🔥 Error yang Berbeda untuk Setiap Titik

Error tidak harus sama.

```python
x = [1, 2, 3]
y = [10, 15, 12]

error = [1, 3, 0.5]

plt.errorbar(
    x,
    y,
    yerr=error,
    fmt='o',
    capsize=5
)
plt.show()
```

Artinya:

```text
Titik 1 → error ±1
Titik 2 → error ±3
Titik 3 → error ±0.5
```

Ini masuk akal karena setiap pengukuran atau estimasi bisa memiliki tingkat ketidakpastianyang berbeda.

---

## 🧩 Hubungan Errorbar dengan Boxplot

Ini bagian penting karena kamu telah belajar Boxplot.

Keduanya bisa berbicara tentang sesuatu yang berkaitan dengan variasi, tetapi tidak sama:

```text
📦 Boxplot
→ Menunjukkan distribusi data
→ Median, Q1, Q3, IQR, outlier

⚠️ Errorbar
→ Menunjukkan rentang error/ketidakpastian yang didefinisikan
→ Bisa berupa SD, SE, CI, atau ukuran lain
```

Jadi jangan otomatiss berpikir:
> "Error bar = Boxplot versi sederhana."

❌ Tidak tepat.

Error bar harus dijelaskan artinya, karena bentuk visual yang sama dapat merepresentasikan ukuran statistik yang berbeda.

---

## 🧠 Insight untuk Data Professional

Misalkan ada dua rata-rata:

```text
Produk A → 100
Produk B → 105
```

Tanpa konteks:
> "Produk B lebih tinggi."

Dengan error bar, kita mungkin melihat:

```text
A → 100 ± 2
B → 105 ± 20
```

Sekarang interpretasinya menjadi lebih hati-hati.

**Pelajaran penting**: jangan membuat keputusan hanya dari angka pusat tanpa memahami variabilitas dan ketidakpastian di baliknya.

Namun, satu kesalahan umum yang harus kita hindari: **overlap atau tidak-overlap error bar saja tidak otomatis membuktikan ada/tidaknya perbedaan statistik yang signifikan**; interpretasinya bergantung pada jenis error bar yang digunakan.

---

## 🧠 Mental Model yang Harus Kamu Pegang

```text
Nilai utama
    +
Informasi tentang ketidakpastian/variasi
    =
Interpretasi yang lebih lengkap
```

Dan ingat:
> **Error bar adalah visualisasi informasi statistik, bukan bukti otomatis bahwa sesuatu signifikan atau tidak signifikan.**

```text
nilai ± error
```

Tetapi terkadang error ke atas dan ke bawah berbeda.

Misalnya:

```text
Nilai utama = 100

Error bawah = 10
Error atas  = 20
```

Maka rentangnya:

```text
90 ─── ● 100 ───────── 120
```

Matplotlib bisa menangani error asimetris dengan format data tertentu. Kita akan bahas ini setelah kamu benar-benar nyaman dengan bentuk dasar error bar—jangan lompat ke kompleksitas dulu. 😎

---

## 🎯 Errorbar — Level Up: Error Asimetris

Sampai sekarang kita memakai error **simetris**:

```text
      +5
       │
       ● 50
       │
      -5
```

Artinya jarak ke atas dan ke bawah sama: 50 ± 5.

Tapi data nyata tidak selalu seperti itu. Kadang ketidakpastian ke atas dan ke bawah **berbeda**.

---

### 1️⃣ Konsep Error Asimetris

Misalnya:

```text
Nilai utama = 100
Error bawah = 10
Error atas  = 20
```

Maka:

```text
       120
        │  +20
        │
       ● 100
        │
        │  -10
        90
```

Jadi kita tidak bisa lagi menulis sederhana:

```text
100 ± ?
```

Karena kedua sisiinya berbeda.

---

### 2️⃣ Cara Memberikannya ke Matplotlib

Untuk `yerr`, Matplotlib dapat menerima error bawah dan error atas secara terpisah.

Contoh:

```python
import matplotlib.pyplot as plt

x = [1, 2, 3]
y = [100, 80, 120]

lower_error = [10, 5, 15]
upper_error = [20, 10, 5]

y_error = [lower_error, upper_error]

plt.errorbar(
    x, y,
    yerr=y_error,
    fmt='o',
    capsize=5
)

plt.grid()
plt.show()
```

#### 🧠 Cara membacanya

Untuk titik pertama:

```text
y = 100
lower error = 10
upper error = 20

Range → 90 sampai 120
```

Titik kedua:

```text
y = 80
lower error = 5
upper error = 10

Range → 75 sampai 90
```

---

### ⚠️ Mental Model Penting: Urutan Error

Formatnya:

```python
y_error = [
    lower_error,
    upper_error
]
```

Jadi:

```text
index pertama → error bawah
index kedua   → error atas
```

Bukan sebaliknya. Kalau tertukar, grafikmu tetap bisa muncul, tetapi makna visualnya menjadi salah. Ini contoh bagus bahwa kode yang berjalan belum tentu analisisnya benar. 😈

---

### 3️⃣ Kita Integrasikan dengan Figure & Axes

Sekarang gunakan pendekatan yang lebih terstruktur:

```python
fig, ax = plt.subplots(figsize=(8, 5))

ax.errorbar(
    x,
    y,
    yerr=y_error,
    fmt='o',
    capsize=5
)
ax.set_title('Data dengan Error Asimetris')
ax.set_xlabel('Observasi')
ax.set_ylabel('Nilai')
ax.grid()

plt.show()
```

🔥 Di sini kamu memakai beberapa skill sekaligus:

```text
Figure & Axes
      +
Styling & Labels
      +
Errorbar
      =
Visualisasi yang lebih informatif
```

Ini yang kita inginkan ke depan: materi baru menambah kemampuan, bukan menggantikan materi lama.

---

### 🧠 Contoh Cara Berpikir

Misalkan hasil pengukuran sebuah metrik:

```text
Produk A → 100
Error bawah → 10
Error atas → 30
```

Rentangnya:

```text
90 ─────── 100 ───────────────── 130
```

Kalau dibandingkan dengan:

```text
Produk B → 110 ± 5
```

Jangan buru-buru mengatakan:
> ❌ “Produk B pasti lebih baik.”

Kita perlu tahu dulu **apa arti error tersebut**. Tetapi kita sudah bisa melakukan observasi yang valid:

> 🟢 Produk A memiliki error/ketidakpastian yang lebih besar dan tidak simetris, sedangkan Produk B memiliki rentang yang lebih sempit berdasarkan error yang ditampilkan.

Itulah perbedaan antara **observasi** dan **kesimpulan yang terlalu jauh**.

---

## 🔑 Mental Model Errorbar Asimetris

Sekarang kamu punya tiga hal yang berbeda yang harus dibedakan:

```text
📏 Ukuran error
→ Seberapa besar error/ketidakpastian

↕️ Rentang error
→ Nilai bawah sampai nilai atas

⚖️ Simetri
→ Apakah error atas dan bawah sama besar?
```

# 📐 Matplotlib : Axis Limits & Scale

## 1️⃣ Axis Limits — Mengatur Area yang Terlihat

Bayangkan kamu punya grafik dengan data dari tahun 2010-2025, tetapi ingin fokus melihat periode 2020-2025 saja.

Kita bisa menggunakan:

```python
ax.set_xlim(2020, 2025)
ax.set_ylim(0, 100)
```

Atau dengan gaya `pyplot`:

```python
plt.xlim(2020, 2025)
plt.ylim(0, 100)
```

Contoh:

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6, 7]
y = [10, 15, 12, 30, 25, 40, 35]

fig, ax = plt.subplots()

ax.plot(x, y)
ax.set_xlim(2, 6)
ax.set_ylim(10, 40)

plt.show()
```

📌 Mental model penting:

> `set_xlim()` dan `set_ylim()` tidak menghapus data. Mereka hanya menentukan bagian mana dari grafik yang ditampilkan.

Secara default, Matplotlib melakukan autoscaling berdasarkan data. Ketika kita menetapkan limits secara eksplisit, kita mengambil kontrol atas _view_ grafik tersebut.

---

## 2️⃣ Axis Scale — Cara Angka Didistribusikan

Ini berbeda dengan limits.

**Axis limits bertanya**:
> "Rentang mana yang ingin saya lihat?"

**Axis scale bertanya**:
> "Bagaimana angka pada rentang tersebut dipetakan ke posisi visual?"

Secara default, Matplotlib menggunakan linear sclae.

Misalnya:

```text
0 ─── 10 ─── 20 ─── 30
```

Jarak antara 0→10 dan 10→20 sama.

Untuk data yang memiliki rentang nilai sangat besar, kita kadang menggunakan logarithmic scale. Matplotlib mendukung pengaturan skala melalui `set_xscale()` dan `set_yscale()`.

### Pilihan Jenis Skala (Nilai String)
- `'linear'` : Skala standar dengan jarak antar angka yang sama.
- `'log'` : Skala logaritmik basis 10 (cocok untuk data yang melonjak drastis).
- `'symlog'` : Skala logaritmik simetris (bisa digunakan jika data memiliki nilai negatif).
- `'logit'` : Skala untuk data probabilitas antara angka 0 dan 1.

---

## 3️⃣ Linear Scale vs Log Scale

Misalnya data pertumbuhan:

```text
1 → 10 → 100 → 1000 → 10000
```

Pada linear sclae, nilai kecil bisa terlihat "menempel" karena nilai besar mendominasi grafik.

Pada log sclae, pola perubahan antar orde besaran menjadi lebih mudah dibandingkan.

Contoh:

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [1, 10, 100, 1000, 10000]

fig, ax = plt.subplots()

ax.plot(x, y, marker='o')
ax.set_yscale('log')

plt.show()
```

⚠️ Catatan penting: **nilai nol dan negatif tidak dapat ditampilkan secara normal pada log scale**, sehingga kita harus berhati-hati memilih data dan skala.

---

## 🧠 Mental Model Utama

```text
DATA
 │
 ▼
AUTOSCALE (default)
 │
 ├── set_xlim / set_ylim
 │       ↓
 │   "Bagian mana yang terlihat?"
 │
 └── set_xscale / set_yscale
         ↓
     "Bagaimana nilai dipetakan secara visual?"
```

### Jangan Tertukar:

| Tool | Fungsi |
|---|---|
| `set_xlim()` | Mengatur batas tampilan sumbu X |
| `set_ylim()` | Mengatur batas tampilan sumbu Y |
| `set_xscale()` | Mengatur skala sumbu X |
| `set_yscale()` | Mengatur skala sumbu Y |

---

## 🎯 Contoh Problem Nyata

Misalkan kamu menganalisis pendapatan perusahaan dari beberapa produk:

```python
produk = ["A", "B", "C", "D"]
pendapatan = [1000, 5000, 50000, 1000000]
```

Jika diplot dengan linear scale, perbedaan antar produk kecil mungkin sulit dianalisis. Dalam situasi seperti ini, log scale bisa membantu melihat perbandingan pola antar orde besaran.

Tapi ingat: log scale bukan sekedar membuat grafik terlihat keren. Pilih scala berdasarkan pertanyaan analisis dan karakteristik data.

# 🏷️ Matplottlib : Legend & Tick Formatting

Sampai sekarang kamu sudah bisa membuat berbagai jenis grafik dan mengatur bagaimana data ditampilkan. Sekarang kita fokus pada satu hal yang sangat penting dalam visualisasi:

> **Apakah orang lain bisa membaca grafikmu dengan mudah?**

Grafik yang benar secara teknis belum tentu komunikatif. Di sinilah legend dan tick formatting berperan.

---

## 1️⃣ Legend — Menjelaskan Data

Legend digunakan ketika sebuah grafik memiliki beberapa data atau kategori yang perlu dibedakan.

Contoh:

```python
import mattplotlib.pyplot as plt


bulan = ["Jan", "Feb", "Mar", "Apr"]
produk_a = [100, 150, 130, 180]
produk_b = [80, 120, 160, 170]

fig, ax = plt.subplots()

ax.plot(bulan, produk_a, label="Produk A")
ax.plot(bulan, produk_b, label="Produk B")

ax.legend()

plt.show()
```

Perhatikan dua hal:

```python
label='Produk A'
```

memberikan identitas pada data.

Sedangkan:

```python
ax.legend()
```

meminta Matplotlib untuk menampilkan kumpulan label tersebut sebagai legend:

### 🧠 Mental model

```text
Data
 ↓
label
 ↓
Legend mengenalkan data kepada pembaca
```

Tanpa `label`, legend tidak tahu nama yang harus ditampilkan.

---

## 2️⃣ Mengatur Posisi Legend

Kamu bisa mengatur posisi legend:

```python
ax.legend(loc='upper right')
```

Beberapa posisi umum:

```text
upper right
upper left
lower right
lower left
best
```

Contoh:

```python
ax.legend(loc='lower right')
```

💡 `best` membiarkan Matplotlib memilih posisi yang dianggap paling sesuai agar tidak terlalu menutupi data.

Tetapi prinsip utamanya bukan menghafal semua posisi.

> **Letakkan legend di tempat yang tidak mengganggu informasi utama.**

---

## 3️⃣ Tick — Angka dan Label pada Sumbu

Perhatikan grafik:

```text
      │
  100 ┤
   80 ┤
   60 ┤
   40 ┤
   20 ┤
      └────────────
        A  B  C
```

Angka `20`, `40`, `60` dan seterusnya adalah *tick labels*.

Kita bisa mengatur tick secara manual.

Contoh:

```python
ax.set_yticks([0, 50, 100, 150])
```

Sekarang sumbu Y akan memiliki tick pada posisi tersebut.

Untuk sumbu X:

```python
ax.set_xticks([0, 1, 2, 3])
```

---

## 4️⃣ Tick Labels

Kita juga bisa menentukan teks yang ditampilkan:

```python
ax.set_xticks(
    [0, 1, 2],
    ['Rendah', 'Sedang', 'Tinggi']
)
```

Jadi ada dua konsep:

```text
Tick position → posisi pada sumbu
Tick label    → teks yang ditampilkan
```

Jangan tertukar. ⚠️

---

## 5️⃣ Memformat Angka agar Lebih Mudah Dibaca

Bayangkan data pendapatan:

```text
1000
25000
1500000
```

Secara teknis benar, tetapi kurang nyaman dibaca.

Kita bisa menggunakan formatter:

```python
from matplotlib.ticker import FuncFormatter
```

Contoh sederhana:

```python
from matplotlib.ticker import FuncFormatter as ticker

def format_rupiah(x, pos):
    return f'Rp{x:, .0f}'

ax.yaxis.set_major_formatter(ticker(format_rupiah))
```

Hasilnya kira-kira:

```text
Rp1,000
Rp25,000
Rp1,500,000
```

🎯 Dalam data visualization, formatting bukan sekadar dekorasi.

> **Formatting mengurangi usaha mental pembaca untuk memahami angka.**

---

## 🧠 MENTAL MODEL UTAMA

```text
📊 GRAPH
│
├── Legend
│     └── "Data mana yang sedang saya lihat?"
│
└── Ticks & Formatting
      └── "Bagaimana saya membaca nilai pada grafik?"
```

Dengan kata lain:

> **Legend membantu pembaca mengenali data.**
>
> **Ticks membantu pembaca membaca skala.**
>
> **Formatting membuat informasi lebih mudah dipahami.**

# 💾 Matplotlib : savefig() — Menyimpan Visualisasi

Sampai sekarang, grafikmu hanya muncul di layar melalui:

```python
plt.show()
```

Tapi dalam workflow Data Science nyata, grafik sering perlu disimpan untuk:

- 📄 laporan analisis
- 📊 presentasi
- 💼 portfolio
- 📓 dokumentasi project
- 🌐 aplikasi atau dashboard

Di sinilah `savefig()` digunakan.

---

## 1️⃣ Menyimpan Grafik dengan `savefig()`

Contoh paling sederhana:

```python
plt.savefig('grafik_penjualan.png')
```

Jika meenggunakan pendekatan `Figure & Axes`, lebih konsisten menggunakan:

```python
fig.savefig('grafik_penjualan.png')
```

Misalnya:

```python
import matplotlib.pyplot as plt

bulan = ["Jan", "Feb", "Mar", "Apr"]
penjualan = [100, 150, 130, 200]

fig, ax = plt.subplots()

ax.plot(bulan, penjualan)
ax.set_title("Penjualan Bulanan")

fig.savefig("grafik_penjualan.png")
plt.show()
```

### 🧠 Mental model:

> `show()` → menampilkan grafik kepada kita.
> `savefig()` → menyimpan grafik untuk digunakan di luar sesi Python.

---

## 2️⃣ Memilih Format File

Format file biasanya ditentukan dari ekstensi nama file:

```python
fig.savefig("grafik.png")
fig.savefig("grafik.jpg")
fig.savefig("grafik.pdf")
```

Secara sederhana:

| Format | Umumnya cocok untuk |
|---|---|
| `.png` | Notebook, laporan, portfolio |
| `.jpg` | Gambar umum dengan ukuran file yang bisa lebih kecil |
| `.pdf` | Dokumen atau kebutuhan cetak |
| `.svg` | Grafik berbasis vektor yang perlu diperbesar |

Untuk workflow Data Science sehari-hari, **PNG adalah pilihan yang sangat umum**.

---

## 3️⃣ DPI — Resolusi Gambar

DPI (_dot per inch_) memengaruhi tingkat detail saat gambar diekspor.

```python
fig.savefig("grafik.png", dpi=300)
```

Semakin tinggi DPI, umumnya semakin tinggi detail dan ukuran file.

Contoh peenggunaan praktis:

```python
fig.savefig(
    "grafik_penjualan.png",
    dpi=300
)
```

📌 Jangan otomatis memakai DPI tertinggi. Pilih berdasarkan kebutuhan output.

---

## 4️⃣ `bbox_inches="tight"`

Kadang ketika menyimpan grafik, ada bagian seperti label atau title yang terlalu dekat dengan batas gambar.

Kita bisa menggunakan:

```python
fig.savefig(
    'grafik_penjualan.png',
    bbox_inches='tight'
)
```

Ini membantu area penyimpanan mengikuti konten grafik dengan lebih rapi.

Kamu juga bisa menggabungkannya:

```python
fig.savefig(
    'grafik_penjualan.png',
    dpi=300,
    bbox_inches='tight'
)
```

---

## ⚠️ 5️⃣ Urutan `savefig()` dan `show()`

Urutan workflow yang aman, biasanya:

```text
Buat Grafik
    ↓
Atur Grafik
    ↓
savefig()
    ↓
show()
```

Contoh:

```python
fig.savefig('hasil.png', dpi=300, bbox_inches='tight')
plt.show()
```

Dengan urutan ini, kamu menyimpan figure sebelum menutup atau membersihkan tampilan figure.

---

## 🧠 MENTAL MODEL UTAMA

```text
📊 Figure selesai dibuat
        │
        ├── 👀 show()
        │       → melihat hasil sekarang
        │
        └── 💾 savefig()
                → menyimpan hasil untuk digunakan nanti
```

Jadi `savefig()` bukan bagian dari proses analisis data secara langsung, tetapi bagian penting dari workflow menyampaikan hasil analisis.

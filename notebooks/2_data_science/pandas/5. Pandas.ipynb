# Pandas — 📈 Statistics

## ⭐ Materi 1: `mean()`

### 📚 1. Apa itu `mean()`?

`mean()` digunakan untuk menghitung **nilai rata-rata** dari data numerik.

Secara matematis:

```text
mean = jumlah seluruh nilai / jumlah data
```

Contoh:

```python
import pandas as pd

df = pd.DataFrame({
    'Nama': ['Andi', 'Budi', 'Citra', 'Doni', 'Eko'],
    'Nilai': [80, 75, 90, 85, 95]
})
```

Menghitung rata-rata:

```python
df['Nilai'].mean()
```

Output:

```text
85.0
```

Karena:

```text
(80 + 75 + 90 + 85 + 95) / 5
= 425 / 5
= 85
```

---

### 📌 2. `mean()` pada DataFrame

Kamu juga bisa menjalankan:

```python
df.mean(numeric_only=True)
```

Ini akan menghasilkan rata-rata untuk kolom numerik.

Contoh Output:

```text
Nilai    85.0
```

`Nama` tidak dihitung karena merupakan data non-numerik.

---

### 📌 3. `mean()` dengan Missing Value

Pandas secara default mengabaikan `NaN` ketika menghitung mean.

```python
df = pd.DataFrame({
    'Nilai': [80, 90, None, 70, 100]
})

df['Nilai'].mean()
```

Yang dihitung:

```text
80 + 90 + 70 + 100
-------------------
        4
```

Output:

```text
85.0
```

Jadi `None/NaN` tidak dianggap sebagai angka `0`.

---

## ⭐ Materi 2: `sum()`

### 📚 1. Apa itu `sum()`?

`sum()` digunakan untuk menghitung nilai jumlah seluruh nilai dalam Series atau kolom.

Contoh:

```python
import pandas as pd

df = pd.DataFrame({
    "Nama": ["Andi", "Budi", "Citra", "Doni"],
    "Nilai": [80, 70, 90, 100]
})

print(df["Nilai"].sum())
```

Output:

```text
340
```

Karena:

```text
80 + 70 + 90 + 100 = 340
```

---

### 📌 2. `sum()` pada DataFrame

Kamu juga bisa menggunakan:

```python
df.sum(numeric_only=True)
```

Hasil:

```text
Nilai    340
```

Berbeda dengan:

```python
df['Nilai'].sum()
```

yang langsung menghasilkan:

```text
340
```

Jadi, kalau hanya ingin total satu kolom, **lebih spesifik menggunakan kolom tersebut**.

---

### 📌 3. `sum()` dengan Missing Value

Pandas secara default mengabaikan `NaN`.

```python
df = pd.DataFrame({
    'Nilai': [80, 90, None, 70]
}])

print(df['Nilai'].sum())
```

Output:

```text
240.0
```

Karena:

```text
80 + 90 + 70 = 240
```

`None` tidak dihitung sebagai `0`secara eksplisit; nilai missing diabaikan dalam operasi agregasi default.


Karena ada 4 nilai yang tersedia.

---

Untuk sekarang fokus pada `count()` :

```python

```

menghitung nilai yang tidak kosong.

Contoh:

```python
df = pd.DataFrame({
```

```text

```

Bukan `5`, karena `None` tidak dihitung.

Ini perbedaan penting:

```text

90
None
70
100

count() → 4
```

```python
df.count()
```

## ⭐ Materi 6: `median()`

### 📚 1. Apa itu `median()`?

`median()` digunakan untuk mencari nilai tengah dari sekumpulan data setelah data tersebut diurutkan.

Contoh:

```python
import pandas as pd

df = pd.DataFrame({
    "Nilai": [70, 80, 90, 100, 60]
})

print(df["Nilai"].median())
```

Data jika diurutkan :

```text
60, 70, 80, 90, 100
```

Nilai tengahnya adalah:

```text
80
```

Hasil:

```text
80.0
```

---

### 📌 2. `median()` dengan jumlah data genap

Kalau jumlah datanya genap, ada dua nilai tengah.

Contoh:

```python
df = pd.DataFrame({
    "Nilai": [60, 70, 80, 90]
})

print(df["Nilai"].median())
```

Data:

```text
60, 70, 80, 90
```

Dua nilai tengah:

```text
70 dan 80
```

Median:

```text
(70 + 80) / 2 = 75
```

Hasil:

```text
75.0
```

---

### 📌 3. `median()` dan Missing Value

Seperti method statistik sebelumnya, `median()` secara default mengabaikan missing value.

```python
df = pd.DataFrame({

})

print(df["Nilai"].median())
```

Data yang digunakan:

```text
60, 80, 90, 100
```

Median:

```text
(80 + 90) / 2 = 85
```

Output:

```text
85.0
```

## ⭐ Materi 7: `mode()`

### 📚 1. Apa itu `mode()`?

`mode()` digunakan untuk mencari **nilai yang paling sering muncul** dalam data.

Contoh:

```python
import pandas as pd

df = pd.DataFrame({
    "Nilai": [70, 80, 80, 90, 100]
})

print(df["Nilai"].mode())
```

Output:

```text
0    80
dtype: int64
```

Karena `80` muncul 2 kali, sedangkan nilai lainnya hanya sekali.

---

### 📌 2. `mode()` bisa menghasilkan lebih dari satu nilai

Ini penting.

Misalnya:

```python
df = pd.DataFrame({
    "Nilai": [70, 70, 80, 80, 90]
})

print(df["Nilai"].mode())
```

Output:

```text
0    70
1    80
dtype: int64
```

Kenapa ada dua?

Karena:

```text
70 → 2 kali
80 → 2 kali
90 → 1 kali
```

Jadi `70` dan `80` sama-sama merupakan mode.

> `mode()` dapat mengembalikan lebih dari satu nilai jika terdapat beberapa nilai dengan frekuensi tertinggi yang sama.

---

### 📌 3. `mode()` pada DataFrame

Kamu juga bisa menggunakan:

```python
df.mode(numeric_only=True)
```

Tetapi untuk sekarang, kita lebih fokus pada:

```python
df['Nilai'].mode()
```

karena kita ingin mengambil mode dari kolom tertentu .


---

### 📌 4. `mode()` dan Missing Value

Secara default, missing value tidak digunakan sebgai kandidat mode.

Contoh:

```python
df = pd.DataFrame({
    "Nilai": [70, 80, None, 80, 90]
})

print(df["Nilai"].mode())
```

Output;

```text
0    80
dtype: int64
```

`80` tetap menjadi mode karena muncul paling sering di antara nilai yang tersedia.

## ⭐ Materi 8: `std()`

### 📚 1. Apa itu `std()`?

`std()` digunakan untuk mengukur **seberapa tersebar data dari nilai rata-ratanya**.

Sederhananya:
- `std()` kecil → data cenderung dekat dengan rata-rata.
- `std()` besar → data lebih tersebar dari rata-rata.

Contoh:

```python
import pandas as pd

df = pd.DataFrame({
    "Nilai": [80, 80, 80, 80, 80]
})

print(df["Nilai"].std())
```

Output:

```text
0.0
```

Karena semua nilai sama, tidak ada penyebaran data.

---

### 📌 2. Contoh data yang lebih tersebar

```python
df = pd.DataFrame({
    "Nilai": [60, 70, 80, 90, 100]
})

print(df["Nilai"].std())
```

Hasilnya akan lebih besar dari `0`, karena nilai-nilainya memiliki jarak dari rata-rata.

Untuk sekarang **tidak perlu menghitung rumus standart deviation secara manual**. Fokus kita adalah memahami fungsi `std()` dan membaca hasilnya.

---

### 📌 3. `std()` pada DataFrame

Seperti method statistik sebelumnya, kamu bisa menggunakan:

```python
df['Nilai'].std()
```

untuk satu kolom.

Atau:

```python
df.std(numeric_only=True)
```

untuk seluruh kolom numerik.

Contoh:

```python
df = pd.DataFrame({
    "Nilai": [60, 70, 80, 90, 100],
    "Umur": [18, 19, 20, 21, 22]
})

print(df.std(numeric_only=True))
```

---

### 📌 4. `std()` dan Missing Value

Pandas secar default mengabaikan missing value.

```python
df = pd.DataFrame({
    "Nilai": [60, None, 80, 90, 100]
})

print(df["Nilai"].std())
```

`None` tidak digunakan sebagai nilai dalam perhitungan standard deviation.

## ⭐ Materi 9: `value_counts()`

### 📚 1. Apa itu `value_counts()`?

`value_counts()` digunakan untuk menghitung **berapa kali setiap nilai muncul** dalam sebuah Series/Kolom.

Contoh:

```python
import pandas as pd

df = pd.DataFrame({
    "Nilai": [80, 75, 80, 90, 75, 80]
})

print(df["Nilai"].value_counts())
```

Output:

```text
80    3
75    2
90    1
Name: count, dtype: int64
```

Artinya:

```text
80 → muncul 3 kali
75 → muncul 2 kali
90 → muncul 1 kali
```

Secara default, hasilnya diurutkan dari **frekuensi terbesar ke terkecil (descenidng)**.

---

### 📌 2. `value_counts()` berbeda dengan `mode()`

Ini penting karena keduanya sudah kamu pelajari.

```text
mode()
   ↓
mencari nilai yang memiliki frekuensi tertinggi
```

Sedangkan:

```text
value_counts()
   ↓
menghitung frekuensi setiap nilai
```

Contoh:

```text
Data: 80, 75, 80, 90, 75, 80

mode()
→ 80

value_counts()
→ 80    3
  75    2
  90    1
```

Jadi `mode()` mengambil **nilai paling sering**, sedangkan `value_counts()` memberikan **distribusi frekuensinya**.

---

### 📌 3. `value_counts()` dan Missing Value

Secara default, missing value tidak dihitung.

```python
df = pd.DataFrame({
    "Nilai": [80, 75, None, 80, 90, None]
})

print(df["Nilai"].value_counts())
```

Nilai `None` tidak muncul dalam list default.

Jika ingin memasukkan missing value dalam hitungan, ada parameter:

```python
df['Nilai'].value_counts(dropna=False)
```

Contoh Output:

```text
NaN     1
```

# 📊 Pandas → GroupBy

## ⭐ Materi 1: `groupby()`

### 1. Apa Fungsi `groupby()`?

`groupby()` digunakan untuk mengelompokkan baris berdasarkan suatu kolom/kategiri.

Misalnya:

```python
import pandas as pd

df = pd.DataFrame({
    "Nama": ["Andi", "Budi", "Citra", "Doni", "Eko"],
    "Kota": ["Jakarta", "Bandung", "Jakarta", "Bandung", "Jakarta"],
    "Nilai": [80, 75, 90, 85, 95]
})

print(df)
```

Data:

```text
    Nama      Kota  Nilai
0   Andi   Jakarta     80
1   Budi   Bandung     75
2  Citra   Jakarta     90
3   Doni   Bandung     85
4    Eko   Jakarta     95
```

Sekarang:

```python
grouped = df.groupby('Kota')

print(grouped)
```

`groupby('Kota')` memberi tahu Pandas:
> "Kelompokkan data berdasarkan nilai yang ada di kolom `Kota`."

Kelompoknya menjadi:

```text
Jakarta
├── Andi
├── Citra
└── Eko

Bandung
├── Budi
└── Doni
```

---

### 2. `groupby()` belum menghtiung apa pun

Ini penting.

```python
df.groupby('Kota')
```

belum berarti menghitung total, rata-rata, atau jumlah.

Ia baru membuat kelompok.

Untuk mendapatkan informasi dari kelompok tersebut, nanti kita kombinasikan dengan aggregation seperti:

```python
df.groupby('Kota')['Nilai'].mean()
```

Tetapi aggregation belum kita pelajari sekarang, Untuk saat ini fokusnya hanya memahami proses grouping.

---

### 3. Melihat kelompoknya

Kita bisa melakukan:

```python
for kota, data in df.groupby('Kota'):
    print(kota)
    print(data)
```

Konsepnya:

```text
groupby("Kota")
      │
      ├── Bandung → data Bandung
      │
      └── Jakarta → data Jakarta
```

Perhatikan bahwa `groupby()` sangat berguna ketika dataset memiliki kategori yang berulang.

Contoh kategori:
- Kota
- Departemen
- Produk
- Kategori
- Gender
- Cabang
- Tahun

---

### 🧠 Inti yang harus kamu kuasai

```text
df.groupby("Kolom")
       │
       ▼
Mengelompokkan data berdasarkan nilai
dalam kolom tersebut.
```

Bedakan:

```python
df['Kota']
```

→ mengambil kolom.

Sedangkan:

```python
df.groupby('Kota')
```

→ **mengelompokkan data berdasarkan kolom tersebut**.

---

## ⭐ Materi 2: Aggregation

### 1. Apa itu Aggregation?

Setelah daata dikelompokkan dengan:

```python
grouped = df.groupby('Kategori')
```

kita bisa melakukan perhitungan pada **setiap kelompok**.

Contoh:

```python
df.groupby('Kategori')['Harga'].sum()
```

Artinya:

```text
groupby("Kategori")
        ↓
kelompokkan berdasarkan Kategori
        ↓
ambil kolom Harga
        ↓
hitung total Harga setiap kelompok
```

Hasilnya kira-kira:

```text
Kategori
Aksesoris      950
Elektronik    15500
```

---

### 2. `sum()`

Menghitung total setiap kelompok.

```python
df.groupby('Kategori')['Harga'].sum()
```

Dengan data kita:

```text
Aksesoris    → 200 + 500 + 250 = 950
Elektronik   → 8000 + 7500 = 15500
```

---

### 3. `mean()`

Menghitung rata-rata setiap kelompok.

```python
df.groupby('Kategori')['Harga'].mean()
```

Hasilnya:

```text
Aksesoris    → 316.67
Elektronik   → 7750.00
```

---

### 4. `count()`

Menghitung jumlah data pada setiap kelompok.

```python
df.groupby('Kategori')['Harga'].count()
```

Hasil:

```text
Aksesoris     3
Elektronik    2
```

Jadi pola pentingnya:

```text
df.groupby("kategori")["kolom"]
                       │
                       ▼
                 aggregation
                       │
             ┌─────────┼─────────┐
             ▼         ▼         ▼
           sum()     mean()    count()
```

**⚠️ Perhatikan**

`groupby()` mengelompokkan.

Aggregation menghitung atau merangkum kelompok tersebut.

Jangan mencampurkan kedua konsep ini.

---

## ⭐ Materi 3: Multiple Aggregation

### 1. Apa itu Multiple Aggregation

Sebelumnya kita hanya menggunakan satu aggregation:

```python
df.groupby('Kategori')['Harga'].sum()
```

Sekarang kita bisa meminta beberapa perhitungan sekaligus.

Gunakan `.agg()`:

```python
df.groupby('Kategori')['Harga'].agg(['sum', 'mean', 'max', 'min'])
```

Konsepnya:

```text
groupby("Kategori")
        ↓
     "Harga"
        ↓
       agg()
        ↓
 ┌──────┼──────┬──────┐
 sum   mean    max    min
```

Hasilnya akan berbentuk *DataFrame*:

```text
            sum        mean   max   min
Kategori
Aksesoris    950  316.666667   500   200
Elektronik 15500 7750.000000  8000  7500
```

---

### 2. Kenapa memakai `.agg()`?

Daripada mengetik:

```python
df.groupby("Kategori")["Harga"].sum()
df.groupby("Kategori")["Harga"].mean()
df.groupby("Kategori")["Harga"].max()
df.groupby("Kategori")["Harga"].min()
```

kita bisa menggabungkannya:

```python
df.groupby("Kategori")["Harga"].agg(
    ["sum", "mean", "max", "min"]
)
```

Lebih ringkas dan hasilnya langsung menjadi satu tabel.

---

### 3. Aggregation dengan nama sendiri

`.agg()` juga memungkinkan kita memberi nama yang lebih jelas:

```python
df.groupby('Kategori')['Harga'].agg(
    total= 'sum',
    rata_rata= 'mean',
    tertinggi= 'max',
    terendah= 'min'
)
```

Output:

```text
            total  rata_rata  tertinggi  terendah
Kategori
Aksesoris     950     316.67        500       200
Elektronik  15500    7750.00       8000      7500
```

Ini berguna ketika hasil aggregation akan digunakan dalam laporan atau project karena nama kolomnya lebih deskriptif.

---

### 🧠 Pola yang perlu diingat

#### Satu Aggregation

```python
df.groupby("Kategori")["Harga"].sum()
```

#### Multi Aggregation

```python
df.groupby("Kategori")["Harga"].agg(
    ["sum", "mean", "max", "min"]
)
```

#### Banyak aggregation dengan nama custom

```python
df.groupby("Kategori")["Harga"].agg(
    total="sum",
    rata_rata="mean",
    tertinggi="max",
    terendah="min"
)
```

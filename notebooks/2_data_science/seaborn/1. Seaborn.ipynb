# 🎨 SEABORN — Statistical Visualization

## 📊 Konsep Dasar

Seaborn adalah library visualisasi statistik yang bekerja erat dengan Matplotlib, tetapi menyediakan antarmuka tingkat lebih tinggi untuk membuat visualisasi statistik. Dokumentasi resminosisikan Seaborn sebagai high-leve API untuk statistical graphics. [Seaborn](https://seaborn.pydata.org/tutorial.html)

Mental model-nya:

```text
Matplotlib
    ↓
Kontrol visualisasi yang detail
    ↓
Seaborn
    ↓
Visualisasi statistik yang lebih praktis
```

Contoh sederhana:

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

Dataset:

```python
data = {
    "Hari": ["Senin", "Selasa", "Rabu", "Kamis", "Jumat"],
    "Penjualan": [120, 150, 135, 180, 170],
    "Pelanggan": [30, 35, 32, 42, 40]
}
```

Kita ubah menjadi DataFrame:

```python
import pandas as pd

df = pd.DataFrame(data)
```

Kemudian:

```python
sns.barplot(data=df, x='Hari', y='Penjualan')

plt.show()
```

Perhatikan bahwa kita tidak sedang belaar syntax baru semata.

Yang ingin aku bangun adalah kemampuan:

> **memilih visualisasi berdasarkan jenis data dan pertanyaan analisis**

## 🔍 Seaborn vs Matplotlib

Keduanya bukan kompetitor yang harus dipilih salah satu.

```text
Pandas
   ↓
Menyiapkan data
   ↓
Seaborn
   ↓
Statistical visualization
   ↓
Matplotlib
   ↓
Fine-tuning / kontrol detail
```

Seaborn sendiri menyediakan fungsi seperti `scatterplot()`, `lineplot()`, `histplot()`, `boxplot()`, `barplot()`, `heatmap()`, dan `pairplot()`, serta fungsi tingkat figure seperti `relplot()`, `displot()`, dan `catplot()`. [Seaborn](https://seaborn.pydata.org/api.html)

# Bab 1: Virtual Environment

## 1.1 Kenapa Perlu Virtual Environment?

Bayangkan kamu mengerjakan dua proyek: Proyek A butuh library versi lama, Proyek B butuh versi terbaru dari library yang **sama**. Kalau semua library diinstall langsung ke sistem (global), keduanya akan bentrok.

**Virtual environment** (sering disingkat *venv*) adalah folder terisolasi yang punya instalasi Python dan library-nya sendiri, terpisah dari sistem utama dan dari proyek lain. Setiap proyek idealnya punya virtual environment sendiri-sendiri.

## 1.2 Membuat & Mengaktifkan Virtual Environment

Perintah-perintah berikut dijalankan di **terminal/command prompt**, bukan di dalam notebook Python:

```bash
# Membuat virtual environment bernama 'venv' di folder proyek
python -m venv venv

# Mengaktifkan di Windows
venv\Scripts\activate

# Mengaktifkan di macOS/Linux
source venv/bin/activate

# Menonaktifkan (dari environment mana saja)
deactivate
```

Saat venv aktif, biasanya nama environment akan muncul di awal baris terminal, misalnya `(venv) C:\proyek-python>` — ini penanda bahwa perintah `python` dan `pip` yang kamu jalankan sekarang memakai instalasi terisolasi milik venv tersebut, bukan instalasi global.

````python
import sys, platform

print("Versi Python yang dipakai notebook ini:", sys.version.split()[0])
print("Sistem operasi:", platform.system(), platform.release())
````

⚠️ **Common Mistakes**
1. Menginstall package langsung ke Python global (tanpa venv aktif) — berisiko bentrok antar proyek dalam jangka panjang
2. Lupa mengaktifkan venv sebelum menjalankan `pip install` atau menjalankan file `.py`
3. Meng-commit folder `venv` ke Git — ukurannya besar dan tidak perlu diikutsertakan, sebaiknya masukkan ke `.gitignore`

💡 **Tips & Best Practice**
- Buat virtual environment baru untuk **setiap** proyek Python yang berbeda
- Tambahkan baris `venv/` ke file `.gitignore` supaya tidak ikut ter-*commit*
- Pastikan nama environment muncul di terminal sebelum menjalankan perintah `pip install`, sebagai tanda venv sudah aktif

🎯 **Latihan Bab 1**
1. Coba praktikkan langsung di terminal komputermu: buat folder proyek baru, buat virtual environment di dalamnya, lalu aktifkan.
2. Setelah venv aktif, jalankan `python --version` dan `pip --version` — bandingkan hasilnya dengan saat venv belum aktif.

# Bab 2: pip (Package Manager)

## 2.1 Instalasi Package

`pip` adalah *package manager* bawaan Python untuk menginstall library dari [PyPI](https://pypi.org) (Python Package Index).

```bash
# Install package terbaru
pip install nama_package

# Install versi tertentu
pip install nama_package==1.2.3

# Upgrade ke versi terbaru
pip install --upgrade nama_package

# Uninstall package
pip uninstall nama_package

# Lihat semua package yang terinstall
pip list
```

## 2.2 `requirements.txt`

File `requirements.txt` mencatat semua dependency proyek beserta versinya, supaya orang lain (atau kamu sendiri di komputer lain) bisa menginstall persis dependency yang sama.

```bash
# Membuat requirements.txt dari environment aktif saat ini
pip freeze > requirements.txt

# Menginstall semua dependency dari requirements.txt
pip install -r requirements.txt
```

````python
import importlib.metadata as metadata

# Menampilkan beberapa package yang sudah terinstall di environment ini (contoh 5 pertama)
jumlah = 0
for dist in metadata.distributions():
    nama = dist.metadata.get("Name") if dist.metadata else None
    if nama:
        print(f"{nama} == {dist.version}")
        jumlah += 1
    if jumlah >= 5:
        break
````

⚠️ **Common Mistakes**
1. Menginstall package tanpa venv aktif — mengotori environment Python global
2. Lupa memperbarui `requirements.txt` setelah menginstall package baru
3. Menulis `pip install nama_package` tanpa versi spesifik di `requirements.txt` — bisa membuat versi berbeda terinstal di komputer lain, berpotensi menyebabkan bug yang tidak konsisten

💡 **Tips & Best Practice**
- Selalu sertakan `requirements.txt` bersama kode proyek, terutama saat berbagi/upload ke Git
- Jalankan `pip freeze > requirements.txt` setiap kali selesai menambah dependency baru
- Untuk proyek yang dibagikan ke publik, pertimbangkan mengunci versi (`==`) supaya environment yang tercipta ulang benar-benar identik

🎯 **Latihan Bab 2**

2. Buat `requirements.txt` dari environment yang sudah kamu buat di Latihan Bab 1, lalu buka isinya dengan text editor.

# Bab 3: Best Practice

## 3.1 PEP 8 — Panduan Gaya Penulisan Kode Python

PEP 8 adalah panduan gaya penulisan kode resmi Python. Beberapa aturan intinya:
- Nama variabel & function: `snake_case` (huruf kecil + underscore)
- Nama class: `PascalCase` (setiap awal kata huruf besar)
- Nama konstanta: `UPPER_CASE` (huruf besar semua)
- Indentasi: 4 spasi per level
- Panjang baris idealnya tidak lebih dari 79–99 karakter

````python
# Kurang mengikuti PEP 8
x = 100
def calc(a,b):
    return a+b

# Lebih mengikuti PEP 8
HARGA_SATUAN = 100

def hitung_total(jumlah, harga):
    return jumlah * harga

print(calc(2, 3))
print(hitung_total(2, HARGA_SATUAN))
````

## 3.2 DRY (*Don't Repeat Yourself*) & Pemecahan Function

Kalau ada logika yang sama ditulis berulang kali, pertimbangkan memecahnya jadi function terpisah.

````python
# Kurang baik: logika validasi diulang berkali-kali
nama1 = "Xenno"
if len(nama1) > 0:
    print(f"Halo, {nama1}")

nama2 = ""
if len(nama2) > 0:
    print(f"Halo, {nama2}")
else:
    print("Nama tidak boleh kosong")

# Lebih baik: logika dikumpulkan jadi satu function
def sapa(nama):
    if len(nama) > 0:
        print(f"Halo, {nama}")
    else:
        print("Nama tidak boleh kosong")

sapa("Xenno")
sapa("")
````

## 3.3 Komentar & Docstring yang Baik

Komentar yang baik menjelaskan **kenapa** sesuatu dilakukan, bukan sekadar mengulang **apa** yang sudah jelas terbaca dari kodenya sendiri.

````python
# Kurang berguna: hanya mengulang apa yang sudah jelas dari kode
x = 5
x = x + 1  # tambahkan 1 ke x

# Lebih berguna: menjelaskan KENAPA
percobaan_ke = 0
percobaan_ke += 1  # hitung ulang percobaan untuk mencegah infinite retry

print(x, percobaan_ke)
````

## 3.4 Type Hints (Pengenalan)

Sejak Python 3.5, kamu bisa menambahkan **petunjuk tipe data** (*type hints*) pada parameter, return value, dan variabel. Ini bersifat **opsional** dan **tidak dipaksakan** Python saat program berjalan — murni membantu dokumentasi dan keterbacaan kode (terutama untuk tim/proyek besar), serta bisa diperiksa dengan tool tambahan seperti `mypy`.

````python
def hitung_luas(panjang: float, lebar: float) -> float:
    return panjang * lebar

nama: str = "Xenno"
umur: int = 20

print(hitung_luas(5.0, 3.0))
print(nama, umur)

# Type hint TIDAK dipaksakan Python saat runtime -> pemanggilan "salah" ini tetap berjalan tanpa error
hasil_aneh = hitung_luas(3, "x")   # seharusnya float, tapi Python tidak melarang
print(hasil_aneh)   # "xxx" -> int * str mengulang string, BUKAN luas yang masuk akal
````

## 3.5 Hindari *Magic Number*

*Magic number* adalah angka mentah yang muncul langsung di tengah kode tanpa penjelasan, sehingga maknanya tidak jelas.

````python
# Kurang baik: angka 0.11 muncul begitu saja, apa maksudnya?
def hitung_total_belanja(harga):
    return harga + (harga * 0.11)

# Lebih baik: diberi nama yang jelas
PPN = 0.11

def hitung_total_belanja_v2(harga):
    return harga + (harga * PPN)

print(hitung_total_belanja(100000))
print(hitung_total_belanja_v2(100000))
````

⚠️ **Common Mistakes**
1. Penamaan yang tidak konsisten (campur `camelCase` dan `snake_case` dalam satu proyek)
2. Function yang terlalu panjang dan melakukan banyak hal sekaligus — sebaiknya dipecah jadi beberapa function kecil dengan tugas jelas
3. Angka atau string "ajaib" yang muncul berulang tanpa nama/konstanta yang jelas

💡 **Tips & Best Practice**
- Konsisten adalah kunci — pilih satu gaya penamaan dan pertahankan di seluruh proyek
- Tool seperti `black` (formatter otomatis) dan `flake8`/`pylint` (linter) bisa membantu menjaga kode tetap rapi tanpa perlu mengecek manual satu-satu
- Kode yang baik itu ditulis untuk dibaca manusia, bukan hanya supaya bisa dijalankan komputer

# Bab 4: Final Project — Aplikasi To-Do List

**Tujuan**: menggabungkan SELURUH materi dari Part 1 sampai Part 5 dalam satu aplikasi utuh:
- Variabel, tipe data, operator, f-string (**Part 1**)
- Percabangan & perulangan (**Part 2**)
- Function (**Part 3**)
- List & string method (**Part 4**)
- Error handling, file I/O, dan OOP (**Part 5**)

**Deskripsi**: aplikasi To-Do List sederhana yang bisa menambah tugas, menandai selesai, menghapus tugas, menyimpan ke file, dan memuat ulang dari file.

````python
class Tugas:
    def __init__(self, judul, prioritas="Sedang"):
        self.judul = judul
        self.prioritas = prioritas
        self.selesai = False

    def tandai_selesai(self):
        self.selesai = True

    def __str__(self):
        status = "[x]" if self.selesai else "[ ]"
        return f"{status} {self.judul} (Prioritas: {self.prioritas})"


class DaftarTugas:
    def __init__(self):
        self.daftar = []

    def tambah(self, judul, prioritas="Sedang"):
        try:
            if not judul.strip():
                raise ValueError("Judul tugas tidak boleh kosong")
            tugas = Tugas(judul.strip(), prioritas)
            self.daftar.append(tugas)
            print(f"Ditambahkan: {tugas}")
        except ValueError as e:
            print(f"Gagal menambahkan tugas: {e}")

    def tandai_selesai(self, index):
        try:
            self.daftar[index].tandai_selesai()
            print(f"Ditandai selesai: {self.daftar[index]}")
        except IndexError:
            print(f"Tugas nomor {index + 1} tidak ditemukan")

    def hapus(self, index):
        try:
            tugas_dihapus = self.daftar.pop(index)
            print(f"Dihapus: {tugas_dihapus}")
        except IndexError:
            print(f"Tugas nomor {index + 1} tidak ditemukan")

    def tampilkan(self):
        if not self.daftar:
            print("Belum ada tugas.")
            return
        for i, tugas in enumerate(self.daftar, start=1):
            print(f"{i}. {tugas}")

    def simpan_ke_file(self, nama_file="tugas.txt"):
        with open(nama_file, "w") as file:
            for tugas in self.daftar:
                status = "selesai" if tugas.selesai else "belum"
                file.write(f"{tugas.judul}|{tugas.prioritas}|{status}\n")
        print(f"Tersimpan ke '{nama_file}'")

    def muat_dari_file(self, nama_file="tugas.txt"):
        self.daftar = []
        try:
            with open(nama_file, "r") as file:
                for baris in file:
                    judul, prioritas, status = baris.strip().split("|")
                    tugas = Tugas(judul, prioritas)
                    if status == "selesai":
                        tugas.tandai_selesai()
                    self.daftar.append(tugas)
        except FileNotFoundError:
            print(f"File '{nama_file}' belum ada, mulai dengan daftar kosong.")
````

**Simulasi Penggunaan**

Bagian di bawah ini mensimulasikan satu sesi penggunaan aplikasi — pengganti input interaktif berulang lewat `input()`, supaya bisa langsung dijalankan dan dites di notebook.

````python
todo = DaftarTugas()

todo.tambah("Belajar Python Part 6", "Tinggi")
todo.tambah("Kerjakan tugas Matematika Diskrit", "Tinggi")
todo.tambah("Nonton film", "Rendah")
todo.tambah("", "Sedang")   # sengaja kosong -> menguji error handling

print("\n=== Daftar Tugas ===")
todo.tampilkan()

todo.tandai_selesai(0)   # tandai tugas nomor 1 selesai
todo.hapus(2)             # hapus tugas nomor 3 ("Nonton film")

print("\n=== Setelah Update ===")
todo.tampilkan()

print()
todo.simpan_ke_file()

print("\n=== Memuat Ulang dari File (Sesi Baru) ===")
todo_baru = DaftarTugas()
todo_baru.muat_dari_file()
todo_baru.tampilkan()
````

**Tantangan Lanjutan (opsional)**, untuk melatih lebih jauh:
- Ganti bagian simulasi dengan menu interaktif sungguhan memakai `while True` + `input()` (pilihan: tambah, tandai selesai, hapus, tampilkan, keluar)
- Tambahkan fitur pengurutan tugas berdasarkan prioritas, memakai `sorted()` dan `lambda` (Part 3 & 4)
- Tambahkan validasi supaya `prioritas` hanya boleh diisi `"Tinggi"`, `"Sedang"`, atau `"Rendah"` (gunakan `raise ValueError` kalau di luar itu)
- Tambahkan method untuk menghitung berapa persen tugas yang sudah selesai

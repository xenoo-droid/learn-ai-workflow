```python
import matplotlib.pyplot as plt

# Data sampel yang mencerminkan kondisi studi kasus
data = [58, 60, 62, 65, 68, 70, 72, 75, 78, 80, 82, 120]

# Membuat figure
plt.figure(figsize=(8, 5))

# Membuat boxplot secara horizontal
plt.boxplot(
    data,
    vert=False,
    patch_artist=True,
    boxprops=dict(facecolor='skyblue', color='blue'),
    medianprops=dict(color='red', linewidth=2),
)

# Menambahkan judul dan label
plt.title(
    'Visualisasi Boxplot: Distribusi Data dengan Outlier',
    fontsize=12,
    fontweight='bold',
)
plt.xlabel('Nilai Data', fontsize=10)
plt.yticks([1], ['Dataset'])

# Menambahkan anotas/keterangan pada grafik
plt.axvline(x=70, color='red', linestyle='--', alpha=0.7, label='Median (70)')
plt.annotate(
    'Outlier (120)',
    xy=(120, 1),
    xytext=(110, 1.15),
    arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=6),
    fontweight='bold',
)

plt.grid(axis='x', linestyle=':', alpha=0.6)
plt.legend(loc='upper left')
plt.tight_layout()

# Tampilkan grafik
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Menyiapkan data dummy
x = np.linspace(0, 10, 100)

# 2. Membuat layout 2 baris x 3 kolom
fig, axes = plt.subplots(2, 3, figsize=(12, 8))

# --- BARIS 1 ---
# Subplot [0, 0] - Kiri Atas
axes[0, 0].plot(x, np.sin(x), color='#1f77b4', lw=2)
axes[0, 0].set_title("1. Grafik Sinus")

# Subplot [0, 1] - Tengah Atas
axes[0, 1].plot(x, np.cos(x), color='#ff7f0e', lw=2)
axes[0, 1].set_title("2. Grafik Kosinus")

# Subplot [0, 2] - Kanan Atas
axes[0, 2].plot(x, x**2, color='#2ca02c', lw=2)
axes[0, 2].set_title("3. Grafik Kuadrat ($X^2$)")

# --- BARIS 2 ---
# Subplot [1, 0] - Kiri Bawah
axes[1, 0].plot(x, np.exp(x/2), color='#d62728', lw=2)
axes[1, 0].set_title("4. Eksponensial ($e^{X/2}$)")

# Subplot [1, 1] - Tengah Bawah
x_rand = np.random.randn(50)
y_rand = np.random.randn(50)
axes[1, 1].scatter(x_rand, y_rand, color='#9467bd', alpha=0.7)
axes[1, 1].set_title("5. Scatter Plot Acak")

# Subplot [1, 2] - Kanan Bawah
data_hist = np.random.normal(loc=0, scale=1, size=500)
axes[1, 2].hist(data_hist, bins=20, color='#8c564b', edgecolor='black', alpha=0.7)
axes[1, 2].set_title("6. Histogram Distribusi Normal")

# Merapikan jarak antar grafik
plt.tight_layout()
plt.show()
```

```python
kelas_a = [70, 72, 73, 74, 75, 76, 77, 78, 80]
kelas_b = [55, 65, 68, 72, 75, 78, 82, 90, 120]

fig, ax = plt.subplots(figsize=(7, 4))

box = ax.boxplot(
    [kelas_a, kelas_b],
    tick_labels=["Kelas A", "Kelas B"],
    patch_artist=True
)

for patch in box["boxes"]:
    patch.set_facecolor("lightblue")
    patch.set_alpha(0.7)

for median in box["medians"]:
    median.set_linewidth(2)
    median.set_color('black')

ax.set_title("Perbandingan Distribusi Nilai")
ax.set_xlabel("Nilai")
ax.grid()
plt.show()
```

```python
systems = ['A', 'B', 'C']
avg_response = [120, 100, 110]
variation = [5, 25, 10]

fig, ax = plt.subplots(figsize=(8, 5))

ax.errorbar(
    systems, 
    avg_response,
    yerr = variation,
    fmt='o',
    capsize=5,
    color='blue'
)
ax.set_title('Rata-rata Error Sistem berdasarkan Variasi')
ax.set_xlabel('Sistem')
ax.set_ylabel('Rata-rata Respon (ms)')
ax.grid()

plt.show()
```

```python
bulan = ["Jan", "Feb", "Mar", "Apr", "Mei", "Jun"]

produk_a = [120000, 150000, 130000, 180000, 200000, 220000]
produk_b = [100000, 140000, 160000, 150000, 190000, 210000]
produk_c = [80000, 110000, 140000, 170000, 160000, 230000]

fig, ax = plt.subplots()

ax.plot(
    bulan, produk_a,
    label='Produk A',
    color='red'
)
ax.plot(
    bulan, produk_b,
    label='Produk B',
    color='blue'
)
ax.plot(
    bulan, produk_c,
    label='Produk C',
    color='green'
)

from matplotlib.ticker import FuncFormatter as ticker

def format_rupiah(x, pos):
    return f'Rp{x:,.0f}'

ax.yaxis.set_major_formatter(ticker(format_rupiah))
ax.set_title('Perbandingan Produk A|B|C')
ax.set_xlabel('Bulan')
ax.set_ylabel('Penjualan Bulanan')
ax.grid()
ax.legend(loc='best')
plt.tight_layout()

fig.savefig('Perbandingan_Produk_ABC.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Produk": ["Produk A", "Produk B", "Produk C"],
    "Penjualan": [250, 180, 320],
    "Target": [300, 200, 300],
}

df = pd.DataFrame(data)

fig, ax = plt.subplots(figsize=(4, 4))

ax.bar(
    df['Produk'], df['Target'],
    color='red',
    label='Target'
)
ax.bar(df['Produk'], df['Penjualan'], color='blue', label='Penjualan')
ax.set_title('Target Vs Penjualan')
ax.set_xlabel('Produk')
ax.set_ylabel('Penjualan & Target')
ax.grid()
ax.legend()
plt.tight_layout()

plt.show()
```

```python
bulan = ["Jan", "Feb", "Mar", "Apr", "Mei", "Jun"]
penjualan = [120, 135, 128, 150, 165, 180]
target = [110, 125, 130, 140, 160, 170]
error = [5, 7, 6, 8, 5, 7]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle('performa penjualan terhadap target'.title())
axes[0, 0].bar(bulan, penjualan, color='red')
axes[0, 0].set_title('Grafik Penjualan')
axes[0, 0].set_xlabel('Bulan')
axes[0, 0].set_ylabel('Performa Penjualan')
axes[0, 0].grid()

axes[0, 1].bar(bulan, target, color='blue')
axes[0, 1].set_title('Grafik Target Penjualan')
axes[0, 1].set_xlabel('Bulan')
axes[0, 1].set_ylabel('Target Penjualan')
axes[0, 1].grid()

axes[1, 0].errorbar(bulan, penjualan, yerr=error, fmt='o', capsize=5)
axes[1, 0].set_title('Grafik Potensi Error Performa Penjualan')
axes[1, 0].set_xlabel('Bulan')
axes[1, 0].set_ylabel('Performa Penjualan')
axes[1, 0].grid()

axes[1, 1].plot(bulan, penjualan, linestyle='-', marker='o', color='lightgreen', label='Performa Penjualan')
axes[1, 1].plot(bulan, target, linestyle='--', marker='o', color='green', label='Target')
axes[1, 1].set_title('Perbandingan Performa Penjualan Vs Target dari Waktu ke waktu')
axes[1, 1].set_xlabel('Bulan')
axes[1, 1].set_ylabel('Performa|Target Penjualan')
axes[1, 1].grid()
axes[1, 1].legend(loc='best')

plt.tight_layout()

fig.savefig(
    'grafik_penjualan.png',
    dpi=300,
    bbox_inches='tight'
)
plt.show()
```

```python
import matplotlib.pyplot as plt

minggu = [
    "W1", "W2", "W3", "W4",
    "W5", "W6", "W7", "W8"
]

tiket_masuk = [120, 135, 128, 150, 165, 172, 160, 185]
tiket_selesai = [115, 130, 125, 142, 155, 168, 158, 176]

waktu_penyelesaian = [8.2, 7.8, 8.5, 7.2, 6.9, 6.5, 7.1, 6.2]
variasi_waktu = [1.2, 1.0, 1.4, 0.9, 1.1, 0.8, 1.0, 0.7]

kepuasan_pelanggan = [82, 84, 81, 85, 87, 89, 86, 91]
```

```python
fig, ax = plt.subplots(figsize=(16, 8), nrows = 2, ncols = 2)

# Ax 1
ax[0, 0].plot(
    minggu, 
    tiket_masuk,
    label = 'Tiket Masuk',
    color = 'blue'
)
ax[0, 0].plot(
    minggu, 
    tiket_selesai,
    label = 'Tiket Selesai',
    color = 'red'
)
ax[0, 0].set_title('Tiket Masuk vs Tiket Selesai')
ax[0, 0].set_xlabel('Minggu')
ax[0, 0].set_ylabel('Jumlah Tiket')
ax[0, 0].legend()
ax[0, 0].grid()

# Ax 2
ax[0, 1].plot(
    minggu,
    waktu_penyelesaian,
    color='lightgreen',
)

ax[0, 1].set_title('Waktu Penyelesaian Selama 8 Minggu')
ax[0, 1].set_xlabel('Minggu')
ax[0, 1].set_ylabel('Waktu Penyelesaian')
ax[0, 1].grid()

# Ax 3
ax[1, 0].bar(
    minggu,
    kepuasan_pelanggan,
    color = 'grey'
)
ax[1, 0].set_title('Perubahan Kepuasan Pelanggan Selama 8 Minggu')
ax[1, 0].set_xlabel('Minggu')
ax[1, 0].set_ylabel('Kepuasan Pelanggan')
ax[1, 0].grid()

# Ax 4
ax[1, 1].errorbar(
    minggu, 
    waktu_penyelesaian,
    yerr=variasi_waktu,
    fmt='o',
    ecolor='green',
    capsize = 5
)
ax[1, 1].set_title('Estimasi Error Value Waktu Penyelesaian')
ax[1, 1].set_xlabel('Minggu')
ax[1, 1].set_ylabel('Waktu Penyelesaian')
ax[1, 1].grid()

plt.tight_layout()
plt.savefig('Analisis Challenge.png',dpi=300)
plt.show()
```

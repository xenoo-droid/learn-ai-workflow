```python
#studi case
nama = "Seno"
nilai = [85,90,78,92,88]

def judul():
    print('''==============================
      LAPORAN NILAI
==============================''')

def hitung_total(data):
    total = 0
    for nilai in data:
        total+=nilai
    return total
total = hitung_total(nilai)

def hitung_rata_rata(data):
    total = 0
    n = 0
    for nilai in data:
        total += nilai
        n +=1

    rataan = round(total/n,2)
    return rataan
rata = hitung_rata_rata(nilai)

def nilai_tertinggi(data):
    maks = data[0]
    for i in range(len(data)):
        if data[0]<data[i]:
            maks = data[i]
    return maks
nilai_tertinggi = nilai_tertinggi(nilai)

def nilai_terendah(data):
    mins = data[0]
    for i in range(len(data)):
        if data[0] > data[i]:
            mins = data[i]
    return mins
nilai_terendah = nilai_terendah(nilai)

def tentukan_grade(rata):
    if rata >= 90:
        return "A"
    elif rata >=80:
        return "B"
    elif rata >=70:
        return "C"
    elif rata >=60:
        return "D"
    else:
        return "E"
grade = tentukan_grade(rata)

def tentukan_status(rata):
    if rata>=75:
        return "Lulus"
    else:
        return "Tidak Lulus"
status = tentukan_status(rata)

matkul = 0
for i in range(len(nilai)):
    matkul+=1

def penutup():
    print("==============================")
def tampilkan_hasil(nama, nilai, matkul, nilai_tertinggi, nilai_terendah, total, rata, grade, status):
    judul()
    print(f"Nama               : {nama}")
    print(f"Daftar Nilai       : {nilai}")
    print(f"Jumlah Mata Kuliah : {matkul}")
    print(f"Nilai Tertinggi    : {nilai_tertinggi}")
    print(f"Nilai Terendah     : {nilai_terendah}")
    print(f"Total Nilai        : {total}")
    print(f"Rata-rata          : {rata}")
    print(f"Grade              : {grade}")
    print(f"Status             : {status}")
    penutup()

tampilkan_hasil(nama, nilai, matkul, nilai_tertinggi, nilai_terendah, total, rata, grade, status)
```

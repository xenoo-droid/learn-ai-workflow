# Bab 1: Variabel

## 1.1 Apa itu Variabel?

Variabel adalah **nama yang dipakai untuk menyimpan data** di memori komputer agar bisa digunakan lagi nanti. Bayangkan variabel seperti kotak berlabel: kamu simpan barang (data) di dalamnya, lalu kotak itu diberi nama supaya gampang ditemukan lagi.

Di Python kamu **tidak perlu** menuliskan tipe data terlebih dahulu (beda dengan Java atau C++). Python otomatis mengenali tipe datanya sendiri — ini disebut *dynamic typing*.

````python
# Membuat variabel dan mengisinya dengan nilai
nama = "Xenno"
umur = 20
tinggi_badan = 170.5

print(nama)
print(umur)
print(tinggi_badan)
````

🧠 **Penjelasan Output**

Kode di atas membuat tiga variabel: `nama`, `umur`, dan `tinggi_badan`, lalu Python menyimpan nilainya masing-masing ke memori. `print()` menampilkan isi variabel tersebut ke layar, satu per baris sesuai urutan pemanggilan.

````python
# Nilai variabel bisa diganti (ditimpa) kapan saja
skor = 80
print("Skor awal   :", skor)

skor = 95   # nilai lama ditimpa nilai baru
print("Skor terbaru:", skor)
````

## 1.2 Multiple Assignment

Python memungkinkan kamu mengisi beberapa variabel sekaligus dalam satu baris.

````python
# Isi beberapa variabel sekaligus, masing-masing nilai berbeda
x, y, z = 10, 20, 30
print(x, y, z)

# Isi beberapa variabel dengan nilai yang SAMA
a = b = c = 0
print(a, b, c)
````

## 1.3 Aturan Penamaan Variabel

- Harus diawali huruf atau underscore (`_`) — **tidak boleh** diawali angka
- Hanya boleh berisi huruf, angka, dan underscore (tidak boleh spasi, `-`, `@`, dll.)
- Bersifat *case-sensitive* — `nama` dan `Nama` dianggap dua variabel berbeda
- Tidak boleh sama dengan *keyword* Python (`class`, `if`, `for`, `return`, dst.)

````python
# ✅ Contoh nama variabel yang VALID
nama_lengkap = "Budi Santoso"
_nilai_privat = 100
umur2 = 25
NAMA_KONSTANTA = "PI"

print("Semua variabel valid berhasil dibuat!")
````

- Gunakan nama yang **deskriptif**: `harga_total` jauh lebih jelas daripada `x` atau `h`
- Konstanta (nilai yang tidak berubah selama program berjalan) biasanya ditulis **huruf besar semua**: `PI = 3.14159`, `MAX_USER = 100`

# Bab 2: Tipe Data

## 2.1 Tipe Data Dasar di Python

Python memiliki beberapa tipe data bawaan (*built-in*) yang paling sering dipakai:

| Tipe Data | Nama di Python | Contoh |
|---|---|---|
| Bilangan bulat | `int` | `10`, `-5`, `0` |
| Bilangan desimal | `float` | `3.14`, `-0.5` |
| Teks | `str` | `"Halo"`, `'Python'` |
| Boolean (benar/salah) | `bool` | `True`, `False` |
| Kosong / tanpa nilai | `NoneType` | `None` |

````python
angka_bulat = 10
angka_desimal = 3.14
teks = "Halo, Python!"
benar_salah = True
kosong = None

print(type(angka_bulat))
print(type(angka_desimal))
print(type(teks))
print(type(benar_salah))
print(type(kosong))
````

🧠 **Penjelasan Output**: fungsi `type()` mengembalikan kelas/tipe data dari sebuah nilai. Inilah cara Python "mengenali" jenis data yang kamu simpan.

## 2.2 String Sekilas

String bisa ditulis dengan petik satu `'...'` atau petik dua `"..."` (sama saja), dan *triple quotes* (tiga tanda kutip dua berurutan) untuk teks yang terdiri dari banyak baris. Pembahasan lengkap tentang string (indexing, slicing, method) ada di **Part 4**.

````python
kalimat = "Saya sedang belajar Python"
kalimat2 = 'Bisa juga pakai petik satu, hasilnya sama'
multiline = """Ini string
yang terdiri
dari beberapa baris"""

print(kalimat)
print(len(kalimat))   # menghitung jumlah karakter, termasuk spasi
````

## 2.3 Boolean & Konsep Truthy/Falsy

Selain `True`/`False` secara eksplisit, Python juga menganggap beberapa nilai sebagai "salah" (*falsy*) secara implisit: `0`, `0.0`, `""` (string kosong), `None`, `[]` (list kosong), `{}` (dict kosong). Semua nilai lain dianggap "benar" (*truthy*).

````python
print(bool(0))
print(bool(1))
print(bool(""))
print(bool("halo"))
print(bool(None))
print(bool([]))
print(bool([0]))   # list berisi satu elemen -> truthy walau elemennya 0!
````

## 2.4 Mengecek Tipe Data: `type()` vs `isinstance()`

````python
nilai = 25

print(type(nilai) == int)          # cara 1
print(isinstance(nilai, int))      # cara 2 -> lebih direkomendasikan
print(isinstance(nilai, (int, float)))   # cek beberapa tipe sekaligus
````

## 2.5 Konversi Tipe Data (*Type Casting*)

Mengubah data dari satu tipe ke tipe lain menggunakan fungsi `int()`, `float()`, `str()`, `bool()`.

````python
angka_teks = "100"
angka = int(angka_teks)
print(angka + 50)

pi_teks = "3.14"
pi = float(pi_teks)
print(pi * 2)

angka_ke_teks = str(100)
print("Skor kamu: " + angka_ke_teks)

print(int(7.9))     # 7  -> dipotong (truncate), BUKAN dibulatkan
print(round(7.9))   # 8  -> dibulatkan secara matematis
````

⚠️ **Common Mistakes**
1. `int("3.14")` akan **error** — string berisi angka desimal tidak bisa langsung diubah ke `int`
2. `int(7.9)` hanya **memotong** bagian desimal (truncate), hasilnya `7`, bukan dibulatkan ke `8` — kalau mau pembulatan matematis pakai `round()`
3. Menggabung string dan angka langsung dengan `+`, misalnya `"Skor: " + 90`, akan menghasilkan `TypeError`

````python
try:
    hasil = int("3.14")
except ValueError as e:
    print("Error:", e)
    print("Solusi: konversi ke float() dulu, baru ke int() -> int(float('3.14'))")

print(int(float("3.14")))
````

💡 **Tips & Best Practice**
- Selalu pastikan tipe data sudah sesuai sebelum melakukan operasi — ini paling sering jadi masalah saat data berasal dari `input()`, yang **selalu** bertipe `str`
- Gunakan `isinstance()` untuk validasi tipe data, bukan `type() ==`, karena `isinstance()` lebih aman terutama saat sudah bekerja dengan OOP/inheritance (dibahas di Part 5)

# Bab 3: Operator

## 3.1 Operator Aritmatika

````python
a = 17
b = 5

print("Penjumlahan     :", a + b)
print("Pengurangan     :", a - b)
print("Perkalian       :", a * b)
print("Pembagian       :", a / b)    # hasil SELALU float
print("Pembagian Bulat :", a // b)   # hasil dibulatkan ke bawah
print("Modulus/Sisa Bagi:", a % b)
print("Pangkat         :", a ** b)
````

🧠 **Penjelasan Output**: operator `/` selalu menghasilkan `float` meskipun hasilnya bulat sempurna. Operator `//` (*floor division*) membulatkan hasil bagi ke bawah. Operator `%` (modulus/sisa bagi) sangat berguna, misalnya untuk mengecek genap/ganjil.

````python
angka = 24
sisa = angka % 2
print("Sisa bagi 2:", sisa)
print("0 berarti genap, 1 berarti ganjil")
````

## 3.2 Operator Perbandingan (*Comparison*)

Selalu menghasilkan nilai boolean (`True`/`False`).

````python
x = 10
y = 20

print(x == y)    # sama dengan
print(x != y)    # tidak sama dengan
print(x > y)     # lebih besar
print(x < y)     # lebih kecil
print(x >= 10)   # lebih besar sama dengan
print(x <= 10)   # lebih kecil sama dengan
````

## 3.3 Operator Logika (*Logical*)

`and`, `or`, `not` — dipakai untuk menggabungkan beberapa kondisi boolean.

| A | B | A and B | A or B |
|---|---|---|---|
| True | True | True | True |
| True | False | False | True |
| False | True | False | True |
| False | False | False | False |

````python
umur = 20
punya_ktp = True

print(umur >= 17 and punya_ktp)   # True hanya jika KEDUANYA True
print(umur >= 17 or punya_ktp)    # True jika SALAH SATU True
print(not punya_ktp)              # membalik nilai boolean
````

## 3.4 Operator Assignment (Penugasan Singkat)

````python
nilai = 10
print(nilai)

nilai += 5    # sama dengan: nilai = nilai + 5
print(nilai)

nilai -= 3    # sama dengan: nilai = nilai - 3
print(nilai)

nilai *= 2    # sama dengan: nilai = nilai * 2
print(nilai)

nilai //= 4   # sama dengan: nilai = nilai // 4
print(nilai)
````

## 3.5 Operator Identity (`is`) & Membership (`in`)

- `is` mengecek apakah dua variabel menunjuk **objek yang sama persis di memori** (bukan sekadar nilainya sama)
- `in` mengecek apakah suatu nilai **ada di dalam** sebuah koleksi (list, string, dll.)

````python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(a == b)   # True  -> isinya sama
print(a is b)   # False -> objek berbeda di memori, walau isinya identik
print(a is c)   # True  -> c menunjuk ke objek yang SAMA dengan a

print(2 in a)       # True
print(5 not in a)   # True
````

⚠️ **Common Mistakes**
1. Tertukar antara `=` (assignment/mengisi nilai) dan `==` (perbandingan) — ini penyebab bug paling umum bagi pemula
2. Tertukar antara `is` dan `==` — gunakan `==` untuk membandingkan **nilai**, gunakan `is` hanya untuk mengecek **identitas objek** (paling sering dipakai untuk `is None`)
3. Lupa urutan operator (*operator precedence*)

````python
hasil = 2 + 3 * 4
print(hasil)     # 14, BUKAN 20 -> perkalian didahulukan dari penjumlahan

hasil2 = (2 + 3) * 4
print(hasil2)    # 20 -> tanda kurung mengubah urutan pengerjaan
````

💡 **Tips & Best Practice**

Urutan pengerjaan operator dari yang **paling didahulukan**:
1. `()` — tanda kurung
2. `**` — pangkat
3. `*`, `/`, `//`, `%` — kali, bagi, bagi bulat, modulus
4. `+`, `-` — tambah, kurang
5. `==`, `!=`, `>`, `<`, dst — perbandingan
6. `and`, `or`, `not` — logika

Meskipun tidak wajib, gunakan tanda kurung `()` untuk memperjelas maksud kode — ini membuat kode lebih mudah dibaca orang lain (dan dirimu sendiri di masa depan).

## 4.3 String Formatting

Ada tiga cara memformat string di Python:
1. `%`-formatting (cara lama)
2. `.format()`
3. **f-string** (Python 3.6+, **paling direkomendasikan**)

````python
nama = "Xenno"
skor = 95.5

# Cara lama (%-formatting)
print("Nama: %s, Skor: %.1f" % (nama, skor))

# .format()
print("Nama: {}, Skor: {}".format(nama, skor))

# f-string -> PALING DIREKOMENDASIKAN karena paling ringkas & mudah dibaca
print(f"Nama: {nama}, Skor: {skor}")
````

````python
angka = 1234567.891

print(f"{angka:.2f}")      # 2 angka di belakang koma
print(f"{angka:,.2f}")     # + pemisah ribuan
print(f"{angka:12.2f}")    # lebar minimal 12 karakter, rata kanan
print(f"{75:>6}|")         # rata kanan, lebar 6
print(f"{75:<6}|")         # rata kiri, lebar 6
print(f"{75:^6}|")         # rata tengah, lebar 6
````

⚠️ **Common Mistakes**
1. Lupa mengonversi hasil `input()` sebelum dipakai operasi matematika → `TypeError: can only concatenate str`
2. Salah menulis format specifier, misalnya `{angka:2f}` (lupa titik sebelum jumlah desimal) — seharusnya `{angka:.2f}`
3. Menggabung string dan angka pakai `+` tanpa `str()` terlebih dahulu

````python
angka = 90
try:
    print("Skor: " + angka)
except TypeError as e:
    print("Error:", e)
    print("Solusi: 'Skor: ' + str(angka)  ATAU  f'Skor: {angka}'")

print("Skor: " + str(angka))
print(f"Skor: {angka}")
````

💡 **Tips & Best Practice**
- Jadikan **f-string** sebagai standar untuk formatting string — lebih ringkas dan mudah dibaca dibanding dua cara lainnya
- Selalu validasi/konversi input pengguna sebelum diproses lebih lanjut
- Untuk menampilkan nilai uang/desimal, gunakan format specifier `:,.2f` supaya rapi dan mudah dibaca

### Kunci Jawaban Mini Project: Kartu Identitas Digital

````python
nama = "Xenno"                  # contoh nilai, ganti dengan input() saat dipakai sendiri
tahun_lahir = 2005
tinggi_cm = 170.0
berat_kg = 65.0

umur = 2026 - tahun_lahir
tinggi_m = tinggi_cm / 100
bmi = berat_kg / (tinggi_m ** 2)

print("=" * 32)
print("    KARTU IDENTITAS DIGITAL")
print("=" * 32)
print(f"Nama          : {nama}")
print(f"Umur          : {umur} tahun")
print(f"Tinggi Badan  : {tinggi_cm:.1f} cm")
print(f"Berat Badan   : {berat_kg:.1f} kg")
print(f"BMI           : {bmi:.2f}")
print("=" * 32)
````

````python
# Membuat Dataset
import numpy as np

np.random.seed(42)

sales = np.random.randint(50, 201, size=30)

print(f'''Data Penjualan 
{sales}''')
print('\n'*5)



# Statistik Dasar
total = np.sum(sales)
mean = np.mean(sales)
maks = np.max(sales)
mins = np.min(sales)
std = np.std(sales)
var = np.var(sales)

print(f'''Total penjualan : {total}
Rata-rata       : {mean}
Nilai maksimum  : {maks}
Nilai minimum   : {mins}
Standar deviasi : {std}
Variansi        : {var}''')
print('\n'*5)


# Hari dengan penjualan tinggi
idx = np.where(sales>150)
nilai = sales[idx]

print("==== HARI DENGAN PENJUALAN TINGGI ====")
print(f'''Indeks Harinya     : {idx}
Nilai Penjualannya : {nilai}''')
print('\n'*5)


# Sorting
asc = np.sort(sales)
print(f'''Data Setelah Diurutkan
{asc}''')
print('\n'*5)



# Reshape
sales_baru = sales.reshape(6, 5)
print(f'''Data dalam Matriks
{sales_baru}''')
print('\n'*5)



# Analisis Per Minggu
total1 = np.sum(sales_baru, axis=1)
mean1 = np.mean(sales_baru, axis=1)

print(f'''Total Penjualan Tiap Minggu : {total1}
Rata-rata Tiap Minggu       : {mean1}''')
print('\n'*5)



# Simulasi Diskon
diskon = sales*0.9
print(f'''==== Perbandingan Harga Sebelum & Setelah Diskon 10% ====
Data Lama : {sales}
Data Baru : {diskon}''')
print('\n'*5)



# Simpan Hasil 
np.save("sales_discount.npy", diskon)
load = np.load("sales_discount.npy")
print(load)

#Bonus Challenge
lebih = np.where(sales > mean)
kurang = np.where(sales < mean)
print(f'''Jumlah Hari dengan Penjualan di atas Rata-rata : {sales[lebih].size}
Jumlah Hari dengan Penjualan di bawah Rata-rata : {sales[kurang].size}''')
````

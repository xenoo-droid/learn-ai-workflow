# 📚 Pandas — Sorting Data

## ⭐ Bagian 1: `sort_values()`

## 🎯 Tujuan Pembelajaran

Setelah materi ini, kamu akan mampu:

- Mengurutkan data berdasarkan satu kolom.
- Mengurutkan secara _ascending_ (kecil → besar / A → Z).
- Mengurutkan secara _descending_ (besar → kecil / Z → A).

---

### 📌 Dataset

```python
import pandas as pd

df = pd.DataFrame({
    "Nama": ["Andi", "Budi", "Citra", "Doni", "Eko"],
    "Umur": [23, 21, 19, 25, 22],
    "Nilai": [88, 75, 90, 80, 95]
})

print(df)
```

Output:

```text
    Nama  Umur  Nilai
0   Andi    23     88
1   Budi    21     75
2  Citra    19     90
3   Doni    25     80
4    Eko    22     95
```

---

### 1️⃣ Mengurutkan Berdasarkan Satu Kolom

Sintaks:

```python
df.sort_values(by = 'Nama')
```

Contoh:

```python
print(df.sort_values(by = 'Nilai'))
```

Output:

```text
    Nama  Umur  Nilai
1   Budi    21     75
3   Doni    25     80
0   Andi    23     88
2  Citra    19     90
4    Eko    22     95
```

Karena `ascending = True` secara default, hasilnya diurutkan dari **nilai terkecil ke terbesar**.

---

### 2️⃣ Ascending (Default)

```python
print(df.sort_values(by = 'Umur', ascending = True))
```

Output:

```text
    Nama  Umur  Nilai
2  Citra    19     90
1   Budi    21     75
4    Eko    22     95
0   Andi    23     88
3   Doni    25     80
```

---

### 3️⃣ Descending

Gunakan:

```python
ascenoding=False
```

Contoh:

```python
print(df.sort_values(by='Nilai', ascending=False))
```

Output:

```text
    Nama  Umur  Nilai
4    Eko    22     95
2  Citra    19     90
0   Andi    23     88
3   Doni    25     80
1   Budi    21     75
```

---

### 💡 Ringkasan

| Fungsi                    | Kegunaan                      |
| ------------------------- | ----------------------------- |
| `sort_values(by="kolom")` | Mengurutkan berdasarkan kolom |
| `ascending=True`          | Urutan naik (default)         |
| `ascending=False`         | Urutan turun                  |

---

## ⭐ Bagian 2: Multi-column Sorting

## 🎯 Tujuan Pembelajaran

Setelah materi ini, kamu akan mampu:

- Mengurutkan data berdasarkan lebih dari satu kolom.
- Menggunakan daftar pada parameter `by`.
- Menggunakan daftar pada parameter `ascending`.

---

### 📌 Dataset

```python
import pandas as pd

df = pd.DataFrame({
    "Nama": ["Andi", "Budi", "Citra", "Doni", "Eko", "Fani"],
    "Kelas": ["A", "A", "B", "B", "A", "B"],
    "Nilai": [88, 88, 90, 80, 95, 90]
})

print(df)
```

Output:

```text
    Nama Kelas  Nilai
0   Andi     A      88
1   Budi     A      88
2  Citra     B      90
3   Doni     B      80
4    Eko     A      95
5   Fani     B      90
```

---

### 1️⃣ Sorting Berdasarkan Dua Kolom

Sintaks:

```python
df.sort_values(by=['Kelas', 'Nilai'])
```

Contoh:

```python
print(df.sort_values(by=['Kelas', 'Nilai']))
```

Hasilnya:
- Data diurutkan berdasarkan **Kelas** terlebih dahulu.
- Jika kelas sama, maka diurutkan berdasarkan **Nilai**.

Output:

```text
    Nama Kelas  Nilai
1   Budi     A      88
0   Andi     A      88
4    Eko     A      95
3   Doni     B      80
2  Citra     B      90
5   Fani     B      90
```

---

### 2️⃣ Mengatur Urutan Tiap Kolom

Kita juga bisa menentukan urutan masing-masing kolom.

Contoh:

```python
print(
    df.sort_values(
        by=['Kelas', 'Nilai'],
        ascending=[True, False]
    )
)
```

Artinya:
- Kelas → A ke B
- Nilai → terbesar ke terkecil

Output:

```text
    Nama Kelas  Nilai
4    Eko     A      95
0   Andi     A      88
1   Budi     A      88
2  Citra     B      90
5   Fani     B      90
3   Doni     B      80
```

---

### 3️⃣ Penjelasan `ascending`

Jika terdapat dua kolom pada `by`, maka `ascending` juga harus memiliki dua nilai.

Contoh:

```python
ascending=[True, False]
```

Berarti:
- Kolom pertama → Ascending
- Kolom kedua → Descending

Contoh lain:

```python
ascending=[False, True]
```

Berarti:
- Kolom pertama → Ascending
- Kolom kedua → Descending

---

### 💡 Ringkasan

| Sintaks                   | Fungsi                                   |
| ------------------------- | ---------------------------------------- |
| `by=["A", "B"]`           | Mengurutkan berdasarkan dua kolom        |
| `ascending=[True, False]` | Atur urutan setiap kolom secara terpisah |

---

## ⭐ Bagian 3: `sort_index()`

## 🎯 Tujuan Pembelajaran

Setelah materi ini, kamu akan mampu:

- Mengurutkan DataFrame berdasarkan **index**.
- Mengurutkan index secara ascending maupun descending.
- Memahami perbedaan `sort_values()` dan `sort_index()`.

---

### 📌 Dataset

Gunakan dataset berikut.

```python
import pandas as pd

df = pd.DataFrame({
    "Nama": ["Andi", "Budi", "Citra", "Doni", "Eko"],
    "Nilai": [88, 75, 90, 80, 95]
})

# Acak urutan data
df = df.sort_values(by="Nilai", ascending=False)

print(df)
```

Output:

```text
    Nama  Nilai
4    Eko     95
2  Citra     90
0   Andi     88
3   Doni     80
1   Budi     75
```

Perhatikan bahwa index sekarang menjadi:

```text
4
2
0
3
1
```

Urutan index sudah tidak beraturan.

---

### 1️⃣ Mengurutkan Berdasarkan Index

Sintaks:

```python
df.sort_index()
```

Contoh:

```python
print(df.sort_index())
```

Output:

```text
    Nama  Nilai
0   Andi     88
1   Budi     75
2  Citra     90
3   Doni     80
4    Eko     95
```

Sekarang index kembali menjadi:

```text
0
1
2
3
4
```

### 2️⃣ Descending Index

```python
print(
    df.sort_index(
        ascending=False
    )
)
```

Output:

```text
    Nama  Nilai
4    Eko     95
3   Doni     80
2  Citra     90
1   Budi     75
0   Andi     88
```

---

### 📌 Perbedaan Penting

#### `sort_values()`

Mengurutkan berdasarkan isi kolom.

```python
df.sort_values(by='Nilai'))
```

Hasil:

```text
75
80
88
90
95
```

---

#### `sort_index()`

Mengurutkan berdasarkan nomor index.

```python
dff.sort_index()
```

Hasil

```text

1
2
3
4
```

Isi kolom tidak menjadi dasar pengurutan.

## ⭐ Bagian 4: `reset_index()`

## 🎯 Tujuan Pembelajaran

Setelah materi ini, kamu akan mampu:

- Mereset index menjadi `0, 1, 2, ...`
- Memahami parameter `drop=True`
- Mengetahui perbedaan `sort_index()` dan `reset_index()`

---

### 📌 Dataset

Gunakan dataset berikut.

```python
import pandas as pd

df = pd.DataFrame({
    "Nama": ["Andi", "Budi", "Citra", "Doni", "Eko"],
    "Nilai": [88, 75, 90, 80, 95]
})

df = df.sort_values(by="Nilai", ascending=False)

print(df)
```

Output:

```text
    Nama  Nilai
4    Eko     95
2  Citra     90
0   Andi     88
3   Doni     80
1   Budi     75
```

Perhatikan bahwa index masih:

```text
4
2
0
3
1
```

---

### 1️⃣ Mereset Index

Sintaks:

```python
df.reset_index()
```

Contoh:

```python
print(df.reset_index())
```

Output:

```text
   index   Nama  Nilai
0      4    Eko     95
1      2  Citra     90
2      0   Andi     88
3      3   Doni     80
4      1   Budi     75
```

Apa yang terjadi?
- Index baru menjadi `0-4`.
- Index lama **tidak hilang**, tetapi dipindahkan menjadi kolombaru bernama `index`.

---

### 2️⃣ Menghilangkan Index Lama

Biasanya kita tidak membutuhkan index lama.

Sintaks:

```python
df.reset_index(drop=True)
```

Contoh:

```python
print(df.reset_index(drop=True))
```

Output:

```text
    Nama  Nilai
0    Eko     95
1  Citra     90
2   Andi     88
3   Doni     80
4   Budi     75
```

Sekarang:
- Index menjadi `0-4`.
- Kolom `index` **tidak dibuat**.

---

### 3️⃣ Menyimpan Perubahan

Sama seperti `sort_value()`, `reset_index()` **mengembalikan DataFrame baru**.

Kalau ingin perubahan tersimpan:

```python
df = df.reset_index(drop=True)
```

atau

```python
df.reset_index(drop=True, inplace=True)
```

> Untuk pembelajaran kita, gunakan cara pertama (`df = ...`) karena lebih konsisten dengan materi sebelumnya.

---

### 📌 Perbedaan Penting

| Fungsi          | Tujuan                                       |
| --------------- | -------------------------------------------- |
| `sort_values()` | Mengurutkan berdasarkan isi kolom            |
| `sort_index()`  | Mengurutkan berdasarkan index yang sudah ada |
| `reset_index()` | Membuat index baru mulai dari 0              |

---

### 💡 Ringkasan

| Sintaks                          | Hasil                                 |
| -------------------------------- | ------------------------------------- |
| `reset_index()`                  | Index baru + index lama menjadi kolom |
| `reset_index(drop=True)`         | Index baru, index lama dibuang        |
| `df = df.reset_index(drop=True)` | Menyimpan hasil reset ke DataFrame    |

# Bab 1: Error Handling

## 1.1 Kenapa Perlu Error Handling?

Tanpa penanganan, error (disebut juga *exception*) akan **menghentikan program secara paksa**. Dengan `try-except`, program bisa menangkap error tersebut dan tetap berjalan dengan baik.

```python
angka1 = 10
angka2 = 0

try:
    hasil = angka1 / angka2
    print(hasil)
except ZeroDivisionError:
    print("Tidak bisa membagi dengan nol!")

print("Program tetap lanjut berjalan setelah error ditangani")
```

## 1.2 Menangkap Beberapa Jenis Exception

Satu blok `try` boleh punya lebih dari satu `except`, masing-masing untuk jenis error yang berbeda.

```python
def bagi(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Error: pembagi tidak boleh nol")
    except TypeError:
        print("Error: kedua nilai harus berupa angka")

print(bagi(10, 2))
print(bagi(10, 0))
print(bagi(10, "dua"))
```

## 1.3 Menangkap Beberapa Error Sekaligus & Mengakses Pesan Error

Beberapa jenis error bisa ditangani sekaligus dengan tuple. Pesan error aslinya bisa diambil dengan `as e`.

```python
def proses(data):
    try:
        angka = int(data)
        return 100 / angka
    except (ValueError, ZeroDivisionError) as e:
        print(f"Terjadi error: {type(e).__name__} -> {e}")

print(proses("10"))
print(proses("abc"))
print(proses("0"))
```

## 1.4 `else` dan `finally`

- Blok `else` dijalankan **hanya jika tidak ada error** di `try`
- Blok `finally` **selalu** dijalankan, apa pun hasilnya — cocok untuk kode "pembersihan" yang wajib jalan

```python
def proses_angka(angka_teks):
    try:
        angka = int(angka_teks)
    except ValueError:
        print("Input bukan angka valid")
    else:
        print(f"Berhasil dikonversi: {angka}")
    finally:
        print("Selesai memproses (selalu dijalankan, error atau tidak)")

proses_angka("25")
print("---")
proses_angka("abc")
```

## 1.5 `raise` — Melempar Exception Sendiri

Kamu bisa memicu error sendiri dengan `raise`, misalnya untuk menolak input yang tidak masuk akal secara logika program (walau secara tipe data sebenarnya valid).

```python
def daftar_umur(umur):
    if umur < 0:
        raise ValueError("Umur tidak boleh negatif")
    if umur > 150:
        raise ValueError("Umur tidak masuk akal")
    return f"Umur {umur} tahun berhasil didaftarkan"

try:
    print(daftar_umur(25))
    print(daftar_umur(-5))
except ValueError as e:
    print("Ditolak:", e)
```

## 1.6 Membuat Exception Sendiri (Custom Exception)

Untuk kasus khusus dalam program, kamu bisa membuat jenis exception sendiri dengan mewarisi (*inherit*) dari `Exception` — konsep pewarisan ini dibahas lebih detail di **Bab 3: OOP**.

```python
class SaldoTidakCukupError(Exception):
    # Exception khusus untuk kasus saldo tidak mencukupi
    pass

def tarik_tunai(saldo, jumlah):
    if jumlah > saldo:
        raise SaldoTidakCukupError(f"Saldo {saldo} tidak cukup untuk menarik {jumlah}")
    return saldo - jumlah

try:
    sisa = tarik_tunai(50000, 100000)
except SaldoTidakCukupError as e:
    print("Transaksi gagal:", e)
```

⚠️ **Common Mistakes**
1. Menangkap error terlalu umum dengan `except:` tanpa jenis spesifik — sulit tahu error apa yang sebenarnya terjadi
2. Meletakkan terlalu banyak baris kode di dalam satu `try`, sehingga tidak jelas baris mana yang benar-benar bisa error
3. Memakai `try-except` untuk menyembunyikan bug alih-alih benar-benar menanganinya (*silent failure* — program terlihat baik-baik saja padahal ada masalah tersembunyi)

```python
try:
    x = 10 / 0
except:   # BURUK: menangkap SEMUA jenis error tanpa pandang bulu
    print("Ada error (tapi tidak jelas error apa)")

try:
    x = 10 / 0
except ZeroDivisionError as e:   # LEBIH BAIK: spesifik dan informatif
    print(f"ZeroDivisionError: {e}")
```

💡 **Tips & Best Practice**
- Tangkap exception se-spesifik mungkin (`ZeroDivisionError`, `ValueError`, dst.), hindari `except:` kosong
- Gunakan `finally` untuk kode yang **wajib** jalan apa pun hasilnya, misalnya menutup koneksi atau file
- Berikan pesan yang jelas saat `raise` exception, supaya mudah dipahami penyebabnya

# Bab 2: File I/O

## 2.1 Mode Membuka File

| Mode | Arti |
|---|---|
| `"r"` | *Read* — membaca file (error kalau file tidak ada) |
| `"w"` | *Write* — menulis file, **menimpa** seluruh isi lama kalau file sudah ada |
| `"a"` | *Append* — menambahkan isi baru di akhir file tanpa menghapus isi lama |
| `"r+"` | Baca sekaligus tulis |

## 2.2 Menulis File dengan Context Manager (`with`)

`with` memastikan file **otomatis tertutup** setelah selesai dipakai, walau terjadi error di tengah proses — ini praktik yang direkomendasikan dibanding `open()`/`close()` manual.

```python
with open("catatan.txt", "w") as file:
    file.write("Baris pertama\n")
    file.write("Baris kedua\n")

print("File berhasil ditulis")
```

## 2.3 Membaca File

```python
with open("catatan.txt", "r") as file:
    isi = file.read()
print(isi)
```

```python
with open("catatan.txt", "r") as file:
    for baris in file:
        print("Baris:", baris.strip())   # .strip() menghapus karakter newline di akhir
```

```python
with open("catatan.txt", "r") as file:
    daftar_baris = file.readlines()
print(daftar_baris)
```

## 2.4 Menambah Isi File (Mode `"a"`)

```python
with open("catatan.txt", "a") as file:
    file.write("Baris ketiga (ditambahkan)\n")

with open("catatan.txt", "r") as file:
    print(file.read())
```

## 2.5 Menangani File yang Tidak Ada

```python
try:
    with open("file_tidak_ada.txt", "r") as file:
        print(file.read())
except FileNotFoundError as e:
    print("FileNotFoundError:", e)
```

⚠️ **Common Mistakes**
1. Membuka file tanpa `with` (pakai `open()`/`close()` manual) dan lupa menutupnya — boros resource, file bisa "terkunci"
2. Salah pilih mode `"w"` padahal maksudnya `"a"` — mode `"w"` **menimpa habis** isi lama, data sebelumnya hilang
3. Mencoba membaca file yang belum ada tanpa penanganan error → `FileNotFoundError` menghentikan program

```python
# Simulasi kesalahan #2: mode "w" menimpa isi lama tanpa disadari
with open("contoh_timpa.txt", "w") as file:
    file.write("Data penting jangan sampai hilang\n")

with open("contoh_timpa.txt", "w") as file:   # niatnya menambah, tapi salah pakai "w"
    file.write("Data baru\n")

with open("contoh_timpa.txt", "r") as file:
    print(file.read())   # "Data penting jangan sampai hilang" SUDAH HILANG!
```

💡 **Tips & Best Practice**
- Selalu gunakan `with open(...) as file:` daripada `open()`/`close()` manual
- Pastikan mode (`"r"`/`"w"`/`"a"`) benar-benar sesuai maksud sebelum menjalankan kode
- Gabungkan dengan `try-except` saat membaca file yang keberadaannya tidak selalu pasti

# Bab 3: OOP (Object-Oriented Programming)

## 3.1 Apa itu Class dan Object?

**Class** adalah "cetakan"/blueprint untuk membuat objek. **Object** adalah hasil konkret dari cetakan tersebut. Analogi: `Class Mobil` adalah rancangan/blueprint mobil secara umum, sedangkan `mobil_saya` adalah satu unit mobil nyata yang dibuat dari rancangan itu.

```python
class Mahasiswa:
    pass

mhs1 = Mahasiswa()
mhs2 = Mahasiswa()

print(type(mhs1))
print(mhs1 is mhs2)   # False -> dua object yang BERBEDA, walau dari class yang sama
```

## 3.2 `__init__` dan Attribute

`__init__` adalah *constructor* — method spesial yang otomatis dijalankan setiap kali object baru dibuat, biasanya dipakai untuk mengisi nilai awal attribute.

```python
class Mahasiswa:
    def __init__(self, nama, jurusan, semester):
        self.nama = nama
        self.jurusan = jurusan
        self.semester = semester

mhs1 = Mahasiswa("Xenno", "Data Science", 3)
mhs2 = Mahasiswa("Budi", "Teknik Informatika", 5)

print(mhs1.nama, mhs1.jurusan, mhs1.semester)
print(mhs2.nama, mhs2.jurusan, mhs2.semester)
```

🧠 **Penjelasan `self`**: `self` merujuk ke **object itu sendiri**. Python otomatis mengisinya setiap kali method dipanggil dari sebuah object — kamu tidak perlu (dan tidak boleh) mengisinya secara manual saat memanggil method. `self` harus selalu jadi parameter **pertama** di setiap method instance.

## 3.3 Method (Instance Method)

Method adalah function yang didefinisikan di dalam class, dan bisa mengakses/mengubah attribute object lewat `self`.

```python
class Mahasiswa:
    def __init__(self, nama, jurusan, semester):
        self.nama = nama
        self.jurusan = jurusan
        self.semester = semester

    def perkenalan(self):
        return f"Halo, saya {self.nama}, mahasiswa {self.jurusan} semester {self.semester}"

    def naik_semester(self):
        self.semester += 1

mhs1 = Mahasiswa("Xenno", "Data Science", 3)
print(mhs1.perkenalan())

mhs1.naik_semester()
print(mhs1.semester)
```

## 3.4 Class Attribute vs Instance Attribute

- **Instance attribute** (didefinisikan lewat `self.nama_attribute`): nilainya **beda-beda** untuk tiap object
- **Class attribute** (didefinisikan langsung di dalam class, di luar method): nilainya **sama** untuk semua object, kecuali sengaja ditimpa per object

```python
class Mahasiswa:
    kampus = "Telkom University"   # class attribute -> SAMA untuk semua object

    def __init__(self, nama):
        self.nama = nama            # instance attribute -> BEDA untuk tiap object

mhs1 = Mahasiswa("Xenno")
mhs2 = Mahasiswa("Budi")

print(mhs1.kampus, "|", mhs2.kampus)   # sama, karena class attribute
print(mhs1.nama, "|", mhs2.nama)        # beda, karena instance attribute

Mahasiswa.kampus = "Universitas Lain"
print(mhs1.kampus, "|", mhs2.kampus)   # ikut berubah KEDUANYA, karena class attribute-nya yang diubah
```

## 3.5 Encapsulation — Public vs "Private"

Python tidak punya konsep private yang benar-benar ketat seperti beberapa bahasa lain, tapi punya **konvensi**:
- Nama biasa (`nama`) → **public**, bebas diakses dari luar
- Awalan underscore tunggal (`_pin`) → **semi-private**, konvensi "sebaiknya jangan diakses langsung dari luar" (tapi secara teknis tetap bisa)
- Awalan underscore ganda (`__saldo`) → Python melakukan *name mangling*, membuatnya jauh lebih sulit diakses langsung dari luar class

```python
class Rekening:
    def __init__(self, pemilik, saldo_awal):
        self.pemilik = pemilik      # public
        self._pin = "1234"           # semi-private (konvensi)
        self.__saldo = saldo_awal     # "private" (name mangling)

    def cek_saldo(self):
        return self.__saldo

    def setor(self, jumlah):
        self.__saldo += jumlah

rek = Rekening("Xenno", 100000)
print(rek.cek_saldo())
rek.setor(50000)
print(rek.cek_saldo())

try:
    print(rek.__saldo)   # error, karena name mangling menyembunyikan atribut ini dari akses langsung
except AttributeError as e:
    print("AttributeError:", e)
```

## 3.6 Inheritance (Pewarisan)

Class baru bisa **mewarisi** semua attribute dan method dari class lain, lalu menambah atau menimpa (*override*) sebagian di antaranya.

```python
class Hewan:
    def __init__(self, nama):
        self.nama = nama

    def bersuara(self):
        return "Suara hewan pada umumnya"

class Kucing(Hewan):        # Kucing MEWARISI semua yang ada di Hewan
    def bersuara(self):      # method overriding -> menimpa method dari class induk
        return "Meong!"

class Anjing(Hewan):
    def bersuara(self):
        return "Guk guk!"

kucing = Kucing("Kitty")
anjing = Anjing("Rex")

print(kucing.nama, "-", kucing.bersuara())
print(anjing.nama, "-", anjing.bersuara())
```

## 3.7 `super()` — Memanggil Method dari Class Induk

`super()` dipakai untuk mengakses method milik class induk, biasanya dipakai di dalam `__init__` class turunan supaya tidak perlu menulis ulang logika yang sama.

```python
class Karyawan:
    def __init__(self, nama, gaji_pokok):
        self.nama = nama
        self.gaji_pokok = gaji_pokok

    def hitung_gaji(self):
        return self.gaji_pokok

class Manajer(Karyawan):
    def __init__(self, nama, gaji_pokok, tunjangan):
        super().__init__(nama, gaji_pokok)   # panggil __init__ milik class induk
        self.tunjangan = tunjangan

    def hitung_gaji(self):
        gaji_dasar = super().hitung_gaji()    # panggil method milik class induk
        return gaji_dasar + self.tunjangan

manajer = Manajer("Xenno", 8000000, 2000000)
print(manajer.hitung_gaji())
```

## 3.8 Dunder/Magic Method: `__str__`

Method dengan awalan dan akhiran garis bawah ganda (*dunder*, singkatan dari *double underscore*) adalah method spesial bawaan Python. `__str__` menentukan bagaimana object ditampilkan saat di-`print()`.

```python
class Buku:
    def __init__(self, judul, penulis):
        self.judul = judul
        self.penulis = penulis

    def __str__(self):
        return f"{self.judul} oleh {self.penulis}"

buku1 = Buku("Python Dasar", "Xenno")
print(buku1)          # otomatis memanggil __str__ saat di-print
print(str(buku1))
```

⚠️ **Common Mistakes**
1. Lupa menulis `self` sebagai parameter pertama di setiap method instance
2. Tidak mendefinisikan attribute lewat `__init__`, menyebabkan `AttributeError` kalau attribute itu diakses sebelum pernah di-*set*
3. Mengira class attribute berperilaku sama seperti instance attribute — mengubah lewat `NamaClass.attribute` akan memengaruhi **semua** object sekaligus

```python
class ContohBug:
    def set_nilai(self, nilai):
        self.nilai = nilai   # baru terdefinisi SETELAH method ini dipanggil, tidak ada __init__

obj = ContohBug()
try:
    print(obj.nilai)   # AttributeError -> belum pernah di-set
except AttributeError as e:
    print("AttributeError:", e)

obj.set_nilai(100)
print(obj.nilai)   # sekarang baru ada, setelah dipanggil
```

💡 **Tips & Best Practice**
- Gunakan `__init__` untuk memastikan semua attribute penting sudah punya nilai awal sejak object dibuat
- Gunakan double underscore (`__attribute`) hanya untuk data yang memang ingin benar-benar disembunyikan/dilindungi dari akses langsung
- OOP paling berguna ketika data dan perilaku (*behavior*) yang berkaitan erat dikelompokkan jadi satu — jangan memaksakan OOP untuk kasus sederhana yang sebenarnya cukup dengan function biasa

# 🚀 Mini Project: Sistem Pencatatan Pengeluaran

**Tujuan**: menggabungkan error handling, file I/O, dan OOP dalam satu program utuh.

**Deskripsi**: buat class `Pengeluaran` untuk merepresentasikan satu catatan pengeluaran, validasi input dengan error handling, simpan seluruh catatan ke file, lalu baca kembali dan tampilkan totalnya.

```python
class Pengeluaran:
    def __init__(self, deskripsi, jumlah, kategori):
        self.deskripsi = deskripsi
        self.jumlah = jumlah
        self.kategori = kategori

    def __str__(self):
        return f"[{self.kategori}] {self.deskripsi}: Rp{self.jumlah:,.0f}"


def tambah_pengeluaran(deskripsi, jumlah_teks, kategori):
    try:
        jumlah = float(jumlah_teks)
        if jumlah <= 0:
            raise ValueError("Jumlah harus lebih besar dari nol")
    except ValueError as e:
        print(f"Gagal menambahkan '{deskripsi}': {e}")
        return None
    return Pengeluaran(deskripsi, jumlah, kategori)


def simpan_ke_file(daftar_pengeluaran, nama_file="pengeluaran.txt"):
    with open(nama_file, "w") as file:
        for p in daftar_pengeluaran:
            file.write(str(p) + "\n")


# Simulasi data mentah (biasanya berasal dari input() pengguna)
data_mentah = [
    ("Makan siang", "25000", "Makanan"),
    ("Bensin", "50000", "Transport"),
    ("Buku catatan", "abc", "Pendidikan"),   # sengaja tidak valid, untuk menguji error handling
    ("Kopi", "18000", "Makanan"),
]

daftar_pengeluaran = []
for deskripsi, jumlah_teks, kategori in data_mentah:
    hasil = tambah_pengeluaran(deskripsi, jumlah_teks, kategori)
    if hasil is not None:
        daftar_pengeluaran.append(hasil)

simpan_ke_file(daftar_pengeluaran)

print("=== Daftar Pengeluaran Tersimpan ===")
with open("pengeluaran.txt", "r") as file:
    print(file.read())

total = sum(p.jumlah for p in daftar_pengeluaran)
print(f"Total pengeluaran: Rp{total:,.0f}")
```

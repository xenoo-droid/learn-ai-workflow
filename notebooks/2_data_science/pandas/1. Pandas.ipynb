## 🌱 Chapter 1: Pandas Series 🎯

### Apa itu pandas?

Sebelum masuk ke Series, kita pahami dulu perannya.

**NumPy** sangat cepat untuk mengolah data numerik, tetapi hanya menyediakan **array** tanpa label.

Contoh NumPy:

```python
import numpy as np

nilai = np.array([90, 85, 95])
print(nilai)
```

Output

```text
[90 85 95]
```

Kalau beberapa hari kemudian kamu melihat angka-angka itu lagi, kamu tidak tahu:
- Milik siapa?
- Nilai mata pelajaran apa?

Nah, **Pandas** menyelesaikan masalah itu.

---

### Apa itu Series?

**Series** adalah struktur data 1 **dimensi** yang memiliki:
- data
- index (label)

Bayangkan seperti ini:

| Index | Value |
| ----: | ----: |
|     0 |    90 |
|     1 |    85 |
|     2 |    95 |

Kalau memakai label:

| Nama  | Nilai |
| ----- | ----: |
| Andi  |    90 |
| Budi  |    85 |
| Citra |    95 |

Itulah Series.

Secara sederhana:
> **Series = NumPy Array + Label (Index)**

---

### Mengimpor pandas

Standar yang hampir selalu digunakan:

```python
import pandas as pd
```

Alias `pd` adalah konversi yang dipakai hampir di semua proyek Data Science.

---

### Membuat Series

#### 1. Dari List

```python
import pandas as pd

nilai = pd.Series([90, 85, 95])

print(nilai)
```

Output

```text
0    90
1    85
2    95
dtype: int64
```

Perthatikan bahwa Pandas otomatis membuat index `0, 1, 2`.

---

#### 2. Memberikan Index Sendiri

```python
nilai = pd.Series(
    [90, 85, 95],
    index = = ['Andi', 'Budi', 'Citra']
)

print(nilai)
```

Output

```text
Andi     90
Budi     85
Citra    95
dtype: int64
```

Sekarang datanya jauh lebih informatif 

---

#### 3. Dari Dictionary

```python
nilai = pd.Series({
    'Andi' : 90,
    'Budi' : 85,
    'Citra' : 95
})

print(nilai)
```

Outputnya sama.

Ini salah satu cara yang paling sering digunakan.

---

### Melihat Tipe Data

```python
print(type(nilail))
```

Output:

```python
<class 'pandas.core.series.Series'>
```

---

### Series Memilikki Index

```python
print(nilai.index)
```

Output:

```python
Index(['Andi', 'Budi', 'Citra'], dtype='object')
```

---

### Mengambil Nilai

Dengan Label:

```python
print(nilai['Andi'])
```

Output:

```python
90
```

---

### Perbedaan NumPy dan Series

| NumPy Array             | Pandas Series                               |
| ----------------------- | ------------------------------------------- |
| Tidak punya label       | Punya label/index                           |
| Fokus komputasi numerik | Fokus analisis data                         |
| Sangat cepat            | Sedikit lebih berat tetapi lebih informatif |

---

### Kapan Menggunakan Series?

Gunakan **Series** ketika data hanya memiliki 1 kolom.

Contoh:
- Umur Mahasiswa
- Harga saham
- Suhu harian
- Penjualan per hari

Kalau datanya sudah memiliki banyak kolom, kita akan menggunakan**DataFrame**, yang akan dipelajari setelah Series.

---

### Ringkasan Materi 

- ✅ Pandas adalah library untuk analisis data.
- ✅ Series adalah struktur data 1 dimensi dengan index.
- ✅ Series bisa dibuat dari list maupun dictionary.
- ✅ Index bisa dibuat otomatis atau ditentukan sendiri.
- ✅ Series sangat cocok untuk merepresentasikan satu kolom data.

## 📋 Chapter 2: Pandas DataFrame 🎯

### Apa itu DataFrame?

Kalau Series adalah data 1 dimensi, maka DataFrame adalah data 2 dimensi.

Bayangkan seperti tabel Excel.

| Nama  | Umur | Nilai |
| ----- | ---: | ----: |
| Andi  |   20 |    88 |
| Budi  |   21 |    78 |
| Citra |   19 |    92 |

Nah, tabel seperti ini adalah **DataFrame**.

Secara sederhana:
> **DataFrame = kumpulan beberapa Series yang memiliki index yang sama.**

---

### Struktur DataFrame

```text
        Kolom
      ↓

      Nama   Umur   Nilai
0     Andi    20      88
1     Budi    21      78
2     Citra   19      92

↑
Baris (Index)
```

DataFrame memiliki:
- Index
- Column
- Values

---

### Membuat DataFrame dari Dictionary

```python
import pandas as pd

data = {
    'Nama' : ['Andi', 'Budi', 'Citra'],
    'Umur' : [20, 21, 19],
    'Nilai' : [88, 78, 92]
}

df = pd.DataFrame(data)

print(df)
```

Output:

```text
    Nama  Umur  Nilai
0  Andi    20      88
1  Budi    21      78
2 Citra    19      92
```

---

### Membuat DataFrame dari List of List

```python
data = [
    ['Andi', 20, 88],
    ['Budi', 21, 78],
    ['Citra', 19, 92]
]

df = pd.DataFrame(
    data,
    columns = ['Nama', 'Umur', 'Nilai']
)

print(df)
```

Hasilnya sama.

---

### Melihat Tipe Data

```python
print(type(df))
```

Output:

```python
<class 'pandas.core.frame.DataFrame'>
```

---

### Melihat Nama Kolom

```python
print(df.columns)
```

Output

```python
Index(['Nama', 'Umur', 'Nilai'], dtype='object')
```

---

### Melihat Index

```python
df.index
```

Output

```python
RangeIndex(start=0, stop=3, step=1)
```

---

### Mengambil Satu Kolom

Karena setiap kolom sebenarnya adalah Series, kita bisa mengambilnya seperti ini:

```python
print(df['Nama'])
```

Output

```text
0     Andi
1     Budi
2    Citra
Name: Nama, dtype: object
```

Perhatikan bahwa hasilnya adalah **Series**, bukan DataFrame.

---

### Ringkasan Materi 

- ✅ DataFrame adalah struktur data 2 dimensi.
- ✅ Terdiri dari baris (rows) dan kolom (columns).
- ✅ Bisa dibuat dari dictionary maupun list.
- ✅ Memiliki `.columns` dan `.index`.
- ✅ Mengambil satu kolom akan menghasilkan Series.

## 🌱 Chapter 3: Creating Data 🎯

### Cara 1 — Dictionary of Lists (Paling Umum)

Ini adalah cara yang paling sering digunakan.

```python
import pandas as pd

data = {
    "Nama": ["Andi", "Budi", "Citra"],
    "Umur": [20, 21, 19],
    "IPK": [3.75, 3.42, 3.91]
}

df = pd.DataFrame(data)

print(df)
```

Output:

```text
    Nama  Umur   IPK
0  Andi    20  3.75
1  Budi    21  3.42
2 Citra    19  3.91
```

---

### Cara 2 — List of Dictionaries

Setiap dictionary mewakili satu baris data.

```python
import pandas as pd

data = [
    {"Nama": "Andi", "Umur": 20},
    {"Nama": "Budi", "Umur": 21},
    {"Nama": "Citra", "Umur": 19}
]

df = pd.DataFrame(data)

print(df)
```

Hasilnya sama.

Cara ini sering digunakan ketika membaca data dari API (format JSON).

---

### Cara 3 — NumPy Array

Karena pandas dibangun diatas NumPy, kita juga bisa membuat DataFrame dari array.

```python
import numpy as np
import pandas as pd

data = np.array([
    ['Andi', 20],
    ['Budi', 21],
    ['Citra', 19]
])

df = pd.DataFrame(
    data, 
    columns = ['Nama', 'Umur']
)

print(df)
```

Output

```text
    Nama Umur
0   Andi   20
1   Budi   21
2  Citra   19
```

---

### Menentukan Index Sendiri 

Secarra default:

```text
0
1
2
```

Kita bisa menggantinya.

```python
df = pd.DataFrame(
    data, 
    index = ['A', 'B', 'C'],
    columns = ['Nama', 'Umur']
)

print(df)
```

Output:

```text
    Nama Umur
A   Andi   20
B   Budi   21
C  Citra   19
```

---

### Mengganti Nama Kolom

Misalnya:

```python
df.columns = [
    'Nama Mahasiswa',
    'Usia'
]

print(df)
```

Output

```text
  Nama Mahasiswa  Usia
A Andi            20
B Budi            21
C Citra           19
```

Ada satu cara lain yang juga sering dipakai untuk mengganti nama kolom:

```python
df = df.rename(columns=[
    'Produk' : 'Nama Produk',
    'Harga' : 'Harga (Rp)'
])
```

Perbedaannya:
- `df.columns = [...]` → mengganti semua nama kolom sekaligus.
- `rename(columns={...})` → hanya mengganti kolom tertentu.

Nanti saat bekerja dengan dataset besar, `rename()` sering menjadi pilihan yang lebih aman.

---

### Ringkasan Materi 

| Cara                 | Kapan Digunakan              |
| -------------------- | ---------------------------- |
| Dictionary of Lists  | Dataset manual (paling umum) |
| List of Dictionaries | JSON / API                   |
| NumPy Array          | Integrasi dengan NumPy       |
| `index=`             | Membuat label baris          |
| `columns=`           | Menentukan nama kolom        |

# 📊 Pandas — Data Inspection

## 🔍 Chapter 1: Melihat Data (`head()`, `tail()`, `shape`)

---

## 🎯 Learning Objectives

Setelah materi ini kamu mampu:

- ✅ Menampilkan beberapa baris pertama dataset.
- ✅ Menampilkan beberapa baris terakhir dataset.
- ✅ Mengetahui ukuran dataset.
- ✅ Menentukan jumlah baris dan kolom.

---

### Dataset Contoh 

Kita akan menggunakan dataset sederhana berikut.

```python
import pandas as pd

data = {
    "Nama": ["Andi", "Budi", "Citra", "Dani", "Eka"],
    "Umur": [20, 21, 19, 22, 20],
    "Jurusan": [
        "Informatika",
        "Fisika",
        "Matematika",
        "Kimia",
        "Biologi"
    ],
    "IPK": [3.75, 3.42, 3.91, 3.58, 3.81]
}

df = pd.DataFrame(data)
```

---

### 1️⃣ `.head()`

Digunakan untuk melihat 10 **baris paling atas**.

```python
print(df.head())
```

Output:

```text
    Nama  Umur      Jurusan   IPK
0  Andi    20  Informatika  3.75
1  Budi    21      Fisika   3.42
2 Citra    19  Matematika   3.91
3  Dani    22       Kimia   3.58
4   Eka    20     Biologi  3.81
```

Karena dataset hanya berisi 5 baris, semuanya ditampilkan.

---

#### Menampilkan n Baris Pertama

Misal: n  = 3

```python
print(df.head(3))
```

Output:

```text
    Nama  Umur      Jurusan
0  Andi    20  Informatika
1  Budi    21      Fisika
2 Citra    19  Matematika
```

`head(3)` berarti tampilkan **3 baris pertama**.

---

### 2️⃣ `.tail()`

Melihat 10 **baris paling bawah/terakhir**.

```python
print(df.tail())
```

Karena hanya ada 5 baris, hasilnya sama dengan `head()`.

---

#### Menampilkan n Baris Terakhir

Misal : n = 2

```python
print(df.tail(2))
```

Output:

```text
   Nama  Umur Jurusan   IPK
3 Dani    22   Kimia  3.58
4 Eka     20 Biologi  3.81
```

### 3️⃣ `.shape`

Melihat ukuran dataset.

```python
print(df.shape)
```

Output:

```text
(5, 4)
```

Artinya:
- 5 baris
- 4 kolom

Ingat urutannya selalu:

```text
(rows, columns)
```

---

### Mengambil Jumlah Baris

```python
print(df.shape[0])
```

Output:

```python
5
```

---

### Mengambil Jumlah Kolom

```python
print(df.shape[1])
```

Output:

```python
4
```

---

### Kapan Digunakan?

Saat menerima dataset baru, workflow yang umum adalah:

```python
df.head()
df.shape
```

Dengan dua perintah ini saja, kamu sudah mengetahui:
- seperti apa bentuk datanya,
- jumlah data,
- nama kolom yang tersedia.

---

### Ringkasan materi 

| Method     | Fungsi                       |
| ---------- | ---------------------------- |
| `head()`   | Menampilkan baris pertama    |
| `head(n)`  | Menampilkan n baris pertama  |
| `tail()`   | Menampilkan baris terakhir   |
| `tail(n)`  | Menampilkan n baris terakhir |
| `shape`    | Menampilkan `(baris, kolom)` |
| `shape[0]` | Jumlah baris                 |
| `shape[1]` | Jumlah kolom                 |

## 🔍 Chapter 3: `.describe()`

## 🎯 Learning Objectives

Setelah materi ini kamu mampu:

- ✅ Menggunakan `.describe()`
- ✅ Membaca statistik dasar dari data numerik
- ✅ Memahami arti `count`, `mean`, `std`, `min`, `max`, dan `kuartil`
- ✅ Menggunakan `.describe()` sebagai langkah awal Exploratory Data Analysis (EDA)

---

### Dataset Contoh

```python
import pandas as pd

data = {
    "Nama": ["Andi", "Budi", "Citra", "Dani", "Eka"],
    "Umur": [20, 21, 19, 22, 20],
    "IPK": [3.75, 3.42, 3.91, 3.58, 3.81]
}

df = pd.DataFrame(data)
```

---

### Apa itu `.describe()`?

Method `.describe()` menghasilkan ringkasan statistik untuk **kolom numerik**.

```python
df.descriibe()
```

Output nya kira-kira seperti ini:

```text
            Umur       IPK
count   5.000000  5.000000
mean   20.400000  3.694000
std     1.140175  0.202558
min    19.000000  3.420000
25%    20.000000  3.580000
50%    20.000000  3.750000
75%    21.000000  3.810000
max    22.000000  3.910000
```

---

### Arti Setiap Baris

#### 1️⃣ `count`

Jumlah data yang **tidak kosong** (non-null).

```text
count = 5
```

Artinya ada **5 data** yang digunakan dalam perhitungan.

---

#### 2️⃣ `mean`

Rata-rata

Misalnya:

```text
Umur

20
21
19
22
20
```

Perhitungan:

```text
(20 + 21 + 19 + 22 + 20) / 5 = 20.4
```

---

#### 3️⃣ `std`

**Standar Deviation** (Simpangan Baku).

Mengukur seberapa jauh penyebaran data dari nilai rata-ratanya.

- Nilai kecil → data cenderung berdekatan.
- Nilai besar → data lebih menyebar.

Untuk saat ini cukup pahami konsepnya. Nanti akan dipelajari lebih dalam dalam materi statistik.

---

#### 4️⃣ `min`

Nilai terkecil.

```text
19
```

---

#### 5️⃣ `25%` (Q1)

Kuartil pertama (Q1).

```text
Umur    

20
```

Artinya sekitar **25% data berada dibawah atau sama dengan nilai ini.
> 25% data $\le$ 20

---

#### 6️⃣ `50%` (Q2)

Median (Q2).

```text
Umur

20
```

Ini adalah nilai tengah data.

Bukan rata-rata.

---

#### 7️⃣ `75%` (Q3)

Kuartil Ketiga (Q3).

```text
Umur

21
```

Artinya sekitar **75% data berada di bawah atau sama dengan** nilai ini.
> 75% data $\le$ 21

---

#### 8️⃣ `max`

Nilai terbesar.

```text
22
```

---

### Hanya Kolom Numerik

Perhatikan bahwa kolom **Nama** tidak muncul.

Mengapa?

Karena:

```text
Nama → object / str
```

Sedangkan `.describe()` secara default hanya menghitung statistik untuk kolom numerik.

---

### `.describe(include = 'all')`

Jika ingin melihat semua kolom, termasuk teks:

```python
df.describe(include = 'all')
```

Output akan menampilkan informasi tambahan seperti:

- `unique`
- `top`
- `freq`

Contoh:

```text
Nama

count      5
unique     5
top      Andi
freq       1
```

Keterangan:
- **unique** → jumlah nilai unik.
- **top** → nilai yang paling sering muncul.
- **freq** → frekuensi kemunculan nilai tersebut.

---

### Kapan Menggunakan `.describe()`?

Urutan kerja umum:

```python
df.head()
df.shape
df.info()
df.describe()
```

Dengan empat langkah ini, kamu sudah memiliki gambaran awal tentang struktur dan karakteristik dataset.

---

### Ringkasan Materi

| Statistik | Arti                 |
| --------- | -------------------- |
| `count`   | Jumlah data non-null |
| `mean`    | Rata-rata            |
| `std`     | Simpangan baku       |
| `min`     | Nilai terkecil       |
| `25%`     | Kuartil 1            |
| `50%`     | Median               |
| `75%`     | Kuartil 3            |
| `max`     | Nilai terbesar       |

> **Catatan:** Kita akan fokus pada `iloc[]` terlebih dahulu karena lebih mudah dipahami.Setelah itu baru belajar `loc[]`.

### Dataset

```python
import pandas as pd

data = {
    "Nama": ["Andi", "Budi", "Citra", "Dani", "Eka"],
    "Umur": [20, 21, 19, 22, 20],
    "Jurusan": [
        "Informatika",
        "Fisika",
        "Matematika",
        "Kimia",
        "Biologi"
    ],
    "IPK": [3.75, 3.42, 3.91, 3.58, 3.81]
}

df = pd.DataFrame(data)
```

---

### Apa itu `iloc[]`?

`iloc` adalah singkatan dari:
> **Integer Location**

Artinya kita memilih data berdasarkan nomor indeks (posisi).

Misalnya:

| Index | Nama  |
| ----: | ----- |
|     0 | Andi  |
|     1 | Budi  |
|     2 | Citra |
|     3 | Dani  |
|     4 | Eka   |

---

### Mengambiil Satu Baris

Ambil baris pertama.

```python
print(df.iloc[0])
```

Output:

```text
Nama              Andi
Umur                20
Jurusan    Informatika
IPK               3.75
Name: 0, dtype: object
```

Perhatikan hasilnya.

Karena hanya mengambil satu baris, hasilnya berupa:

```python
print(type(df.iloc[0]))
```

Output

```python
<class 'pandas.core.series.Series'>
```

---

### Contoh Lain

Baris kedua:

```python
print(df.iloc[1])
```

Output:

```text
Nama        Budi
Umur          21
Jurusan   Fisika
IPK         3.42
```

---

### Mengambil Beberapa Baris

Misalnya ingin mengambil:
- Baris 0
- Baris 1
- Baris 2

Gunakan slicing.

```python
print(df.iloc[0:3])
```

Output:

```text
    Nama  Umur      Jurusan   IPK
0   Andi    20  Informatika  3.75
1   Budi    21       Fisika  3.42
2  Citra    19   Matematika  3.91
```

Sekrang cek tipenya.

```python
print(type(df.iloc[0:3]))
```

Ouput:

```python
<class 'pandas.core.frame.DataFrame'>
```

---

### Kenapa `0:3` menghasilkan 3 Baris?

Karena aturan slicing python adalah:

```python
awal : akhir
```
- Awal → ikut
- Akhir → tidak ikut

Contoh:

```python
print(df.iloc[0:3])
```

Artinya:

- ✅ 0
- ✅ 1
- ✅ 2
- ❌ 3

---

### Contoh Lain

```python
print(df.iloc[2:5]])
```

Output:

```text
        Nama  Umur     Jurusan   IPK
2  Citra    19  Matematika  3.91
3   Dani    22       Kimia  3.58
4    Eka    20     Biologi  3.81
```

---

### Mengambil Baris Terakhir

Cara pertama:

```python
print(df.iloc[4])
```

Cara yang lebih fleksibel:

```python
print(df.iloc[-1])
```

Output:

```text
Nama            Eka
Umur             20
Jurusan    Biologi
IPK           3.81
```

Kenapa menggunakan `-1`?

Dalam Python:

```text
-1 = elemen terakhir
-2 = elemen sebelum terakhir
-3 = tiga dari belakang
```

Ini sangat berguna jika jumlah baris berubah-ubah.

---

### Ringkasan

| Sintaks        | Hasil                              |
| -------------- | ---------------------------------- |
| `df.iloc[0]`   | Baris pertama (Series)             |
| `df.iloc[2]`   | Baris ketiga (Series)              |
| `df.iloc[-1]`  | Baris terakhir (Series)            |
| `df.iloc[0:3]` | Tiga baris pertama (DataFrame)     |
| `df.iloc[2:5]` | Baris ke-3 sampai ke-5 (DataFrame) |

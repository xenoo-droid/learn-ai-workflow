```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv('data_mahasiswa.csv')
df
```

```python
sns.histplot(
    data= df,
    x = 'jam_belajar',
    hue= 'jurusan',
    bins = 50,
    kde=True,
)
plt.show()
```

```python
sns.kdeplot(
    data=df,
    x='jam_tidur',
    hue='jurusan',
    fill=True,
    bw_adjust=2
)
plt.show()
```

```python
sns.displot(
    data=df,
    x="jam_belajar",
    kind="hist",
    col='jenis_kelamin',
    bins=50,
    hue='jurusan'
)
plt.show()
```

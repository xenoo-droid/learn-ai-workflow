# 📊 Pandas → Column Manipulation

## ⭐ Materi 1: Menambah Kolom

Ini dasar penting karena dalam data science kita sering membuat kolom baru dari data yang sudah ada.

---
### DataFrame

```python
import pandas as pd

df = pd.DataFrame({
    "Nama": ["Andi", "Budi", "Citra", "Doni"],
    "Umur": [20, 22, 19, 25],
    "Nilai": [85, 90, 95, 80]
})

print(df)
```

Output

```text
    Nama  Umur  Nilai
0   Andi    20     85
1   Budi    22     90
2  Citra    19     95
3   Doni    25     80
```

---

### 📚 1. Menambah Kolom

Sintaks:

```python
df['Status'] = 'Lulus'
```

Hasil:

```text
    Nama  Umur  Nilai  Status
```

Artinya:

```python
df['nama_kolom_baru'] = nilai
```

### 📚 2. Menambah Kolom Berdasarkan Kolom Lain

Ini jauh lebih sering digunakan dalam data analysis.

Contoh:

```python
df['Nilai_10'] = df['Nilai'] * 10
```

Hasil:

```text
    Nama  Umur  Nilai  Nilai_10
```

Perhatikan bahwa kita **tidak menggunakan `for` loop**.

### 📚 3. Menambah Kolom dari Beberapa Kolom

Misalnya kita ingin membuat kategori umur:

```python
df['Umur_2x'] = df['Umur'] * 2
```

Atau menghtung nilai akhir:

```python
df['Nilai_Akhir'] = df['Nilai'] + df['Umur']
```

Pandas akan melakukan perhitungan baris demi baris secara otomatis.

---

### 📚 4. Menambah Kolom dengan Kondisi

Misalnya:
> Nilai ≥ 90 → `"Lulus"`
> 
> Nilai < 90 → `"Belum"`

Kita bisa menggunakan `np.where()`:

```python
import numpy as np

df['Status'] = np.where(
    df['Nilai'] >= 90,
    'Lulus',
    'Belum'
)
```

Hasil:

```text
    Nama  Umur  Nilai  Status
0   Andi    20     85  Belum
1   Budi    22     90  Lulus
2  Citra    19     95  Lulus
3   Doni    25     80  Belum
```

Ini menghubungkan skill yang sudah kamu pelajari sebelumnya di NumPy dengan Pandas.

## ⭐ Materi 3: Rename Column

Perhatikan bahwa kita menyimpan hasilnya kembali:

```python
df = df.rename(...)
```

---

### 📚 2. Rename Beberapa Kolom

Kita bisa mengganti beberapa nama kolom sekaligus:

```python
df = df.rename(
    columns={
        'Nama': 'Nama_Siswa',

```

Hasil:

```text

0       Andi           85
1       Budi           90
```

### 📚 3. `rename()` Tidak Mengubah Isi Data

Misalnya:

```python

```

Yang berubah hanya:

```text

```

Nilainya tetap:

```text

```

Jadi:

> **Rename Column = mengubah label/nama kolom, bukan data di dalamnya**.

---

Namun cara ini harus hati-hati karena **jumlah nama baru harus sama dengan jumlah kolom**.

Untuk mengganti beberapa kolom tertentu, `.rename()` biasanya lebih aman:

```python

```

## ⭐ Materi 4: Drop Column

```python
df.drop(columns=['kolom1', 'kolom2'])

```

## ⭐ Materi 5: Insert Column

### 📚 1. Apa itu `insert()`?

`insert()` digunakan untuk menambahkan kolom baru pada posisi tertentu.

Format:

```python
df.insert(posisi, 'Nama_Kolom', value)
```

Contoh:

```python
import pandas as pd

df = pd.DataFrame({
    "Nama": ["Andi", "Budi", "Citra"],
    "Nilai": [85, 90, 95]
})

df.insert(1, "Umur", [20, 22, 19])

print(df)
```

Output:

```text
    Nama  Umur  Nilai
0   Andi    20     85
1   Budi    22     90
2  Citra    19     95
```

Perhatikan:

```python
df.insert(1, ...)
```

Angka `1` berarti kolo dimasukkan pada posisi index kolom ke-1.

---

### 📚 2. Memahami Posisi

Misalnya awalnya:

```text
Posisi 0 → Nama
Posisi 1 → Nilai
```

Kalau:

```python
df.insert(1, 'Umur', [20, 22, 19])
```

maka:

```text
Posisi 0 → Nama
Posisi 1 → Umur   ← baru
Posisi 2 → Nilai
```

Jadi `insert()` berbeda dari:

```python
df['Umur'] = [...]
```

karen assignment biasa biasanya menambahkan kolomdi bagian paling akhir, sedangkan `inset()` memungkinkan kita menentukan posisinya.

---

### 📚 3. Insert Kolom dari Perhitungan

Nilai yang dimasukkan juga bisa berasal dari kolom lain.

Contoh:

```python
df.insert(
    2,
    'Nilai_2x',
    df['Nilai'] * 2
)
```

Outpu:

```text
    Nama  Umur  Nilai_2x  Nilai
0   Andi    20       170     85
1   Budi    22       180     90
2  Citra    19       190     95
```

---

### 📚 4. `insert()` Memodifikasi DataFrame Langsung

Ini sedikit berbeda dengan beberapa method yang sebelumnya kita pelajari.

Kita cukup:

```python
df.insert(...)
```

Tidak perlu:

```python
df = df.insert(...)
```

Karena `insert()` melakukan perubahan **langsung pada DataFrame** dan tidak mengembalikan DataFrame baru.

Jadi **jangan** melakukan:

```python
df = df.insert(...)
```

---

### 🧠 Perbandingan

#### Assignment

```python
df['Kelas'] = ['A', 'B', 'A']
```

➡️ Kolom baru ditambahkan di bagian akhir.

#### `insert()`

```python
df.insert(1, 'Kelas', ['A', 'B', 'A'])
```

➡️ Kolom baru ditempatkan pada posisi tertentu.

# 📊 Pandas → Row Manipulation

## ⭐ Materi 1: Menambah Baris

Ada beberapa cara yang perlu kamu pahami, tetapi untuk workflow Pandas modern kita akan fokus pada `pd.concat()`.

Misalnya punya DataFrame:

```python
import pandas as pd

df = pd.DataFrame({
    "Nama": ["Andi", "Budi", "Citra"],
    "Umur": [20, 21, 19],
    "Nilai": [85, 90, 88]
})

print(df)
```

Output:

```text
    Nama  Umur  Nilai
0   Andi    20     85
1   Budi    21     90
2  Citra    19     88
```

Kita ingin menambahkan mahasiswa baru:

```text
Doni | 22 | 95
```

Buat DataFrame:

```python
data_baru = pd.DataFrame({
    'Nama': ['Doni'],
    'Umur': [22],
    'Nilai': [95]
})
```

Kemudian gabungkan:

```python
df = pd.concat(
    [df, data_baru],
    ignore_index=True
)
```

Output:

```text
    Nama  Umur  Nilai
0   Andi    20     85
1   Budi    21     90
2  Citra    19     88
3   Doni    22     95
```

#### 🔑 Yang perlu kamu pahami

```python
pd.concat([df, data_baru], ignore_index=True)
```

- `df`→ DataFrame lama.
- `data_baru` → baris/data yang ingin ditambahkan.
- `ignore_index=True` → membuat index baru `0, 1, 2, 3, 4`.
- Hasil `concat()` **dikembalikan sebagai DataFrame baru**, jadi kalau ingin `df` berubah, simpann kembali:

```python
df = pd.concat([df, data_baru], ignore_index=True)
```

Ini penting karena kamu sebelumnya sudah membangun kebiasaan yang benar saat `sort_values()` dan `reset_index()`: **hasil operasi disimpan kembali jika memang ingin mempertahankan perubahan**.

#### ⚠️ Catatan penting

Jangan memakai:

```python
df.append(...)
```

untuk workflow Pandas modern. Kita gunakan `pd.concat()`.

---

### 🧠 Pola yang harus kamu ingat

```text
DataFrame lama
      +
DataFrame baru
      ↓
pd.concat()
      ↓
DataFrame hasil
      ↓
simpan kembali ke df
```

---

## ⭐ Materi 3: Reset Index

### 📚 Konsep Dasar

Setelah menghapus baris, index sering menjadi tidak berurutan.

```python
df = df.drop(index=1)
```

Output:

```text
    Nama  Nilai
0   Andi     85
2  Citra     90
3   Doni     95
```

Untuk membuat index kembali `0, 1, 2`, gunakan:

```python
df = df.reset_index(drop=True)
```

Output:

```text
    Nama  Nilai
0   Andi     85
1  Citra     90
2   Doni     95
```

🔑 `drop=True`

Ini penting:

```python
df.reset_index(drop=True)
```

`drop=True` berarti **index lama dibuang**.

Kalau tidak menggunakan `drop=True`:

```python
df.reset_index()
```

index lama akan dimasukkan menjadi kolom baru bernama `index`.

## ⭐ Materi 4: Set Index

### 📚 Konsep: set_index()

`set_index()` digunakan untuk menjadikan **kolom tertentu sebagai index DataFrame**.

Misalnya:

```python
import pandas as pd

df = pd.DataFrame({
    "Nama": ["Andi", "Budi", "Citra", "Doni"],
    "Umur": [20, 21, 19, 22],
    "Nilai": [85, 90, 88, 95]
})

print(df)
```

Index awal:

```text
    Nama  Umur  Nilai
0   Andi    20     85
1   Budi    21     90
2  Citra    19     88
3   Doni    22     95
```

Kalau kita ingin `Nama` menjadi index:

```python
df = df.set_index('Nama')
```

Output:

```text
       Umur  Nilai
Nama
Andi     20     85
Budi     21     90
Citra    19     88
Doni     22     95
```

Sekarang `Nama` bukan lagi kolom biasa. Ia menjadi index.

---

### 🔑 `set_index()` vs `reset_index()`

Keduanya adalah kebalikan secara konsep:

```text
Kolom
  │
  │ set_index()
  ▼
Index
```

dan:

```text
Index
  │
  │ reset_index()
  ▼
Kolom
```

Contoh:

```python
df = df.set_index("Nama")
```

kemudian untuk mengembalikan `Nama` menjadi kolom:

```python
df = df.reset_index()
```

---

### ⚠️ drop pada `set_index()`

Secara default:

```python
df.set_index('Nama')
```

kolom `Nama` dipindahkan menjadi index.

Kalau ingin **mempertahankan kolom** `Nama`, bisa menggunakan:

```python
df.set_index('Nama', drop=False)
```

Jadi:

```text
       Nama  Umur  Nilai
Nama
Andi   Andi    20     85
Budi   Budi    21     90
```

Untuk sekarang, fokus utama kita adalah pola standar:

```python
df = df.set_index('Nama')
```

# 📊 Pandas → Missing Value

Sekarang fokusnya mendeteksi data yang kosong (`missing value`).

## ⭐ Materi 1: `isnull()`

### 1. Apa itu Missing Value?

Missing Value adalah data yang tidak memiliki nilai.

Contoh:

```python
import pandas as pd

df = pd.DataFrame({
    "Nama": ["Andi", "Budi", "Citra", "Doni"],
    "Umur": [20, None, 19, 22],
    "Nilai": [85, 90, None, 88]
})

print(df)
```

Outputnya kira-kira:

```text
    Nama  Umur  Nilai
0   Andi  20.0   85.0
1   Budi   NaN   90.0
2  Citra  19.0    NaN
3   Doni  22.0   88.0
```

`NaN` menandakan nilai yang hilang.

---

### 2. `isnull()`

Untuk mengecek apakah setiap nilai kosong:

```python
print(df.isnull())
```

Output:

```text
    Nama   Umur  Nilai
0  False  False  False
1  False   True  False
2  False  False   True
3  False  False  False
```

Artinya:
- `False` → data tersedia.
- `True` → data missing.

---

### 3. Mencari Missing Value per Kolom

Biasanya kita tidak ingin melihat seluruh tabel boolean. Kita ingin tahu *berapa banyak missing value di setiap kolom*.

Gunakan:

```python
print(df.isnull().sum())
```

Output:

```text
Nama     0
Umur     1
Nilai    1
dtype: int64
```

Cara membacanya:

```text
Nama  → 0 missing
Umur  → 1 missing
Nilai → 1 missing
```

#### 🔑 Pola penting

```python
df.isnull()
```

→ mendeteksi posisi missing value.

```python
df.isnull().sum()
```

→ menghitung jumlah missing value.

---

## ⭐ Materi : `notnull()`

### 1. Apa fungsi `notnull()`?

Kalau:

```python
df.isnull()
```

mencari data yang kosong.

maka:

```python
df.notnull()
```

mencari **data yang tidak kosong**.

Contoh:

```python
import pandas as pd

df = pd.DataFrame({
    "Nama": ["Andi", "Budi", "Citra", "Doni", "Eko"],
    "Umur": [20, None, 19, 22, None],
    "Nilai": [85, 90, None, 88, 95],
    "Kota": ["Jakarta", "Bandung", None, "Surabaya", "Malang"]
})

print(df.notnull())
```

Outputnya:

```text
    Nama   Umur  Nilai   Kota
0   True   True   True   True
1   True  False   True   True
2   True   True  False  False
3   True   True   True   True
4   True  False   True   True
```

#### 🔑 Ingat pasangan ini

```text
isnull()   → True kalau data KOSONG
notnull()  → True kalau data TERISI
```

---

### 2. Menghitung Data yang Tidak Kosong

Sama seperti `isnull()`, kita bisa menggunakan `.sum()`:

```python
print(df.notnull().sum())
```

Output:

```text
Nama     5
Umur     3
Nilai    4
Kota     4
```

Artinya:
- `Nama` → 5 data terisi
- `Umur` → 3 data terisi
- `Nilai` → 4 data terisi
- `Kota` → 4 data terisi

Perhatikan hubungan dengan `isnull()`:

```text
Umur
Total data     = 5
Missing        = 2
Tidak missing  = 3
```

Jadi:

```text
5 - 2 = 3
```

## ⭐ Materi 3: `fillna()`

Mengisi Missing Value.

### 1. Apa fungsi `fillna()`?

`fillna()` digunakan untuk **menggantikan nilai kosong(`NaN`/`None`) dengan nilai yang kita tentukan**.

Gunakan DataFrame dibawah ini:

```python
import pandas as pd

df = pd.DataFrame({
    "Nama": ["Andi", "Budi", "Citra", "Doni", "Eko"],
    "Umur": [20, None, 19, 22, None],
    "Nilai": [85, 90, None, 88, 95],
    "Kota": ["Jakarta", "Bandung", None, "Surabaya", "Malang"]
})
```

Misalnya semua missing value ingin diganti dengan `0`:

```python
df.fillna(0)
```

Output:

```text
    Nama  Umur  Nilai      Kota
0   Andi  20.0   85.0   Jakarta
1   Budi   0.0   90.0   Bandung
2  Citra  19.0    0.0         0
3   Doni  22.0   88.0  Surabaya
4    Eko   0.0   95.0    Malang
```

#### 🔑 Konsep penting

```python
df.fillna(0)
```

menghasilkan DataFrame baru.

Jadi `df` asli belum berubah.

Kalau ingin menyimpan hasilnya:

```python
df = df.fillna(0)
```

---

### 2. Mengisi Missing Value dengan Nilai Berbeda

Kita juga bisa menentukan nilai berdasarkan kolom.

Misalnya:

```python
df.fillna({
    'Umur' : 0,
    'Nilai': 0,
    'Kota' : 'Unknown'
})
```

Artinya:
- `Umur` kosong → `0`
- `Nilai` kosong → `0`
- `Kota` kosong → `"Unknown"`

Ini lebih masuk akal daripada semua kolom diberi `0`.

---

### 3. `fillna()` Tidak Sama dengan `dropna()`

Ini penting untuk nanti.

```text
fillna() → mempertahankan baris, tetapi mengisi data kosong
dropna() → menghapus data yang memiliki nilai kosong
```

Contoh:

```python
df.fillna(0)
```

→ jumlah baris tetap 5.

## ⭐ Materi 4: `dropna()`

Menghapus data yang memiliki missing value.

### DataFrame

```python
import pandas as pd

df = pd.DataFrame({
    "Nama": ["Andi", "Budi", "Citra", "Doni", "Eko"],
    "Umur": [20, None, 19, 22, None],
    "Nilai": [85, 90, None, 88, 95],
    "Kota": ["Jakarta", "Bandung", None, "Surabaya", "Malang"]
})
```

---

### 1. Apa fungsi `dropna()`?

`dropna()` digunakan untuk **menghapus baris yang memiliki missing value**.

```python
df.dropna()
```

Karena terdapat missing value pada:

- Budi → `Umur`
- Citra → `Nilai` dan `Kota`
- Eko → `Umur`

maka hasilnya hanya:

```text
    Nama  Umur  Nilai      Kota
0   Andi  20.0   85.0   Jakarta
3   Doni  22.0   88.0  Surabaya
```

Jadi:

```text
5 baris
↓
3 baris dihapus
↓
2 baris tersisa
```

#### 🔑 Pola dasarnya

```python
df.dropna()
```

→ hapus baris yang memiliki missing value.

Dan seperti `fillna()`, hasilnya merupakan DataFrame baru.

```python
df_clean = df.dropna()
```

---

### 2. `dropna()` vs `fillna()`

Ini harus benar-benar kamu bedakan:

```text
fillna()
   ↓
Missing Value
   ↓
DIISI
```

sedangkan:

```text
dropna()
   ↓
Missing Value
   ↓
DIHAPUS
```

Contoh:

```python
df.fillna(0)
```

→ jumlah baris tetap.

```python
df.dropna()
```

→ baris yang memiliki missing value dihapus.

---

### 3. `dropna(subset=...)`

Kita juga bisa menentukan **kolom mana yang dijadikan patokan**.

Misalnya hanya ingin menghapus baris yang `Nilai`-nya kosong:

```python
df.dropna(subset=['Nilai'])
```

Output:

```text
    Nama  Umur  Nilai      Kota
0   Andi  20.0   85.0   Jakarta
1   Budi   NaN   90.0   Bandung
3   Doni  22.0   88.0  Surabaya
4    Eko   NaN   95.0    Malang
```

Citra saja yang yang dihapus karena `Nilai` Citra kosong.

#### Polanya:

```python
df.dropna(subset['NamaKolom'])
```

→ hapus baris jika kolom tersebut memiliki missing value.

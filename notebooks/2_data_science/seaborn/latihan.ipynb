```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.DataFrame({
    "Produk": ["A", "B", "C", "D", "E"],
    "Penjualan": [120, 180, 150, 220, 190],
    "Pelanggan": [30, 45, 38, 55, 48]
})

plt.title('Penjualan Per Bulan')
sns.barplot(data=df, x='Produk', y='Penjualan')

plt.show()
```

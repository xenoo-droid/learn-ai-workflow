```python
# Dataset Filtering

df = pd.DataFrame({
    "Nama": ["Andi", "Budi", "Citra", "Dina", "Eko"],
    "Umur": [20, 22, 19, 21, 23],
    "Nilai": [85, 70, 90, 75, 95]
})
```

```python
df = pd.DataFrame({
    "Produk": ["Laptop", "Mouse", "Laptop", "Keyboard", "Mouse"],
    "Kategori": ["Elektronik", "Aksesoris", "Elektronik", "Aksesoris", "Aksesoris"],
    "Harga": [8000, 200, 7500, 500, 250]
})

grouped = df.groupby('Kategori')
grouped['Harga'].agg(
    total= 'sum',
    rata_rata= 'mean',
    tertinggi= 'max',
    terendah= 'min'
)
```

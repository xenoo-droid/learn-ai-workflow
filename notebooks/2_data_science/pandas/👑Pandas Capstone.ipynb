````python
import pandas as pd
import numpy as np
````

````python
# Dataset

data = {
    "Tanggal": [
        "2026-01-03", "2026-01-05", "2026-01-08",
        "2026-01-12", "2026-01-15", "2026-01-18",
        "2026-01-22", "2026-01-25", "2026-01-28",
        "2026-01-30"
    ],
    "Produk": [
        "Laptop", "Mouse", "Keyboard",
        "Laptop", "Monitor", "Mouse",
        "Keyboard", "Laptop", "Monitor", "Mouse"
    ],
    "Kategori": [
        "Elektronik", "Aksesoris", "Aksesoris",
        "Elektronik", "Elektronik", "Aksesoris",
        "Aksesoris", "Elektronik", "Elektronik", "Aksesoris"
    ],
    "Kota": [
        "Surabaya", "Malang", "Surabaya",
        "Jakarta", "Malang", "Jakarta",
        "Surabaya", "Jakarta", "Surabaya", "Malang"
    ],
    "Jumlah": [2, 10, 5, 1, 3, 8, 7, 2, 4, 12],
    "Harga": [8000000, 150000, 300000, 8000000, 2500000,
              150000, 300000, 8000000, 2500000, 150000]
}

df = pd.DataFrame(data)
````

````python
# 🗺️ QUEST 1 — DATA INSPECTION
print('''==== DATA INSPECTION ====''')

print(f'''info(): 
{df.info()}
\n
head(): 
{df.head()}
\n
tail(): 
{df.tail()}
\n
describe(): 
{df.describe()}
\n
\n
shape(): {df.shape}
\n
missing value: {df.isna().sum().sum()}''')
````

````python
# 🧹 QUEST 2 — DATA PREPARATION
print('==== DATA PREPARATION ====')
df['Tanggal'] = pd.to_datetime(df['Tanggal'])
df['Total'] = df['Jumlah'] * df['Harga']
df
````

# 🎯 QUEST 3 — DATA ANALYSIS

Jawab pertanyaan berikut menggunakan Pandas:

---

## A. Produk
- Produk apa yang menghasilkan total penjualan terbesar?
- Produk mana yang memiliki jumlah unit terjual terbanyak?

---

````python
produk_total_max = df.groupby('Produk')['Total'].sum().idxmax()
print(produk_total_max)

produk_jumlah_max = df.groupby('Produk')['Jumlah'].sum().idxmax()
print(produk_jumlah_max)
````

# Jawaban A:

- Produk yang menghasilkan total penjualan terbesar:
  > Laptop
  
- Produk yang memiliki jumlah unit terjual terbanyak:
  > Mouse

---

# B. Kota
- Kota mana yang menghasilkan **revenue terbesar**?
- Berapa total revenue setiap kota?

---

````python
kota_revenue_max = df.groupby('Kota')['Total'].sum().idxmax()
print(kota_revenue_max)
kota_total_revenue = df.groupby('Kota')['Total'].sum()
print(kota_total_revenue)
````

# Jawaban B

- Kota yang menghasilkan revenue terbesar:
  > Surabaya
  
- Total Revenue Tiap Kota:
```text
Jakarta   :  25200000
Malang    :  10800000
Surabaya  :  29600000
```

---

# C. Kategori
- Kategori mana yang menghasilkan revenue terbesar?
- Berapa rata-rata `Harga` setiap kategori?

---

````python
kategori_revenue_max = df.groupby('Kategori')['Total'].sum().idxmax()
print(kategori_revenue_max)
kategori_harga_mean = df.groupby('Kategori')['Harga'].mean()
print(kategori_harga_mean)
````

# Jawaban C

- Kategori yang menghasilkan revenue terbesar:
  > Elektronik
  
- Rata-Rata Harga setiap Kategori:

```text
Aksesoris   : Rp   210.000
Elektronik  : Rp 5.800.000
```

# D. Statistik Revenue

````python
print(f'''
Rata-Rata Revenue      : {int(df['Total'].mean())}
Total Revenue          : {df['Total'].sum()}
Revenue Terendah       : {df['Total'].min()}
Revenue Tertinggi      : {df['Total'].max()}
Kuartil Tengah Revenue : {int(df['Total'].median())}
''')
````

---

````python
# 📊 QUEST 4 — GROUPBY
grouped_1 = df.groupby('Produk', as_index=False)['Jumlah'].sum()
grouped_2 = df.groupby('Produk', as_index=False)['Total'].sum()
grouped_3 = df.groupby('Produk', as_index=False)['Harga'].mean()

df_group_1 = pd.merge(
    grouped_1,
    grouped_2,
    on='Produk'
)
df_group_2 = pd.merge(
    df_group_1,
    grouped_3,
    on='Produk'
)

#rename kolom
df_group_2.columns = ['Produk', 'Total Unit', 'Total Revenue', 'Rata-rata Harga']

#Sorting
sort = df_group_2.sort_values(
    by='Total Revenue',
    ascending=False
)

sort
````

````python
# 🔗 QUEST 5 — MERGE / JOIN

# Buat DataFrame Kedua
target = pd.DataFrame({
    "Kota": ["Surabaya", "Malang", "Jakarta"],
    "Target": [10000000, 8000000, 15000000]
})

analisis_kota = df.groupby(
    'Kota', as_index=False
)['Total'].sum()

quest_5 = pd.merge(
    analisis_kota,
    target,
    on='Kota'
)

# rename kolom
quest_5.columns = ['Kota', 'Revenue', 'Target']

# add status
quest_5['Status'] = np.where(
    quest_5['Revenue'] >= quest_5['Target'],
     'Tercapai',
     'Belum Tercapai'
) 
quest_5
````

# 💾 QUEST 6 — IMPORT / EXPORT

dataframe utama:

- quest_5 = dataframe hasil analisis target revenue per kota.

````python
# 💾 QUEST 6 — IMPORT / EXPORT

df.to_csv('capstone_dataframe_mentah.csv', index=False)
sort.to_csv('capstone_dataframe_analyze_1.csv', index=False)
quest_5.to_csv('capstone_dataframe_analyze_2.csv', index=False)

df_mentah = pd.read_csv('capstone_dataframe_mentah.csv')
df_analyze_1 = pd.read_csv('capstone_dataframe_analyze_1.csv')
df_analyze_2 = pd.read_csv('capstone_dataframe_analyze_2.csv')

print(df_mentah)
print('\n'*2)
print(df_analyze_1)
print('\n'*2)
print(df_analyze_2)
````

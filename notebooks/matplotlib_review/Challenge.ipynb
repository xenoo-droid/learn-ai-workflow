```python
import matplotlib.pyplot as plt

minggu = [
    "W1", "W2", "W3", "W4",
    "W5", "W6", "W7", "W8"
]

tiket_masuk = [120, 135, 128, 150, 165, 172, 160, 185]
tiket_selesai = [115, 130, 125, 142, 155, 168, 158, 176]

waktu_penyelesaian = [8.2, 7.8, 8.5, 7.2, 6.9, 6.5, 7.1, 6.2]
variasi_waktu = [1.2, 1.0, 1.4, 0.9, 1.1, 0.8, 1.0, 0.7]

kepuasan_pelanggan = [82, 84, 81, 85, 87, 89, 86, 91]

fig, ax = plt.subplots(figsize=(16, 8), nrows = 2, ncols = 2)

# Ax 1
ax[0, 0].plot(
    minggu, 
    tiket_masuk,
    label = 'Tiket Masuk',
    color = 'blue'
)
ax[0, 0].plot(
    minggu, 
    tiket_selesai,
    label = 'Tiket Selesai',
    color = 'red'
)
ax[0, 0].set_title('Tiket Masuk vs Tiket Selesai')
ax[0, 0].set_xlabel('Minggu')
ax[0, 0].set_ylabel('Jumlah Tiket')
ax[0, 0].legend()
ax[0, 0].grid()

# Ax 2
ax[0, 1].plot(
    minggu,
    waktu_penyelesaian,
    color='lightgreen',
)

ax[0, 1].set_title('Waktu Penyelesaian Selama 8 Minggu')
ax[0, 1].set_xlabel('Minggu')
ax[0, 1].set_ylabel('Waktu Penyelesaian')
ax[0, 1].grid()

# Ax 3
ax[1, 0].bar(
    minggu,
    kepuasan_pelanggan,
    color = 'grey'
)
ax[1, 0].set_title('Perubahan Kepuasan Pelanggan Selama 8 Minggu')
ax[1, 0].set_xlabel('Minggu')
ax[1, 0].set_ylabel('Kepuasan Pelanggan')
ax[1, 0].grid()

# Ax 4
ax[1, 1].errorbar(
    minggu, 
    waktu_penyelesaian,
    yerr=variasi_waktu,
    fmt='o',
    ecolor='green',
    capsize = 5
)
ax[1, 1].set_title('Estimasi Error Value Waktu Penyelesaian')
ax[1, 1].set_xlabel('Minggu')
ax[1, 1].set_ylabel('Waktu Penyelesaian')
ax[1, 1].grid()

plt.tight_layout()
plt.savefig('Analisis Challenge.png',dpi=300)
plt.show()
```

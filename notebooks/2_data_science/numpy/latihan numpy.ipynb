```python
import numpy as np
```

```python
arr = np.arange(1, 21)
shape = arr.shape
reshape = arr.reshape(4, 5)
shape1 = reshape.shape

print("="*100)
print(f"Array 1 Dimensi : {arr}\n")

print(f'''Matriks 2 Dimensi : 
{arr.reshape(4, 5)}\n''')

print(f"Shape sebelum Reshape : {shape}\n")

print(f"Shape sesudah Reshape : {shape1}\n")

print(f'''Ndim : {reshape.ndim}\n
Size : {reshape.size}\n
Dtype : {reshape.dtype}\n
Itemsize : {reshape.itemsize}\n
Nbytes : {reshape.nbytes}\n
Hasil Transpose:
{reshape.T}''')
print("="*100)
```

## Kesimpulan singkat (2-3 kalimat)

- Apa itu fungsi `reshape()`
> fungsi reshape() merupakan salah satu fungsi pada library numpy yang digunakan untuk mengubah ukuran dan dimensi dalam artian mengubah bentuk nya tanpa mengubah elemen pada array.

- Hubungan antara `shape`, `size`, dan `ndim`
> hubungan antaraa ketiga atribut array tersebut lebih seperti untuk menujukkan properti yang dimiliki oleh sebuah Matriks, shape untuk bentuk nya (baris dan kolom), lalu size merupakan jumlah elemen nya, dan ndim merupakan jumlah dimensi yang dimiliki oleh array/matriks tersebut.

- Mengapa jumlah elemen tidak berubah setelah reshape()
> karena fungsi reshape hanya mengubah bentuk dari array yaitu baris dan kolom sekaligus dimensi nya.

```python
penjualan = np.array([
    [101, 250000],
    [102, 175000],
    [103, 320000],
    [104, 90000],
    [105, 450000],
    [106, 150000]
])

# Keterangan
# Kolom 0 Id produk, kolom 1 total penjualan(Rp)

# 1. Tampilkan semua produk dengan penjualan lebih dari Rp200.000.
print(penjualan[penjualan[:,1] >200000])
print("\n")

# 2. Ambil hanya ID Produk yang penjualannya lebih dari Rp200.000.
print(penjualan[penjualan[:,1] > 200000, 0])
print("\n")

# 3. Ubah seluruh penjualan yang kurang dari Rp150.000 menjadi RP150.000
ubah = penjualan[penjualan[:,1] < 150000, 1] = 150000 

# 4. Cetak Seluruh isi Array `penjualan` setelah perubahan tersebut.
print(penjualan)
```

```python
sales = np.array([
    [12, 15, 18, 20],
    [8, 10, 9, 11],
    [20, 18, 22, 24],
    [14, 16, 15, 17],
    [10, 12, 13, 14]
])

# axis 0 -> minggu
# axis 1 -> produk

#Total penjualan setiap produk.
print(np.sum(sales, axis = 1))

#Total penjualan setiap minggu.
print(np.sum(sales, axis = 0))

#Rata-rata penjualan tiap produk
print(np.mean(sales, axis = 1))

#Rata-rata penjualan tiap minggu
print(np.mean(sales, axis = 0))

#Total penjualan seluruh toko.
print(np.sum(sales))

#Rata-rata penjualan seluruh toko
print(np.mean(sales))

#Penjualan tertinggi dari seluruh data
print(np.max(sales))

#Penjualan terendah dari seluruh data
print(np.min(sales))
```

```python
produk = np.array([
    "Keyboard",
    "Mouse",
    "Monitor",
    "Speaker",
    "Laptop"
])

harga = np.array([
    350000,
    120000,
    2500000,
    500000,
    8500000
])

idx = np.argsort(harga)

for i, urutan in enumerate(idx, start=1):
    nama_barang = produk[urutan]
    harga_barang = harga[urutan]
    print(f"{i}. {nama_barang} : Rp {harga_barang}")
```

```python
nilai = np.array([
    45, 80, 90, 60, 55,
    100, 75, 30, 88, 95
])

#semua nilai lulus
lulus = np.extract(nilai >= 75, nilai)

#semua nilai ga lulus
galulus = np.extract(nilai < 75, nilai)

#jumlah siswa yang lulus
jumlah_lulus = lulus.size

#jumlah siswa ga lulus
jumlah_galulus = galulus.size

#persentase kelulusan
persen = (jumlah_lulus/nilai.size)*100

print(f'''Semua nilai lulus : {lulus} \n
Semua nilai ga lulus        : {galulus} \n
Jumlah siswa yang lulus     : {jumlah_lulus} \n
Jumlah siswa yang ga lulus  : {jumlah_galulus} \n
Persentase kelulusan        : {persen}
''')
```
